## Current notation and verification conventions

The screened interaction includes the coupling: $W_\lambda=\lambda v/(1-\lambda v\Pi_0)$, with $v=q^{-2}$ and $\Pi_0=\chi_0=-q_{\rm TF}^2\mathcal L$ in the declared dimensionless units. Its coupling average is $\overline W=\int_0^1W_\lambda\,d\lambda$, so $2\overline W=vR(c)$ and $e_{\rm kite}=2\mathcal K[\overline W]$. The bare limit is $\overline W=v/2$.

Imaginary frequency is $i\nu$, the auxiliary width ratio is $s_\Lambda=\Lambda/q$, and the local profile is $p_{\rm loc}(s_\Lambda)$. The shell coordinate in the manuscript is $\eta_{\rm sh}$, distinct from frequency. All response corrections use the same units as $\Pi_0$: $\Pi_{\rm vtx}^{(1)}=-q_{\rm TF}^4X/(32\pi^2)$ and the self-energy response has the corresponding factor $C=\pi^2q_{\rm TF}^2$ multiplying its loop integral. The numerical kernel $X$ and $A_0$ are unchanged.

Equation labels remain stable for source linkage; displayed equation numbers are refreshed from the compiled manuscript by the coverage audit. The checks below retain their existing check identifiers.

## Setup and verification helpers

`exact` simplifies a residual to zero under the stated assumptions. `num` compares values with the tolerance printed for that check. `check` records a specified test.

```python
import sympy as sp
import mpmath as mp
mp.mp.dps = 30
pi, R, oo = sp.pi, sp.Rational, sp.oo

import time, platform, numpy as np, scipy
START_TIME = time.perf_counter()
RESULTS = []
SKIPPED = []
RUN_SLOW = True  # included in the distributed executed result; False skips Check 21
print("Python", platform.python_version(), "| NumPy", np.__version__, "| SciPy", scipy.__version__)
def check(label, ok, detail=""):
    """Record and print one verification item."""
    ok = bool(ok)
    RESULTS.append((label, ok))
    print(("PASS  " if ok else "FAIL  ") + label + (("   [" + detail + "]") if detail else ""))

def exact(label, expr):
    """EXACT: SymPy must reduce expr to 0."""
    d = sp.simplify(expr)
    check(label, d == 0, "simplify -> %s" % d)

def num(label, a, b, tol=1e-12):
    """NUM: |a-b| <= tol * max(1,|b|), a,b are mpmath numbers."""
    a, b = mp.mpf(a), mp.mpf(b)
    err = abs(a - b) / max(1, abs(b))
    check(label, err <= tol, "scaled abs. diff = %s" % mp.nstr(err, 3))

print("SymPy", sp.__version__, "| mpmath", mp.__version__, "| working precision", mp.mp.dps, "digits")
```

## Check 1 — Eq. (A13): the normalization $A_0$

**Where we are in the argument.** The starting point is the functional $\mathcal K[f]$ of Eq. (A15) in reduced units.  Its prefactor $A_0$
collects the Rydberg conversion, the spin sum, the volume per particle and the $\tfrac14$ of the diagram; every closed-form coefficient of the paper
is proportional to it, so it is checked first.

**Claim.** $A_0=2_{\rm Ha\to Ry}\,2_{\rm spin}\,(4\pi)^2(2\pi)^{-9}(3\pi^2)\cdot\tfrac14=\dfrac{3}{32\pi^5}=3.0635\times10^{-4}$.

**Manuscript route.** The factors are listed after Eq. (A13).

**How this notebook verifies it.** The first `exact` multiplies the listed factors and subtracts $3/(32\pi^5)$; the `num` line compares with the
printed five-digit value (absolute tolerance $5\times10^{-9}$ accordingly). No departure from the manuscript; this cell defines `A0` for later use.

**Hand-first.** *Which* factors appear (the $\tfrac14$ of Eq. (A7), the spin sum, $\mathcal V/N=3\pi^2/k_F^3$) is a bookkeeping statement of Sec. 2.

```python
A0 = R(3)/(32*pi**5)
exact("Eq.(A13)  2*2*(4pi)^2 (2pi)^-9 (3pi^2)/4 = 3/(32 pi^5)",
      2*2*(4*pi)**2*(2*pi)**(-9)*(3*pi**2)*R(1,4) - A0)
num("Eq.(A13)  A0 = 3.0635e-4  (tolerance = half a unit in the last printed digit)", mp.mpf(3)/(32*mp.pi**5), mp.mpf('3.0635e-4'), tol=5e-9)
```

## Check 2 — Eqs. (D2), (F1), (F24): the screening kernel $\Xi(z)$ and its Mellin transform

**Where we are in the argument.** For a single Lorentzian kernel $b_\Lambda$ the frequency integral factorizes and leaves the *screening kernel*
$\Xi(\Lambda y)$ (Eq. (D2)); all $\Lambda$ dependence of the Lorentzian-kernel energy is carried by this one function.  Its Mellin transform $\widetilde\Xi(s)$ is what
produces the large-$\Lambda$ asymptotics in Sec. 4.4: the poles of $K$ at $s=0,2,4,\dots$ and $s=-1,-3,\dots$ and the values $\widetilde\Xi(1)$, $\widetilde\Xi'(1)$ appear in the
residues of Check 5.

**Claim.**
1. $\Xi(z)=z\!\int_0^\infty\!\frac{e^{-zt}}{1+t^2}dt$ (Eq. (D2)) $=z\big[\operatorname{Ci}(z)\sin z+(\tfrac\pi2-\operatorname{Si}(z))\cos z\big]$ (Eq. (F1))
   $=\int_0^\infty e^{-w}\frac{z^2}{z^2+w^2}dw$ (Eq. (F26)).
2. $\widetilde\Xi(s)=\int_0^\infty z^{s-1}\Xi(z)\,dz=\Gamma(s+1)\int_0^\infty\frac{t^{-s-1}}{1+t^2}dt=-\frac\pi2\frac{\Gamma(s+1)}{\sin(\pi s/2)}$, $-1<\Re s<0$ (Eq. (F24)).
3. Continuation: $\widetilde\Xi(1)=-\pi/2$, $\widetilde\Xi'(1)=-\pi(1-\gamma)/2$ (used in Eq. (F6)).

**Manuscript route.** For $\widetilde\Xi(s)$: integrate over $z$ first (Laplace transform of $z^s$), then the Beta integral
$\int_0^\infty t^{-s-1}(1+t^2)^{-1}dt=\tfrac12\Gamma(-\tfrac s2)\Gamma(1+\tfrac s2)$ and the reflection formula.

**How this notebook verifies it.**
1. The three representations of $\Xi$ are coded literally (`Xi_lap`, `Xi_ci`, `Xi_w`; `mp.ci`/`mp.si` are the cosine/sine integrals) and
   compared pairwise at $z=0.3,2,9$ to 12+ digits. In `Xi_w` the breakpoint at $w=z$ marks where the Lorentzian factor turns over.
2. `K_beta` is the manuscript's intermediate form $\Gamma(s+1)\cdot\tfrac12\Gamma(-\tfrac s2)\Gamma(1+\tfrac s2)$; `sp.gammasimp(K_beta/K_closed) - 1`
   applies the reflection formula and must give 0. (The *ratio* is used rather than the difference because `gammasimp` works on products of Gamma functions.)
3. **Departure from the manuscript — the Mellin transform is computed directly.** `K_num` integrates $z^{s-1}\Xi(z)$ over $(0,\infty)$ at $s=-\tfrac12$,
   inside the strip, using the Ci/Si form; breakpoints at $1,10,100,1000$ follow the slow $z^{-3/2}$ decay. Agreement with the closed form to $10^{-10}$
   (tolerance set by the slow tail, not by the formula). This tests the $z$-first interchange and the Beta step together.
4. $\widetilde\Xi(1)$ and $\widetilde\Xi'(1)$ are obtained by substituting into and differentiating the *continued* closed form (`exact`).

**Reading the output.** Six `num` lines for the representations, one `exact` (reflection), one `num` (Mellin at $s=-\tfrac12$), two `exact` (values at $s=1$).

**Hand-first.** The strip of convergence $-1<\Re s<0$ and the pole/double-pole pattern of the continuation (Sec. 4.4).

```python
# --- 1. three representations of Xi(z) agree (NUM)
Xi_lap = lambda z: z*mp.quad(lambda t: mp.exp(-z*t)/(1+t*t), [0, 1, mp.inf])
Xi_ci  = lambda z: z*(mp.ci(z)*mp.sin(z) + (mp.pi/2 - mp.si(z))*mp.cos(z))
Xi_w   = lambda z: mp.quad(lambda w: mp.exp(-w)*z*z/(z*z+w*w), [0, z, mp.inf])
for z in [mp.mpf('0.3'), mp.mpf(2), mp.mpf(9)]:
    num("Eq.(D2)/(F1)  Laplace form = Ci/Si form, z=%s" % z, Xi_lap(z), Xi_ci(z))
    num("Eq.(F26)      w-form = Ci/Si form,      z=%s" % z, Xi_w(z),  Xi_ci(z))

# --- 2. XiT(s): Beta-integral form equals the closed form (EXACT via reflection formula)
s = sp.symbols('s')
XiT_beta   = sp.gamma(s+1) * R(1,2)*sp.gamma(-s/2)*sp.gamma(1+s/2)
XiT_closed = -pi/2*sp.gamma(s+1)/sp.sin(pi*s/2)
exact("Eq.(F24)  Gamma(s+1)*(1/2)Gamma(-s/2)Gamma(1+s/2) = -(pi/2)Gamma(s+1)/sin(pi s/2)",
      sp.gammasimp(XiT_beta/XiT_closed) - 1)
# numerical Mellin transform inside the strip, s = -1/2
sv = -mp.mpf(1)/2
XiT_num = mp.quad(lambda z: z**(sv-1)*Xi_ci(z), [0, 1, 10, 100, 1000, mp.inf])
num("Eq.(F1)   int z^(s-1) Xi dz at s=-1/2 = closed form", XiT_num,
    -mp.pi/2*mp.gamma(sv+1)/mp.sin(mp.pi*sv/2), tol=1e-10)

# --- 3. values at s=1 from the continuation (EXACT)
exact("Eq.(F1)   XiT(1) = -pi/2  (value used in Eq.(F6))", XiT_closed.subs(s, 1) + pi/2)
exact("Eq.(F1)   XiT'(1) = -pi(1-gamma)/2  (value used in Eq.(F6))", sp.diff(XiT_closed, s).subs(s, 1) + pi*(1-sp.EulerGamma)/2)
```

## Check 3 — Eqs. (D5), (D6): the static Lorentzian kernel constant $D_0$ and the Onsager energy

**Where we are in the argument.** The other factor of the one-dimensional formula is the geometric profile $\Phi(y)$.  For the static Lorentzian kernel
($\Xi\equiv1$) only $D_0=\int\Phi\,dy/y$ is needed, and it must reproduce the Onsager–Mittag–Stephen energy: $\tfrac6{\pi^3}D_0=e_{2x}$.  This is the
normalization check of the whole Lorentzian reduction (Eqs. (D2)–(D6)); if it failed, nothing downstream would be trustworthy.

**Claim.** With $\int_0^\infty j_1(ay)j_1(by)\,dy=\pi b/(6a^2)$ ($a\ge b$) the $y$-integral of $\Phi$ collapses to $D_0=\frac\pi3\int_1^\infty L(u)h_0(u)\,du$, with
$h_0(u)=-\log u\big[\frac2{u+1}-\log(1+\tfrac1u)\big]+\operatorname{Li}_2(-\tfrac1u)+2\log(1+\tfrac1u)$ (Eq. (D18)), and
$$D_0=\frac{\pi^3}{18}\log2-\frac\pi4\zeta(3)=0.2499018971\ldots\ \text{(Eq. (D5))},\qquad \mathcal K[b_\infty]=\frac{6}{\pi^3}D_0=\frac13\log2-\frac{3}{2\pi^2}\zeta(3)=e_{2x}\ \text{(Eq. (D6))}.$$

**Manuscript route.** Bessel identity → $u,v$ integral → $x=1/u$ and integration of the logarithms (the supplied derivation note gives the longer analytic route).

**How this notebook verifies it.**
1. `j1` is the spherical Bessel function $j_1(x)=\sin x/x^2-\cos x/x$ written out. The Bessel identity is checked with `mp.quadosc` (oscillatory tail,
   `omega=a+b` is the sum frequency) at $a=2$, $b=0.7$.
2. `h0` is the closed form; `h0_def` is its *definition* $\int_0^1\log\frac{1}{uv}\frac{u-v}{(u+v)^2}dv$ integrated numerically. They agree at $u=1.5,3,20$ to 30 digits.
3. **Departure from the manuscript — $D_0$ is obtained by quadrature, not by the polylogarithm chain.** `D0_num` integrates $\frac\pi3L(u)h_0(u)$ over $u\in(1,\infty)$
   (breakpoints $2,10,100$; $L(u)$ has a log singularity at $u=1$ which tanh–sinh handles) and agrees with the closed form to 30 digits. This confirms the entire
   analytic evaluation of $D_0$ from the $h_0$ representation without repeating it.
4. `exact`: $6D_0/\pi^3=e_{2x}$ and $64\pi^2A_0=6/\pi^3$ (the prefactor of Eq. (D6)).

**Reading the output.** The `(pi/3) int L h0 du = ...` line is the substantive one (rel. diff $\sim10^{-32}$). Since Check 21 (optional) evaluates $\Phi(y)$ itself,
it tests two local profile values; it does not numerically integrate the full profile over $y$. The global static normalization is independently sampled in Check 29.1.

**Hand-first.** The Frullani step giving $\log[1/(2|R_k|)]$ and the prolate-spheroidal change of variables in Sec. 4.2 (companion entries 13–14).

```python
j1 = lambda x: mp.sin(x)/x**2 - mp.cos(x)/x
a, b = mp.mpf(2), mp.mpf('0.7')
num("Eq.(D5)  int j1(ay) j1(by) dy = pi b/(6 a^2)",
    mp.quadosc(lambda y: j1(a*y)*j1(b*y), [0, mp.inf], omega=a+b), mp.pi*b/(6*a**2), tol=1e-10)

L  = lambda u: mp.log((u+1)/(u-1))
h0 = lambda u: -mp.log(u)*(2/(u+1) - mp.log(1+1/u)) + mp.polylog(2, -1/u) + 2*mp.log(1+1/u)
h0_def = lambda u: mp.quad(lambda v: mp.log(1/(u*v))*(u-v)/(u+v)**2, [0, 1])
for u in [mp.mpf('1.5'), mp.mpf(3), mp.mpf(20)]:
    num("Eq.(D5)  h0(u) closed form = its v-integral, u=%s" % u, h0(u), h0_def(u))

D0_num = mp.pi/3*mp.quad(lambda u: L(u)*h0(u), [1, 2, 10, 100, mp.inf])
D0_cl  = mp.pi**3/18*mp.log(2) - mp.pi/4*mp.zeta(3)
num("Eq.(D5)  (pi/3) int L h0 du = pi^3/18 log2 - pi/4 zeta(3)", D0_num, D0_cl, tol=1e-18)
num("Eq.(D5)  D0 = 0.2499018971...", D0_cl, mp.mpf('0.2499018971'), tol=1e-10)

# Eq. (D19): the two pieces of the x = 1/u integral, to 30 digits
with mp.workdps(40):
    Lx = lambda x: mp.log((1+x)/(1-x))
    I_log = mp.quad(lambda x: Lx(x)*mp.log(x)*(2*x/(1+x) - mp.log(1+x))/x**2, [0, mp.mpf(1)/2, 1])
    I_li  = mp.quad(lambda x: Lx(x)*(mp.polylog(2, -x) + 2*mp.log(1+x))/x**2, [0, mp.mpf(1)/2, 1])
    num("Eq.(D19)  log piece = 3 zeta(3)/8 - pi^2/4 + log^2 2", I_log, 3*mp.zeta(3)/8 - mp.pi**2/4 + mp.log(2)**2, tol=1e-30)
    num("Eq.(D19)  Li2 piece = pi^2 log2/6 - 9 zeta(3)/8 + pi^2/4 - log^2 2", I_li,
        mp.pi**2*mp.log(2)/6 - 9*mp.zeta(3)/8 + mp.pi**2/4 - mp.log(2)**2, tol=1e-30)

D0  = pi**3/18*sp.log(2) - pi/4*sp.zeta(3)
e2x = sp.log(2)/3 - 3*sp.zeta(3)/(2*pi**2)
exact("Eq.(D6)  6 D0/pi^3 = e2x (Onsager-Mittag-Stephen)", 6/pi**3*D0 - e2x)
exact("Eq.(D6)  64 pi^2 A0 = 6/pi^3", 64*pi**2*A0 - 6/pi**3)
print("e2x =", sp.N(e2x, 12), "Ry")
```

## Check 4 — Eq. (F2), (F9)–(F11): endpoint coefficients $a_0$, $b_0$ of the profile

**Where we are in the argument.** The Mellin transform $\widetilde\Phi(s)$ of $\Phi$ has poles dictated by the endpoint behaviour of $\Phi$.  The small-$y$
form $a_0y\log y+b_0y$ produces a *double* pole at $s=-1$ in $\widetilde\Phi(s)$, i.e. at $s=1$ in $\widetilde\Phi(-s)$; this is the origin of the $\log\Lambda/\Lambda$ term of the
Lorentzian-kernel energy.  The coefficients $a_0,b_0$ follow from a Weber–Schafheitlin integral.

**Claim.** $\Phi(y)=a_0y\log y+b_0y+O(y^2\log^2y)$ with $a_0=2\int_0^\infty j_1(t/2)^2\,dt/t=2I_j(0)$, $b_0=2\big[(1-\log2)I_j(0)-I_j'(0)\big]$, where
$$I_j(s)=\int_0^\infty x^{s-1}j_1(x)^2dx=\frac\pi2\frac{\Gamma(2-s)\Gamma(1+s/2)}{2^{2-s}\Gamma((3-s)/2)^2\Gamma(3-s/2)},\qquad -2<\Re s<2\quad\text{(F11)},$$
so that $I_j(0)=\tfrac14$, $I_j'(0)=\tfrac14(\tfrac74-\gamma-\log2)$, $a_0=\tfrac12$, $b_0=\tfrac\gamma2-\tfrac38$ (F9), (F10).

**Manuscript route.** Rescale $t=uy$ in the large-$u$ region of Eq. (D3), then read the coefficients from the Weber–Schafheitlin formula.

**How this notebook verifies it.**
1. The Weber–Schafheitlin formula is checked numerically at $s=-1$ (`mp.quadosc`, `omega=2` because $j_1^2$ oscillates at twice the Bessel frequency).
   *Why $s=-1$ and not $s=\tfrac12$:* the integrand decays as $x^{s-3}$; at $s=-1$ the tail is $x^{-4}$ and the oscillatory quadrature converges to $10^{-13}$,
   whereas at $s=\tfrac12$ the $x^{-5/2}$ tail limits it to $\sim10^{-7}$. Any $s$ inside the strip tests the same formula.
2. `sp.limit(Is, s, 0)` and `sp.limit(sp.diff(Is, s), s, 0)` evaluate $I_j(0)$ and $I_j'(0)$ from the Gamma-function form (the derivative involves digamma values, which SymPy reduces exactly).
3. `a0`, `b0` are assembled exactly as in (F9)–(F10) and compared with $\tfrac12$ and $\tfrac\gamma2-\tfrac38$.

**Reading the output.** One `num` (formula), four `exact`. The printed line shows $b_0=-0.0864$.

**Hand-first.** That the rescaling $t=uy$ isolates the correct region, and the error estimate $O(y^2\log^2y)$ (App. F.2.1). Check 21 tests the resulting
two-term form against the full 2-D integral.

```python
I_ws = lambda s: mp.pi/2*mp.gamma(2-s)*mp.gamma(1+s/2)/(2**(2-s)*mp.gamma((3-s)/2)**2*mp.gamma(3-s/2))
sv = mp.mpf(-1)          # inside the strip -2 < Re s < 2; fast x^-4 decay makes the oscillatory quadrature accurate
num("Eq.(F11)  Weber-Schafheitlin I_j(-1): quadrature = Gamma formula",
    mp.quadosc(lambda x: x**(sv-1)*j1(x)**2, [0, mp.inf], omega=2), I_ws(sv), tol=1e-10)

s = sp.symbols('s')
Is = pi/2*sp.gamma(2-s)*sp.gamma(1+s/2)/(2**(2-s)*sp.gamma((3-s)/2)**2*sp.gamma(3-s/2))
I0  = sp.limit(Is, s, 0)
I0p = sp.limit(sp.diff(Is, s), s, 0)
exact("Eq.(F11)  I_j(0) = 1/4", I0 - R(1,4))
exact("Eq.(F11)  I_j'(0) = (7/4 - gamma - log 2)/4", I0p - (R(7,4) - sp.EulerGamma - sp.log(2))/4)
a0 = 2*I0
b0 = 2*((1-sp.log(2))*I0 - I0p)
exact("Eq.(F9)  a0 = 1/2", a0 - R(1,2))
exact("Eq.(F10)  b0 = gamma/2 - 3/8", b0 - (sp.EulerGamma/2 - R(3,8)))
print("a0 =", a0, "  b0 =", b0, "=", sp.N(b0, 10))
```

## Check 5 — Eqs. (F5), (F6), (F26)–(F29): the two large-$\Lambda$ coefficients

**Where we are in the argument.** Mellin inversion writes $S(\Lambda)$ as a contour integral of $\widetilde\Xi(s)\widetilde\Phi(-s)\Lambda^{-s}$; shifting the contour
to the right across $s=0$ (residue $-1$, giving the static limit $S=1$) and $s=1$ (the double pole) gives the first two terms of the large-$\Lambda$
expansion.  Appendix F.4 reaches the same two coefficients by real integration; both routes are checked here.

**Claim.** With $\widetilde\Phi(-s)=-a_0/(s-1)^2-b_0/(s-1)+O(1)$, $\widetilde\Xi(s)=\widetilde\Xi(1)+(s-1)\widetilde\Xi'(1)+\dots$, $\Lambda^{-s}=\Lambda^{-1}[1-(s-1)\log\Lambda+\dots]$:
$$\operatorname{Res}_{s=1}I=\frac{a_0\widetilde\Xi(1)\log\Lambda-a_0\widetilde\Xi'(1)-b_0\widetilde\Xi(1)}{D_0\Lambda}\ \text{(Eq. (F5))},\qquad
S(\Lambda)=1+\frac{C_1^\Lambda\log\Lambda+C_0^\Lambda}{\Lambda}+\dots,\ C_1^\Lambda=\frac{\pi}{4D_0}=3.14282593\ldots,\ C_0^\Lambda=-\frac{\pi}{16D_0}\ \text{(Eq. (F6))}.$$
The clockwise closure *subtracts* the residues: $S=1-\operatorname{Res}_{s=1}I+\dots$ (Sec. 4.4 and Appendix F).
Appendix F.4: $\int_0^\infty[\Xi(z)-1]\,dz=-\tfrac\pi2=\widetilde\Xi(1)$ and $\int_0^\infty\log z\,[\Xi(z)-1]\,dz=-\tfrac\pi2(1-\gamma)=\widetilde\Xi'(1)$, which give the same coefficients
by real integration (Eq. (F29)).

**Manuscript route.** Mellin inversion, contour shift across $s=0,1$, residues.

**How this notebook verifies it.**
1. The residue at the double pole is extracted symbolically: `laurent` is the product of the three Laurent/Taylor factors with $\epsilon=s-1$ (the overall
   $1/(D_0\Lambda)$ is left out), and `.coeff(eps, -1)` picks the coefficient of $\epsilon^{-1}$. This is a check of the *algebra* of Eq. (F5), with
   $a_0,b_0,\widetilde\Xi(1),\widetilde\Xi'(1),\log\Lambda$ as free symbols.
2. `C1 = -(a0*K1v)/D0`: the sign here encodes “subtract the residue.” `C0` likewise. Both are compared exactly with $\pi/(4D_0)$ and $-\pi/(16D_0)$, and $C_1$ numerically.
3. **Departure from the manuscript's main route — Appendix F.4 is evaluated numerically.** The two real moments of $\Xi-1$ are integrated directly.
   *Numerical detail:* $\Xi(z)-1\sim-2/z^2$, so the integrals converge slowly and the Ci/Si evaluation loses accuracy at very large $z$. The code therefore
   truncates at $Z=2000$ and adds the analytic tail from $\Xi-1=-2/z^2+24/z^4-\dots$ (i.e. $\int_Z^\infty(\Xi-1)\,dz\approx-2/Z+8/Z^3$, and the corresponding
   expression with $\log z$). Agreement to $\sim10^{-14}$ with $\widetilde\Xi(1)$ and $\widetilde\Xi'(1)$ (tolerance $10^{-8}$ leaves margin for the truncation).

   Because Eq. (F29) obtains $C_1^\Lambda$, $C_0^\Lambda$ from these two moments *without any contour shift*, this cell verifies the two coefficients by two
   logically independent routes: the residue algebra (items 1–2) and the real moments (item 3).

**Reading the output.** Three `exact` + one `num` (residue route); two `num` (moment route).

**Hand-first.** The direction of closure and its sign (Appendix F); vanishing of the horizontal segments (App. F.3, Stirling bound).

```python
eps, A0s, B0s, Lg, XiT1, XiT1p = sp.symbols('epsilon a_0 b_0 logLambda XiT1 XiT1p')
laurent = (-A0s/eps**2 - B0s/eps)*(XiT1 + eps*XiT1p)*(1 - eps*Lg)      # eps = s-1, overall 1/(D0 Lambda) omitted
res = sp.expand(laurent).coeff(eps, -1)
exact("Eq.(F5)  Res_{s=1} = a0 XiT(1) logL - a0 XiT'(1) - b0 XiT(1)", res - (A0s*XiT1*Lg - A0s*XiT1p - B0s*XiT1))

XiT1_val, XiT1p_val = -pi/2, -pi*(1-sp.EulerGamma)/2
C1 = -(a0*XiT1_val)/D0                    # S = 1 - Res  =>  coefficient of logL/L is -a0 XiT(1)/D0
C0 = (a0*XiT1p_val + b0*XiT1_val)/D0
exact("Eq.(F6)  C1 = pi/(4 D0)",   C1 - pi/(4*D0))
exact("Eq.(F6)  C0 = -pi/(16 D0) = -C1/4", C0 + pi/(16*D0))
num("Eq.(F6)  C1 = 3.14282593...", sp.N(C1, 25), mp.mpf('3.14282593'), tol=1e-8)

# App. F moments: truncate at Z and add the asymptotic tail  Xi-1 = -2/z^2 + 24/z^4 - ...
Z = mp.mpf(2000)
m0 = mp.quad(lambda z: Xi_ci(z)-1,            [0, 1, 10, 100, Z]) + (-2/Z + 8/Z**3)
m1 = mp.quad(lambda z: mp.log(z)*(Xi_ci(z)-1), [0, 1, 10, 100, Z]) + (-2*(mp.log(Z)+1)/Z + 8*(mp.log(Z)+mp.mpf(1)/3)/Z**3)
num("App.F  int (Xi-1) dz = -pi/2 = XiT(1)",                m0, -mp.pi/2,               tol=1e-8)
num("App.F  int log z (Xi-1) dz = -pi(1-gamma)/2 = XiT'(1)", m1, -mp.pi/2*(1-mp.euler), tol=1e-8)
```

## Check 6 — Eq. (H3): the large-momentum Lorentzian kernel profile $p_{\rm loc}(s_\Lambda)$

**Where we are in the argument.** From here on the Lorentzian-kernel energy is resolved in momentum transfer, $\mathcal K[b_\Lambda]=\int e_\Lambda(q)dq$.
At large $q$ only the same-sphere sectors survive at leading order, with both energies at $\pm q^2/2$; the frequency integral with the Lorentzian kernel gives
the profile $p_{\rm loc}(s_\Lambda)$, $s_\Lambda=\Lambda/q$, which will be integrated along the dispersion curve in Check 12.  Its maximum at $s_\Lambda=\tfrac12$ is the
free-particle branch.

**Claim.** For $q>2$ the two same-sphere sectors give
$e_\Lambda(q)=32\pi^2A_0q^{-2}\int_0^\infty du\,\frac{s_\Lambda^2(\tfrac14-u^2)}{(u^2+s_\Lambda^2)(u^2+\tfrac14)^2}=\frac{3}{\pi^3}\frac{p_{\rm loc}(s_\Lambda)}{q^2}$, $p_{\rm loc}(s_\Lambda)=\frac{2\pi s_\Lambda}{(1+2s_\Lambda)^2}$,
maximal at $s_\Lambda=\tfrac12$ (free-particle branch).

**Manuscript route.** $\frac{\tfrac14-u^2}{(u^2+\tfrac14)^2}=-\partial_b\frac{b}{u^2+b^2}\big|_{b=1/2}$ and $\int_0^\infty\frac{du}{(u^2+s_\Lambda^2)(u^2+b^2)}=\frac{\pi}{2s_\Lambda b(s_\Lambda+b)}$.

**How this notebook verifies it.**
1. `exact`: the derivative identity that generates the double pole.
2. `Ib` is the elementary integral quoted in the manuscript (not re-derived; it is a standard partial-fraction result). `p_loc_from_route` $=-s_\Lambda^2\partial_b[b\,I_b]_{b=1/2}$
   reproduces the manuscript's two-line evaluation and is compared with $p_{\rm loc}(s_\Lambda)$ exactly. Note the factor $b$ inside the derivative — it is the numerator
   of $b/(u^2+b^2)$ and is easy to drop by hand.
3. `exact`: $32\pi^2A_0=3/\pi^3$ and $P'(\tfrac12)=0$.
4. **Departure from the manuscript — the $u$-integral is also done by quadrature** at $s_\Lambda=0.3,0.8,4$ and compared with $p_{\rm loc}(s_\Lambda)$ (12+ digits; the integrand is smooth
   and the agreement is in fact $\sim10^{-31}$).

**A caution recorded here for CAS users.** During development, SymPy 1.14 (`sp.integrate(..., (u,0,oo))`) returned
`2*pi*s_lambda*(4*s_lambda**2+1)/(16*s_lambda**4-8*s_lambda**2+1)` for this integral, which is *wrong* (e.g. 7.35 instead of 0.7436 at $s_\Lambda=0.8$). The manuscript's derivative route
and the 30-digit quadrature agree with each other. This is why the notebook never relies on a symbolic definite integral with parameters without a numerical cross-check.

**Reading the output.** Four `exact`, three `num`.

**Hand-first.** The sector decomposition for $q>2$ and the error terms $O(q^{-3}/(1+s_\Lambda))+O(q^{-4})$ (Sec. 5.3).

```python
u, s_lambda, bb = sp.symbols('u s_lambda b', positive=True)
exact("Eq.(H3)  (1/4-u^2)/(u^2+1/4)^2 = -d/db [b/(u^2+b^2)] at b=1/2",
      -sp.diff(bb/(u**2+bb**2), bb).subs(bb, R(1,2)) - (R(1,4)-u**2)/(u**2+R(1,4))**2)
Ib = pi/(2*s_lambda*bb*(s_lambda+bb))                                   # int_0^inf du /[(u^2+s_lambda^2)(u^2+b^2)]
p_loc_from_route = -s_lambda**2*sp.diff(bb*Ib, bb).subs(bb, R(1,2))   # -s_lambda^2 d/db [ b * Ib ]
p_loc_expr = 2*pi*s_lambda/(1+2*s_lambda)**2
exact("Eq.(H3)  -s_lambda^2 d/db[b Ib]_{b=1/2} = p_loc(s_lambda) = 2 pi s_lambda/(1+2s_lambda)^2", p_loc_from_route - p_loc_expr)
exact("Eq.(H3)  32 pi^2 A0 = 3/pi^3", 32*pi**2*A0 - 3/pi**3)
exact("Eq.(H3)  P'(s_lambda)=0 at s_lambda=1/2", sp.diff(p_loc_expr, s_lambda).subs(s_lambda, R(1,2)))
for s_lambda_value in [mp.mpf('0.3'), mp.mpf('0.8'), mp.mpf(4)]:
    Pn = mp.quad(lambda uu: s_lambda_value**2*(mp.mpf(1)/4-uu**2)/((uu**2+s_lambda_value**2)*(uu**2+mp.mpf(1)/4)**2), [0, 1, mp.inf])
    num("Eq.(H3)  quadrature of the u-integral = p_loc(s_lambda), s_lambda=%s" % s_lambda_value, Pn, 2*mp.pi*s_lambda_value/(1+2*s_lambda_value)**2)
```

## Check 7 — Eqs. (G2)–(G6): the Fermi-surface shell and $\kappa(\Lambda)$

**Where we are in the argument.** At small $q$ the occupied region is a thin shell around the Fermi surface and the Coulomb coincidence of the
two loop momenta produces a logarithm $\log(1/q)$.  Its coefficient $\kappa(\Lambda)$ is the first ingredient of the high-density analysis
(Check 13, where the same angular integral appears as $A(z)$), and $\kappa(\infty)=0$ is again the statement that a static line kills the coincidence.

**Claim.** The coincidence term of the shell integral is governed by $\int_0^\infty\frac{\Lambda^2(c^2-z^2)\,dz}{(\Lambda^2+z^2)(c^2+z^2)^2}=\frac{\pi\Lambda}{2(\Lambda+c)^2}$ ($c>0$, display after Eq. (G6)), and
$$\kappa(\Lambda)=\frac{3}{2\pi^2}\Lambda\int_0^1\frac{c^2dc}{(c+\Lambda)^2}=\frac{3\Lambda}{2\pi^2}\Big[1-2\Lambda\log\big(1+\tfrac1\Lambda\big)+\frac{\Lambda}{1+\Lambda}\Big]\quad\text{(Eq. (G5))},$$
$\kappa\simeq\frac{3\Lambda}{2\pi^2}$ ($\Lambda\to0$), $\kappa\simeq\frac{1}{2\pi^2\Lambda}$ ($\Lambda\to\infty$); $\beta(\infty)=\tfrac6{\pi^3}c_0=\tfrac{3(2\log2-1)}{4\pi^2}$ (after Eq. (G6));
the shell bound $|ab\,R(a,b;z)|=O((1+z^2)^{-1})$ (Eq. (G4)).

**How this notebook verifies it.**
1. **The $z$-integral is done by quadrature** (two parameter pairs, 30-digit agreement) rather than symbolically — see the caution in Check 6.
2. The $c$-integral is elementary and SymPy handles it reliably (`exact`); the two limits are `sp.limit` calls on $\kappa/\Lambda$ and $\kappa\Lambda$.
3. `exact`: $\beta(\infty)$ from $c_0$ of Eq. (I3).
4. Eq. (G4) is a bound, not an identity; the code *samples* $|ab\,R|(1+z^2)$ on a grid of $(a,b,z)$ including $z\to0$ and $z=30$ and reports the largest value
   (`check` with the threshold 3). This is a plausibility check of the constant, not a proof of the bound.

**Reading the output.** Two `num`, four `exact`, one sampled `check` (prints the maximum found, 1.000).

**Hand-first.** The thin-shell reduction itself (radial thickness $q|c_p|$, the $\operatorname{Fp}$ prescription, and the $\tfrac32$ in Eq. (G2)) — this is the geometric
core of Sec. 5.2 and is not touched by the CAS.

```python
c, Lam, z = sp.symbols('c Lambda z', positive=True)
for (Lv, cv) in [(mp.mpf('1.3'), mp.mpf('0.4')), (mp.mpf('0.2'), mp.mpf('0.9'))]:
    Iz = mp.quad(lambda zz: Lv**2*(cv**2-zz**2)/((Lv**2+zz**2)*(cv**2+zz**2)**2), [0, 1, mp.inf])
    num("Eq.(G5)  frequency integral = pi Lam/(2(Lam+c)^2), Lam=%s c=%s" % (Lv, cv), Iz, mp.pi*Lv/(2*(Lv+cv)**2))
kc = sp.integrate(c**2/(c+Lam)**2, (c, 0, 1))
bracket = 1 - 2*Lam*sp.log(1+1/Lam) + Lam/(1+Lam)
exact("Eq.(G5)  int_0^1 c^2/(c+Lam)^2 dc = 1 - 2 Lam log(1+1/Lam) + Lam/(1+Lam)", kc - bracket)
kappa = R(3,2)/pi**2*Lam*bracket
exact("Eq.(G5)  kappa ~ 3 Lam/(2 pi^2) as Lam->0", sp.limit(kappa/Lam, Lam, 0) - R(3,2)/pi**2)
exact("Eq.(G5)  kappa ~ 1/(2 pi^2 Lam) as Lam->oo", sp.limit(kappa*Lam, Lam, oo) - 1/(2*pi**2))
c0_static = pi/8*(2*sp.log(2)-1)
exact("Eq.(G6)  beta(inf) = 6 c0/pi^3 = 3(2 log2 - 1)/(4 pi^2)", 6*c0_static/pi**3 - 3*(2*sp.log(2)-1)/(4*pi**2))
# Eq.(G4): shell bound, sampled
Rab = lambda a_, b_, z_: (a_*b_ - z_**2)/((a_*b_ - z_**2)**2 + z_**2*(a_+b_)**2)
worst = max(abs(a_*b_*Rab(a_, b_, z_))*(1+z_**2) for a_ in [0.1, 0.5, 1.0, -0.7] for b_ in [0.05, 0.9, -1.0] for z_ in [0.01, 0.5, 2.0, 30.0])
check("Eq.(G4)  |ab R(a,b;z)| (1+z^2) bounded (sampled max = %.3f)" % worst, worst < 3)
```

## Check 8 — (C6)–(C12) and Eq. (I1): the dynamical frequency kernel $J(\omega;a,b)$

**Where we are in the argument.** The frequency integral of two energy denominators against a Lorentzian kernel, $J(\omega;a,b)$, is the elementary kernel
behind $p_{\rm loc}(s_\Lambda)$, $\kappa(\Lambda)$, $\beta(\Lambda)$ and the plasmon limit.  Its closed form has two sign cases; its static limit $\mathcal W(a,b)$ vanishes for
$ab>0$.  This is the mechanism of Sec. 5.4: a static line loses the same-sphere sectors (hence $r_s^{-3/4}$), a dynamical line keeps them
(hence $r_s^{-1/4}$).  The $1/(2\omega)$ correction is what generates the companion's $A_2$.

**Claim.** With $\alpha=|a|,\beta=|b|$ and the partial fraction $r_c=\prod_{c'\ne c}(c'^2-c^2)^{-1}$ (C7)–(C8):
$\frac1\pi\int_0^\infty\frac{d\nu}{\mathcal D}=\frac{\alpha+\beta+\omega}{2\alpha\beta\omega(\alpha+\beta)(\beta+\omega)(\omega+\alpha)}$ (C10),
$\frac1\pi\int_0^\infty\frac{\nu^2d\nu}{\mathcal D}=\frac{1}{2(\alpha+\beta)(\beta+\omega)(\omega+\alpha)}$ (C11), and
$$J=\begin{cases}\dfrac{1}{2\omega(\alpha+\omega)(\beta+\omega)},&ab>0\\[6pt]-\dfrac{\alpha+\beta+2\omega}{2\omega(\alpha+\beta)(\alpha+\omega)(\beta+\omega)},&ab<0\end{cases}\quad\text{(C12)}.$$
Static limit: $\omega^2J\to\mathcal W(a,b)=\frac{\operatorname{sgn}(ab)-1}{2(|a|+|b|)}$ (Eq. (I1)) with first correction $+1/(2\omega)$ for both signs (Sec. 5.3).

**Manuscript route.** Partial fractions, $\int_0^\infty(c^2+\nu^2)^{-1}d\nu=\pi/(2c)$, and for the $\nu^2$ numerator the fact that $\sum_cr_c=0$.

**How this notebook verifies it.**
1. `exact`: the partial-fraction identity in the auxiliary variable $x=\nu^2$, and $\sum_cr_c=0$ (needed so that $\nu^2/(c^2+\nu^2)=1-c^2/(c^2+\nu^2)$ produces no divergent constant).
2. `exact`: (C10) as $\sum_cr_c/(2c)$ and (C11) as $-\sum_cr_c\,c^2/(2c)$; then (C12) for both signs by substituting $P=\pm\alpha\beta$.
3. `sp.series(om**2*Jp, om, oo, 2)` expands $\omega^2J$ at $\omega=\infty$ keeping the constant and $1/\omega$ terms; `exact` shows they are $0+\frac{1}{2\omega}$ for $ab>0$ and
   $-\frac{1}{\alpha+\beta}+\frac{1}{2\omega}$ for $ab<0$. The first is $\mathcal W=0$ for same-sign arguments (the same-sphere cancellation of Sec. 5.4); the $1/(2\omega)$
   is the term that generates $A_2=8/15$ in the companion paper.
4. **Departure from the manuscript — $J$ is integrated numerically** at five parameter sets, including the removable points $\omega=|a|$ and $|a|=|b|$ and a mixed-sign case,
   against the closed forms (30-digit agreement). The partial-fraction route is singular at those points; the closed forms are not, and the quadrature confirms them there.

**Reading the output.** Eight `exact`, five `num`.

**Hand-first.** Why the *same-sphere* sectors ($ab>0$) drop out of the static kernel but not of the Lorentzian kernel (Sec. 5.4) — the physics behind $r_s^{-1/4}$ vs $r_s^{-3/4}$;
the CAS only shows $\mathcal W=0$ for equal signs.

```python
al, be, om, x = sp.symbols('alpha beta omega x', positive=True)
ra = 1/((be**2-al**2)*(om**2-al**2)); rb = 1/((al**2-be**2)*(om**2-be**2)); ro = 1/((al**2-om**2)*(be**2-om**2))
exact("Eq.(C7) partial fractions of 1/[(x+a^2)(x+b^2)(x+w^2)]",
      1/((x+al**2)*(x+be**2)*(x+om**2)) - (ra/(x+al**2) + rb/(x+be**2) + ro/(x+om**2)))
exact("Eq.(C7) sum of residues r_c = 0 (needed for the nu^2 numerator)", ra + rb + ro)
I0c = (al+be+om)/(2*al*be*om*(al+be)*(be+om)*(om+al)); I2c = 1/(2*(al+be)*(be+om)*(om+al))
exact("Eq.(C10) (1/pi) int dnu/D  = sum r_c/(2c)",            ra/(2*al) + rb/(2*be) + ro/(2*om) - I0c)
exact("Eq.(C11) (1/pi) int nu^2 dnu/D = -sum r_c c^2/(2c)",  -(ra*al/2 + rb*be/2 + ro*om/2) - I2c)
Jp = al*be*I0c - I2c; Jm = -al*be*I0c - I2c
exact("Eq.(C12) J for ab>0", Jp - 1/(2*om*(al+om)*(be+om)))
exact("Eq.(C12) J for ab<0", Jm + (al+be+2*om)/(2*om*(al+be)*(al+om)*(be+om)))
# static limit and 1/(2 omega) correction (Sec. 5.4, Eq. (I1))
sp_ = sp.series(om**2*Jp, om, oo, 2).removeO(); sm_ = sp.series(om**2*Jm, om, oo, 2).removeO()   # constant + 1/omega terms
exact("Eq.(I1)  omega^2 J -> W = 0 for ab>0, next term 1/(2 omega)", sp.simplify(sp_ - 1/(2*om)))
exact("Eq.(I1)  omega^2 J -> W = -1/(|a|+|b|) for ab<0, next term 1/(2 omega)", sp.simplify(sm_ - (-1/(al+be) + 1/(2*om))))

def J_num(w, a_, b_):
    return mp.quad(lambda nu: (a_*b_-nu**2)/((a_*a_+nu*nu)*(b_*b_+nu*nu)*(nu*nu+w*w)), [0, 1, mp.inf])/mp.pi
def J_cl(w, a_, b_):
    A_, B_ = abs(a_), abs(b_)
    return 1/(2*w*(A_+w)*(B_+w)) if a_*b_ > 0 else -(A_+B_+2*w)/(2*w*(A_+B_)*(A_+w)*(B_+w))
for (w, a_, b_) in [('1.5', '0.7', '2.1'), ('1.5', '-0.7', '2.1'), ('0.7', '0.7', '0.7'), ('0.7', '-0.7', '0.7'), ('2', '2', '-0.3')]:
    w, a_, b_ = mp.mpf(w), mp.mpf(a_), mp.mpf(b_)
    num("App.C  J(%s;%s,%s): quadrature = closed form" % (w, a_, b_), J_num(w, a_, b_), J_cl(w, a_, b_))
```

## Check 9 — Eq. (H6): the small-$q$ logarithm of the signed overlap $Z(q)$

**Where we are in the argument.** The signed overlap $Z(q)=\int\Delta\Delta/|\bm p-\bm k|^2$ multiplies the $1/(2\omega)$ term of $J$ in the
plasmon limit and is the Fermi-sphere factor of the companion's outer region.  Its small-$q$ behaviour $\tfrac{8\pi^2}{3}q^2\log(1/q)$, stated in Eq. (H6)
from a Fourier representation, is confirmed numerically here.

**Claim.** $Z(q)=\frac{2\pi^2}{(2\pi)^3}\int d^3t\,\frac{|\hat\chi_B(t)|^2}{t}\,2[1-\cos(qt_z)]$, $\hat\chi_B(t)=4\pi\frac{\sin t-t\cos t}{t^3}$, behaves as
$Z(q)=\frac{8\pi^2}{3}q^2\log\frac1q+O(q^2)$; “the momentum interval $1\ll t\ll1/q$ supplies $8\pi^2/3$.”

**Manuscript route.** Asymptotics: for $t\gg1$, $t|\hat\chi_B|^2\to8\pi^2/t^3$ on average and $1-\sin x/x\to x^2/6$, so the logarithmic interval gives the coefficient. No numerical value is displayed.

**How this notebook verifies it — independent numerical route (not in the manuscript).**
1. *Angular integration (not written in the manuscript).* $\int d\Omega\,2[1-\cos(qt\cos\theta)]=8\pi\big[1-\frac{\sin qt}{qt}\big]$, so
   $Z(q)=2\int_0^\infty t\,|\hat\chi_B(t)|^2\big[1-\frac{\sin qt}{qt}\big]dt=32\pi^2\int_0^\infty\frac{(\sin t-t\cos t)^2}{t^5}\,g(qt)\,dt$ with $g(x)=1-\sin x/x$.
2. *Splitting for accuracy.* The integrand oscillates out to $t\sim1/q$ and beyond, which tanh–sinh cannot resolve at $q=10^{-4}$. The code therefore writes
   $(\sin t-t\cos t)^2=\frac{1+t^2}{2}+\big[-\frac{\cos2t}{2}-t\sin2t+\frac{t^2\cos2t}{2}\big]$ and treats
   - `A`: $t\in[0,1]$ with the exact integrand (no oscillation issue);
   - `B`: the non-oscillatory part on $[1,\infty)$, rewritten in $x=qt$ as $\frac{q^4}2\int_q^\infty\frac{g(x)}{x^5}dx+\frac{q^2}{2}\int_q^\infty\frac{g(x)}{x^3}dx$;
   - `C`: the oscillatory part on $[1,\infty)$ with `mp.quadosc(..., omega=2)`.
   `g_` uses the series $x^2/6-x^4/120+\dots$ below $x=10^{-3}$ to avoid cancellation in $1-\sin x/x$.
3. $Z(q)$ is evaluated at $q=10^{-2},10^{-3},10^{-4}$ and the finite-difference slope $\Delta[Z/q^2]/\Delta\log(1/q)$ is formed. If $Z=Aq^2\log(1/q)+Bq^2+o(q^2)$, the slope
   converges to $A$; the printed slopes are $26.3193$ and $26.31895$ against $8\pi^2/3=26.318945$. The test requires agreement to $10^{-5}$ between the last slope
   and the target (the residual $o(q^2)$ terms are what limits it).
4. The working precision is lowered to 20 digits inside this cell for speed and restored afterwards.

**Reading the output.** The printed slopes and one `num`. This is a quantitative confirmation of a coefficient that the manuscript obtains only by an asymptotic argument.

**Hand-first.** That the *signed* weight $Z$ (not $\bar Z$) multiplies $1/(2\omega)$ in the plasmon limit, and the $O(q^2\log(e/q))$ bound on $\bar Z$ (Sec. H2).

```python
def g_(x): return x**2/6 - x**4/120 + x**6/5040 if x < mp.mpf('1e-3') else 1 - mp.sin(x)/x
def Zsigned(q):
    # split (sin t - t cos t)^2 = (1+t^2)/2 + [oscillatory]; t<1 handled exactly, t>1 non-osc./osc. separately
    A = mp.quad(lambda t: (mp.sin(t)-t*mp.cos(t))**2*g_(q*t)/t**5, [0, mp.mpf(1)/2, 1])
    B = q**4/2*mp.quad(lambda xx: g_(xx)/xx**5, [q, 1, 10, 100, mp.inf]) + q**2/2*mp.quad(lambda xx: g_(xx)/xx**3, [q, 1, 10, 100, mp.inf])
    C = mp.quadosc(lambda t: (-mp.cos(2*t)/2 - t*mp.sin(2*t) + t**2*mp.cos(2*t)/2)*g_(q*t)/t**5, [1, mp.inf], omega=2)
    return 32*mp.pi**2*(A + B + C)          # 2 * (4 pi)^2 * ...
mp.mp.dps = 20
qs = [mp.mpf('1e-2'), mp.mpf('1e-3'), mp.mpf('1e-4')]
Zs = [Zsigned(q) for q in qs]
slopes = [(Zs[i+1]/qs[i+1]**2 - Zs[i]/qs[i]**2)/mp.log(qs[i]/qs[i+1]) for i in range(2)]
mp.mp.dps = 30
print("slopes d[Z/q^2]/dlog(1/q):", [mp.nstr(s_, 10) for s_ in slopes], "  target 8pi^2/3 =", mp.nstr(8*mp.pi**2/3, 10))
num("Eq.(H6)  Z(q) ~ (8 pi^2/3) q^2 log(1/q): slope between q=1e-3 and 1e-4", slopes[1], 8*mp.pi**2/3, tol=1e-5)
```

## Check 10 — Eq. (H8): large-momentum expansion of the imaginary-axis Lindhard function

**Where we are in the argument.** To apply the Lorentzian representation to the physical RPA line one needs the imaginary-axis Lindhard function $\mathcal L$ at
large momentum transfer: $\mathcal L(q,q^2u)\simeq\frac4{3q^2(1+4u^2)}$, a single Lorentzian in $u$, which is exactly the frequency dependence of a Lorentzian kernel.
The same expansion, with its $q^{-4}$ correction, is Eq. (H3) of the companion.

**Claim.** From $\mathcal L=\frac{1}{2q}\int_0^1k\log\frac{\varpi+k/q+k^2/q^2}{\varpi-k/q+k^2/q^2}dk$, $\varpi=u^2+\tfrac14$ (display after Eq. (H8)),
$$\mathcal L(q,q^2u)=\frac{4}{3q^2(1+4u^2)}\Big[1+\frac{g_4(u)}{q^2}+O(q^{-4})\Big],\qquad g_4(u)=-\frac{4(12u^2-1)}{5(1+4u^2)^2},$$
odd powers absent. Also $\mathcal L(q,0)=f_L(q/2)$ and $\mathcal L(q,qz)\to h(z)=1-z\arctan(1/z)$ as $q\to0$ (Sec. 5.2).

**How this notebook verifies it.**
1. *Reparametrization.* With $w=1/q$ the logarithm is $\log\frac{A+wk+w^2k^2}{A-wk+w^2k^2}$ (`Lw`). `sp.series` expands it to $w^6$, `sp.integrate` does the $k$-integral term by term,
   the prefactor $w/2$ is the $\frac1{2q}$, and `A` is then replaced by $u^2+\tfrac14$. `lindhard_series.coeff(w, n)` reads off the $q^{-n}$ coefficients.
2. `exact` for the $q^{-2}$ and $q^{-4}$ coefficients; odd powers tested as a sum of squares.
3. **Departure from the manuscript — numerical checks from the exact integral.** `lindhard_shape_exact(q, nu)` is the imaginary-axis Lindhard function of the companion paper,
   Eq. (F3) there, integrated numerically (breakpoint at $k=q/2$ where the log argument is smallest). Three limits are tested: the static limit against $f_L(q/2)$ (`fL`),
   the small-$q$ shell limit against $h(z)$ at $q=10^{-4}$ (tolerance $10^{-3}$, set by the $O(q)$ corrections), and the two-term large-$q$ form at $q=30$ (tolerance $10^{-6}$,
   set by the omitted $O(q^{-6})$).

**Reading the output.** Three `exact`, three `num`.

**Hand-first.** Uniformity of the expansion in $u$ ($\varpi\ge\tfrac14$ keeps the log arguments away from zero, stated after Eq. (H8)).

```python
w, A, k = sp.symbols('w A k', positive=True)
Lw  = sp.log((A + w*k + w**2*k**2)/(A - w*k + w**2*k**2))
ser = sp.series(Lw, w, 0, 6).removeO()
lindhard_series   = sp.expand(sp.integrate(sp.expand(k*ser), (k, 0, 1))*w/2)
lindhard_u_series  = lindhard_series.subs(A, u**2 + R(1,4))
lead = R(4,3)/(1+4*u**2);  g4 = -4*(12*u**2-1)/(5*(1+4*u**2)**2)
exact("Eq.(H8)  coefficient of q^-2 is 4/[3(1+4u^2)]", lindhard_u_series.coeff(w, 2) - lead)
exact("Eq.(H8)  coefficient of q^-4 is (4/[3(1+4u^2)]) g4(u)", lindhard_u_series.coeff(w, 4) - lead*g4)
exact("Eq.(H8)  no odd powers q^-1, q^-3", lindhard_u_series.coeff(w, 1)**2 + lindhard_u_series.coeff(w, 3)**2)

def lindhard_shape_exact(q, nu):
    f = lambda kk: kk*mp.log(((q*kk+q*q/2)**2 + nu**2)/((q*kk-q*q/2)**2 + nu**2))
    return mp.quad(f, [0, q/2, 1])/(2*q) if q < 2 else mp.quad(f, [0, 1])/(2*q)
fL = lambda x_: mp.mpf(1)/2 + (1-x_**2)*mp.log(abs((1+x_)/(1-x_)))/(4*x_)
num("L(q,0) = f_L(q/2), q=0.8", lindhard_shape_exact(mp.mpf('0.8'), mp.mpf('1e-30')), fL(mp.mpf('0.4')), tol=1e-10)
h = lambda zz: 1 - zz*mp.atan(1/zz)
num("L(q,qz) -> h(z)=1-z arctan(1/z), q=1e-4, z=0.7", lindhard_shape_exact(mp.mpf('1e-4'), mp.mpf('0.7e-4')), h(mp.mpf('0.7')), tol=1e-3)
num("L(q,q^2u) -> 4/[3q^2(1+4u^2)] (1+g4/q^2), q=30, u=0.6",
    lindhard_shape_exact(mp.mpf(30), mp.mpf(30)**2*mp.mpf('0.6')),
    mp.mpf(4)/(3*900*(1+4*mp.mpf('0.36')))*(1 - 4*(12*mp.mpf('0.36')-1)/(5*(1+4*mp.mpf('0.36'))**2)/900), tol=1e-6)
```

## Check 11 — Eq. (H11): pole weight and dispersion of the large-momentum screening pole

**Where we are in the argument.** Inserting the Lorentzian into $1/(1+\lambda c)$ gives, at large $q$,
$W_\lambda\simeq\lambda v-\lambda\theta_\lambda(q)b_{\Lambda_\lambda(q)}$.
Thus the Lorentzian kernel weight in the full screened interaction is $\lambda\theta_\lambda$, while $\theta_\lambda$ is its dimensionless fraction.
The pole parameter is $\Lambda_\lambda(q)$, and the pole frequency $\omega_\lambda=q\Lambda_\lambda$ interpolates
between the plasmon frequency $\sqrt{\lambda q_{\rm TF}^{2}/3}$ at $q\ll q_*$ and the free-particle energy $q^2/2$ at $q\gg q_*$.  This dispersion curve is the
path along which the Lorentzian kernel is integrated in the next check.

**Claim.** With $c_0=(q_*/q)^4/(1+4u^2)$, $u=\nu/q^2$:
$$\frac1{1+\lambda c_0}=1-\frac{\lambda q_*^4/4}{\nu^2+\omega_\lambda^2},\quad\omega_\lambda^2=\frac{q^4+\lambda q_*^4}{4},\quad
\theta_\lambda=\frac{\lambda q_*^4}{q^4+\lambda q_*^4},\quad\Lambda_\lambda=\frac q2\sqrt{1+\lambda(q_*/q)^4},\quad\theta_\lambda\Lambda_\lambda^2q^2=\frac{\lambda q_*^4}{4}.$$
Limits: $\Lambda_\lambda\to q/2$, $\theta_\lambda\to0$ for $q\gg q_*$; $\omega_\lambda\to\sqrt{\lambda q_{\rm TF}^{2}/3}$, $\theta_\lambda\to1$ for $q\ll q_*$ (with $q_*^4=\tfrac43q_{\rm TF}^{2}$).

**How this notebook verifies it.** Pure algebra: `exact` on the rational identity that produces the single-pole form, on $\theta_\lambda\Lambda_\lambda^2$ (weight × strength, which
must equal the residue $\lambda q_*^4/4q^2$), on $\Lambda_\lambda^2=\omega_\lambda^2/q^2$, and `sp.limit` for the three limits. No departure from the manuscript.

**Reading the output.** Six `exact`.

**Hand-first.** That this is a statement about the imaginary axis in the joint large-$q$, low-density limit, not about a real-frequency RPA pole (end of Sec. H3); the error
term $\rho_\lambda$ and its integrated size (Sec. 5.3).

```python
lam, q, qs_, q_tf_sq = sp.symbols('lambda q q_* q_tf_sq', positive=True)
c0 = (qs_/q)**4/(1+4*u**2); om2 = (q**4 + lam*qs_**4)/4; nu = q**2*u
exact("Eq.(H11)  1/(1+lam c0) = 1 - (lam q*^4/4)/(nu^2+omega^2)", 1/(1+lam*c0) - (1 - (lam*qs_**4/4)/(nu**2 + om2)))
theta = lam*qs_**4/(q**4 + lam*qs_**4);  Lam_l = q/2*sp.sqrt(1 + lam*(qs_/q)**4)
exact("Eq.(H11)  theta Lambda^2 = lam q*^4/(4 q^2)  [pole weight x strength]", theta*Lam_l**2 - lam*qs_**4/(4*q**2))
exact("Eq.(H11)  Lambda_lambda^2 = omega_lambda^2/q^2", Lam_l**2 - om2/q**2)
exact("Eq.(H11)  omega -> sqrt(lam q_tf_sq/3) for q<<q* with q*^4 = 4 q_tf_sq/3",
      sp.limit(om2.subs(qs_, (R(4,3)*q_tf_sq)**R(1,4)), q, 0) - lam*q_tf_sq/3)
exact("Eq.(H11)  theta -> 1 for q -> 0", sp.limit(theta, q, 0) - 1)
exact("Eq.(H11)  Lambda/q -> 1/2 for q -> oo", sp.limit(Lam_l/q, q, oo) - R(1,2))
```

## Check 12 — Eqs. (H14)–(H17): the leading low-density coefficient $A_1$

**Where we are in the argument.** At low density the dominant transfers are $q\sim q_*$, where $s_{\Lambda,\lambda}=\Lambda_\lambda/q\ge\tfrac12$ and the
large-$q$ kernel $p_{\rm loc}(s_\Lambda)$ applies.  Integrating $\theta_\lambda p_{\rm loc}(s_{\Lambda,\lambda})/q^2$ along the dispersion curve, then over $2\lambda\,d\lambda$, gives
$e_{\rm kite}\simeq A_1/q_*$ with $A_1$ in closed form — the leading $r_s^{-1/4}$ law, identical to the matched-asymptotics result of the companion.

**Claim.**
$$A_1=-\frac{12}{\pi^2}\int_0^1\lambda\,d\lambda\int_0^\infty\frac{dx}{x^2}\frac{\lambda}{x^4+\lambda}\frac{s_{\Lambda,\lambda}(x)}{[1+2s_{\Lambda,\lambda}(x)]^2},\quad s_{\Lambda,\lambda}=\tfrac12\sqrt{1+\lambda x^{-4}}\quad\text{(Eq. (H15))},$$
$x=\lambda^{1/4}t$ factorizes $\int_0^1\lambda^{3/4}d\lambda=\tfrac47$; $s=\sqrt{1+t^{-4}}$ and $z=(s-1)/(s+1)$ give $A_1=-\frac{48}{7\pi^2}\frac1{4\sqrt2}B(\tfrac54,\tfrac12)$ (Eq. (H17)),
and the Beta function gives $A_1=-\frac{4}{7\sqrt2\pi^{3/2}}\frac{\Gamma(1/4)}{\Gamma(3/4)}=-0.21469376098909765\ldots$ (Eq. (H16)); $e_{\rm kite}\sim-0.2213782293\,r_s^{-1/4}$ Ry.

**Manuscript route.** The three substitutions above.

**How this notebook verifies it.**
1. **Departure from the manuscript — the unreduced 2-D integral is evaluated directly.** `inner(lv)` integrates over $x$ at fixed $\lambda$ (breakpoint at $x=\lambda^{1/4}$,
   the crossover of $s_{\Lambda,\lambda}$); the outer $\lambda$-integral is done by `mp.quad` as well. `A1_direct` agrees with the Gamma closed form to $10^{-31}$ (tolerance $10^{-15}$).
   This one number tests all three substitutions and the Beta evaluation at once.
2. The individual steps are then confirmed: $\int_0^1\lambda^{3/4}d\lambda=\tfrac47$ (`exact`); the $s$-integral $\int_1^\infty(s-1)^{1/4}(s+1)^{-7/4}ds$ against $B(\tfrac54,\tfrac12)/\sqrt2$
   (numerically, using the general formula $\int_1^\infty(s-1)^\alpha(s+1)^\beta ds=2^{\alpha+\beta+1}B(\alpha+1,-\alpha-\beta-1)$); the Beta→Gamma rewriting and the assembly (`exact`
   via `expand_func`/`gammasimp`); and the Beta integral itself in $z$ (numerically).
3. `aL`, `a_`: $\alpha_L=(4/9\pi)^{1/3}$, $a_{\rm TF}=4\alpha_L/\pi$; with $q_*=(4a_{\rm TF}r_s/3)^{1/4}$ the $r_s$ coefficient is $A_1(4a_{\rm TF}/3)^{-1/4}$.

**Reading the output.** The `direct 2-D quadrature` line is the substantive one.

**Hand-first.** The remainder bookkeeping $O(q_*^{-3/2})$ — the splits at $q_1=q_*^{1/2}$ and $q_0$, and the pole-approximation error $\mathcal P_\lambda$ (Sec. 5.3).

```python
def inner(lv):
    s_lambda_ = lambda xx: mp.sqrt(1 + lv*xx**-4)/2
    return mp.quad(lambda xx: lv/(xx**4+lv)*s_lambda_(xx)/(1+2*s_lambda_(xx))**2/xx**2, [0, lv**mp.mpf(0.25), mp.inf])
A1_direct = -12/mp.pi**2*mp.quad(lambda lv: lv*inner(lv), [0, 1])
A1_cl = -4/(7*mp.sqrt(2)*mp.pi**mp.mpf(1.5))*mp.gamma(mp.mpf(1)/4)/mp.gamma(mp.mpf(3)/4)
num("Eqs.(H15)/(H16)  direct 2-D quadrature of the dispersion integral = closed form", A1_direct, A1_cl, tol=1e-15)
num("Eq.(H16)  A1 = -0.21469376098909765", A1_cl, mp.mpf('-0.21469376098909765'), tol=1e-16)

exact("Eq.(H17)  int_0^1 lam^(3/4) dlam = 4/7", sp.integrate(lam**R(3,4), (lam, 0, 1)) - R(4,7))
# s-integral -> Beta via z=(s-1)/(s+1):  int_1^oo (s-1)^a (s+1)^b ds = 2^(a+b+1) B(a+1, -a-b-1); here a=1/4, b=-7/4
num("Eq.(H17)  int_1^oo (s-1)^(1/4)(s+1)^(-7/4) ds = B(5/4,1/2)/sqrt(2)",
    mp.quad(lambda ss: (ss-1)**mp.mpf(0.25)*(ss+1)**mp.mpf(-1.75), [1, 2, mp.inf]), mp.beta(mp.mpf(5)/4, mp.mpf(1)/2)/mp.sqrt(2))
Bv = sp.beta(R(5,4), R(1,2))
exact("Eq.(H16)  B(5/4,1/2) = (sqrt(pi)/3) Gamma(1/4)/Gamma(3/4)", sp.gammasimp(sp.expand_func(Bv)) - sp.sqrt(pi)/3*sp.gamma(R(1,4))/sp.gamma(R(3,4)))
A1_sym = -R(48,7)/pi**2/(4*sp.sqrt(2))*Bv
exact("Eq.(H16)  -(48/7pi^2)(1/4sqrt2) B(5/4,1/2) = -4 Gamma(1/4)/(7 sqrt2 pi^(3/2) Gamma(3/4))",
      sp.gammasimp(sp.expand_func(A1_sym)) - (-4/(7*sp.sqrt(2)*pi**R(3,2))*sp.gamma(R(1,4))/sp.gamma(R(3,4))))
# numeric check of the Beta integral over z itself
num("Eq.(H17)  int_0^1 z^(1/4)(1-z)^(-1/2) dz = B(5/4,1/2)",
    mp.quad(lambda zz: zz**mp.mpf(0.25)*(1-zz)**mp.mpf(-0.5), [0, 1]), mp.beta(mp.mpf(5)/4, mp.mpf(1)/2))
alphaL_mp = (4/(9*mp.pi))**(mp.mpf(1)/3); aTF_mp = 4*alphaL_mp/mp.pi
num("Eq.(H16)  e_kite ~ A1 (4 a_TF rs/3)^(-1/4): coefficient -0.2213782293", A1_cl/(4*aTF_mp/3)**mp.mpf(0.25), mp.mpf('-0.2213782293'), tol=1e-9)
```

## Check 13 — Eqs. (G1)–(G10): the high-density double logarithm

**Where we are in the argument.** High density.  In the window $q_{\rm TF}\ll q\ll1$ screening is weak ($c\ll1$), $\Upsilon(c)-1\simeq-\tfrac23c$, and
the shell logarithm of $X$ (Check 7) is multiplied by the $\log(1/q)$ of the momentum integral over the window: two logarithms, hence
$r_s\log^2r_s$.  The frequency integral $\mathcal A=4\pi\int h^2dz$ is the Gell-Mann–Brueckner integral.

**Claim.** $A(z)=2\pi\!\int d\Omega_p\,c_p^2\frac{c_p^2-z^2}{(c_p^2+z^2)^2}=8\pi^2\big[1-2z\arctan\tfrac1z+\tfrac{z^2}{1+z^2}\big]=8\pi^2[h+zh']$ (Eq. (G9));
$\mathcal A=\int_0^\infty\frac{dz}{\pi}h(z)A(z)=4\pi\int_0^\infty h^2dz=\frac{4\pi^2}{3}(1-\log2)=4.0380212$ (Eq. (G10));
$\int_{q_{\rm TF}}^1\frac{dq}{q}\log\frac1q=\tfrac18\log^2q_{\rm TF}^{2}$; $C_{\log^2}=-\frac{\pi A_0}{3}a\mathcal A=-\frac{\alpha_L(1-\log2)}{6\pi^3}=-0.000859$ Ry (Eq. (G1)).

**Manuscript route.** Angular integral over $c_p$; integration by parts turns $\int h(h+zh')$ into $\tfrac12\int h^2$; the Gell-Mann–Brueckner integral $\int_0^\infty h^2=\tfrac\pi3(1-\log2)$.

**How this notebook verifies it.**
1. `exact`: the $c$-integral $\int_0^1c^2\frac{c^2-z^2}{(c^2+z^2)^2}dc$ equals the bracket $1-2z\arctan\tfrac1z+\tfrac{z^2}{1+z^2}$, and the bracket equals $h+zh'$. The prefactor $8\pi^2$ of Eq. (G9) is $2\pi$ (explicit) $\times\,2\pi$ (azimuth of $d\Omega_p$) $\times\,2$ (the integrand is even in $c$, so $\int_{-1}^1=2\int_0^1$).
2. **The Gell-Mann–Brueckner integral is not taken from the literature but computed.** `calA_num` integrates $hA/\pi$ directly; `calA_h2` integrates $4\pi h^2$; both are compared
   with $\frac{4\pi^2}{3}(1-\log2)$ to 20 digits. Their mutual agreement also verifies the integration by parts.
3. `exact`: the window integral $\int_{q_{\rm TF}}^1\log(1/q)\,dq/q=\tfrac18\log^2q_{\rm TF}^{2}$ and the assembly $-\tfrac{\pi A_0}{3}a\mathcal A=-\tfrac{\alpha_L(1-\log2)}{6\pi^3}$.
4. `num`: the printed value $-0.000859$.

**Reading the output.** Four `exact`, four `num`.

**Hand-first.** The window argument itself — $\Upsilon(c)-1=-\tfrac23c$ for $q_{\rm TF}\ll q\ll1$, and the estimates showing $q>1$ and $q<q_{\rm TF}$ are subleading (Sec. 5.2) — and the coupling
weight $\lambda^2d\lambda$ of the first screening correction.

```python
z_ = sp.symbols('z', positive=True); cc = sp.symbols('c', real=True)
h_s = 1 - z_*sp.atan(1/z_);  br = 1 - 2*z_*sp.atan(1/z_) + z_**2/(1+z_**2)
exact("Eq.(G9)  int_0^1 c^2 (c^2-z^2)/(c^2+z^2)^2 dc = bracket",
      sp.integrate(cc**2*(cc**2-z_**2)/(cc**2+z_**2)**2, (cc, 0, 1)) - br)
exact("Eq.(G9)  bracket = h + z h'", h_s + z_*sp.diff(h_s, z_) - br)
Abar = lambda zz: 1 - 2*zz*mp.atan(1/zz) + zz**2/(1+zz**2)
calA_num = mp.quad(lambda zz: h(zz)*8*mp.pi**2*Abar(zz), [0, 1, mp.inf])/mp.pi
calA_h2  = 4*mp.pi*mp.quad(lambda zz: h(zz)**2, [0, 1, mp.inf])
calA_cl  = 4*mp.pi**2/3*(1-mp.log(2))
num("Eq.(G10)  int h A dz/pi = 4pi^2/3 (1-log2)", calA_num, calA_cl, tol=1e-20)
num("Eq.(G10)  4 pi int h^2 dz = 4pi^2/3 (1-log2)  [Gell-Mann-Brueckner]", calA_h2, calA_cl, tol=1e-20)
num("Eq.(G10)  calA = 4.0380212", calA_cl, mp.mpf('4.0380212'), tol=1e-7)
exact("Eq.(G1)  int_{sqrt q_tf_sq}^1 log(1/q)/q dq = log(q_tf_sq)^2/8", sp.integrate(sp.log(1/q)/q, (q, sp.sqrt(q_tf_sq), 1)) - sp.log(q_tf_sq)**2/8)
alphaL = (4/(9*pi))**R(1,3); aTF = 4*alphaL/pi
calA_s = 4*pi**2/3*(1-sp.log(2))
exact("Eq.(G1)  -(pi A0/3) a_TF calA = -alpha_L (1-log2)/(6 pi^3)", -pi*A0/3*aTF*calA_s + alphaL*(1-sp.log(2))/(6*pi**3))
num("Eq.(G1)  C_log2 = -0.000859 Ry  (tolerance = half a unit in the last printed digit)", -alphaL_mp*(1-mp.log(2))/(6*mp.pi**3), mp.mpf('-0.000859'), tol=5e-7)
```

## Check 14 — Exchange vertex and self-energy response normalization

The two first-order corrections are expressed in the same response units. With $C=\pi^2q_{\rm TF}^2$, $\Sigma_x(k)=-q_{\rm TF}^2f_x(k)/2$, and $v(q)=q^{-2}$,
$$\Pi_{\rm vtx}^{(1)}=-\frac{q_{\rm TF}^4}{32\pi^2}X,\qquad
\Pi_{\rm SE}^{(1)}=2C\int\frac{d^3p}{(2\pi)^3}\Delta_{\bf p}\operatorname{Re}\frac{\Sigma_x(|\mathbf p+\mathbf q|)-\Sigma_x(p)}{(i\nu-\delta_{\bf p})^2}.$$
The integral $\int_{-1}^1 c^2(c^2-z^2)/(c^2+z^2)^2\,dc=2[h(z)+zh'(z)]$ supplies the common shell function. The Fock endpoint expansion is checked from both sides of the Fermi surface. The normalization check follows the two factors of $C$: one from the interaction in the vertex loop, the other from the common response unit.

The relative diagrammatic sign and the common selected-line energy weight remain part of the analytic argument. The calculation checks their resulting equal-and-opposite logarithmic coefficients without changing the diagram family.


```python
exact("After Eq.(G18)  int_{-1}^{1} c^2(c^2-z^2)/(c^2+z^2)^2 dc = 2[h + z h']",
      sp.integrate(cc**2*(cc**2-z_**2)/(cc**2+z_**2)**2, (cc, -1, 1)) - 2*(h_s + z_*sp.diff(h_s, z_)))
uu = sp.symbols('u', positive=True)
f_x_plus  = R(1,2) + (-uu*(2+uu))*sp.log((2+uu)/uu)/(4*(1+uu))     # k = 1+u
f_x_minus = R(1,2) + ( uu*(2-uu))*sp.log((2-uu)/uu)/(4*(1-uu))     # k = 1-u
lead_p = sp.series(f_x_plus - R(1,2), uu, 0, 2).removeO().coeff(uu, 1)
lead_m = sp.series(f_x_minus - R(1,2), uu, 0, 2).removeO().coeff(uu, 1)
exact("Eq.(G18)  F(1+u)-F(1) = -(u/2) log(2/u) + ...   (u>0)", sp.expand(lead_p + sp.log(2/uu)/2))
exact("Eq.(G18)  F(1-u)-F(1) = +(u/2) log(2/u) + ...   (i.e. -(u'/2)log(2/|u'|) with u'=-u)", sp.expand(lead_m - sp.log(2/uu)/2))
exact("Eq.(A16)  f_x(0) = 1  =>  Sigma_x(0) = -q_tf_sq/2", sp.limit(R(1,2) + (1-k**2)*sp.log((1+k)/(1-k))/(4*k), k, 0) - 1)
# The common response unit supplies a second C = pi^2 q_tf_sq.
exact("Eq.(A17) response normalization -2 (2pi)^-6 C^2 = -q_tf_sq^2/(32 pi^2)",
      -2*(2*pi)**(-6)*(pi**2*q_tf_sq)**2 + q_tf_sq**2/(32*pi**2))
```

## Check 15 — Eq. (S14): spin dependence of the double logarithm

**Where we are in the argument.** Repeating the window analysis with spin-resolved Fermi spheres gives the $\zeta$ dependence of $C_{\log^2}$
in closed form through one function $\mathcal F_{\rm s}(a)$; $\zeta=1$ must reproduce the exact rescaling $e_{\rm kite}(r_s,1)=e_{\rm kite}(2^{-4/3}r_s,0)$.

**Claim.** $\int_0^\infty h(\theta z)\bar A(z)\,dz=\frac{\pi}{6a}\mathcal F_{\rm s}(a)$, $a=1/\theta$, with $\mathcal F_{\rm s}(a)=2a^2-a+(1-2a^3)\log(1+a)+2a^3\log a$; $\mathcal F_{\rm s}(1)=1-\log2$
(recovers $C_{\log^2}$ at $\zeta=0$), $\mathcal F_{\rm s}(0)=0$; $C_{\log^2}(1)=2^{-4/3}C_{\log^2}(0)$; the ratio table $0.990,0.956,0.894,0.781,0.683,0.397$ at $\zeta=0.2,\dots,1$.

**Manuscript route.** Write $h(z/a)=a^{-1}\int_0^at^2dt/(t^2+z^2)$ and integrate in $z$ first.

**How this notebook verifies it.**
1. **Departure from the manuscript — the $z$-integral is done by quadrature** at $a=0.6,1,1.7$ and compared with $\frac{\pi}{6a}\mathcal F_{\rm s}(a)$ to 20 digits.
2. `exact`: $\mathcal F_{\rm s}(1)$ and the limit $\mathcal F_{\rm s}(0)=0$ (which needs $a^3\log a\to0$).
3. `Cratio(zeta)` implements Eq. (S14) literally, $\sum_{\sigma\sigma'}x_\sigma^2\mathcal F_{\rm s}(x_{\sigma'}/x_\sigma)$ with $x_\sigma=(1\pm\zeta)^{1/3}$, normalized by its $\zeta=0$ value
   $4(1-\log2)$; terms with a vanishing $x$ are skipped, as the manuscript prescribes. The $\zeta=1$ ratio is compared with $2^{-4/3}$ (the exact rescaling identity) and the six table
   entries are compared at three significant figures (`check`).

**Reading the output.** Three `num`, two `exact`, one `num` (ratio), one table `check`.

**Hand-first.** The spin-resolved loop weights $x_\sigma^3/2$ and the two-component response $H_\zeta$ (Supplementary Material S2 and Supplementary Material S1).

```python
F_s = lambda av: 2*av**2 - av + (1-2*av**3)*mp.log(1+av) + 2*av**3*mp.log(av)
for av in [mp.mpf('0.6'), mp.mpf(1), mp.mpf('1.7')]:
    th = 1/av
    num("Eq.(S14)  int h(theta z) Abar(z) dz = pi/(6a) F_s(a), a=%s" % av,
        mp.quad(lambda zz: h(th*zz)*Abar(zz), [0, 1, mp.inf]), mp.pi/(6*av)*F_s(av), tol=1e-20)
aS = sp.symbols('a', positive=True)
F_s_sym = 2*aS**2 - aS + (1-2*aS**3)*sp.log(1+aS) + 2*aS**3*sp.log(aS)
exact("Eq.(S14)  F_s(1) = 1 - log 2", F_s_sym.subs(aS, 1) - (1-sp.log(2)))
exact("Eq.(S14)  F_s(0) = 0", sp.limit(F_s_sym, aS, 0))
def Cratio(zeta):
    xs = [(1+zeta)**(mp.mpf(1)/3), (1-zeta)**(mp.mpf(1)/3)]
    tot = mp.mpf(0)
    for x1 in xs:
        for x2 in xs:
            if x1 > 0 and x2 > 0: tot += x1**2*F_s(x2/x1)
    return tot/(4*(1-mp.log(2)))
num("Eq.(S14)  C_log2(1)/C_log2(0) = 2^(-4/3)", Cratio(mp.mpf(1)), mp.mpf(2)**(-mp.mpf(4)/3), tol=1e-20)
table = [mp.nstr(Cratio(mp.mpf(zv)), 3) for zv in ['0.2', '0.4', '0.6', '0.8', '0.9', '1']]
print("ratio table:", table)
check("Eq.(S14)  table 0.990,0.956,0.894,0.781,0.683,0.397 reproduced", table == ['0.99', '0.956', '0.894', '0.781', '0.683', '0.397'])
```

## Check 16 — Eqs. (I2)–(I4), (I19)–(I21): the static kernel coefficients $c_0$, $c_1=0$, $c_2$ and $C_\infty$

**Where we are in the argument.** Section 5.4 and Appendix I treat the *same diagram with a static screened line* as a reference.  With the
static kernel $\mathcal W(a,b)$ only the cross-sphere sectors survive, and the radial kernel $X_c(q)$ can be written as an integral over a fixed
rectangle in which $q$ enters analytically; its Taylor coefficients are the small-$q$ coefficients $c_0$ and $c_2$ (with $c_1=0$), and its large-$q$
tail is $C_\infty/q^4$.

**Claim (App. I).** On the fixed square, with $L_t=\tfrac{qt}2+\sqrt{1-q^2(1-t^2)/4}$ and $F=L_t^2L_s^2uv(L_tu+L_sv)^{-2}H^{-1/2}$, the average over $t,s\in[-1,1]$ is
$$\langle\mathsf F\rangle=\frac{uv}{(u+v)^2}+q^2\,\frac{u^2v^2(3uv-3u-3v-5)}{12(u+v)^4}+O(q^4)\qquad\text{(I19)},$$
so $c_0=\frac\pi4\int_0^1\frac{r\,dr}{(1+r)^2}=\frac\pi8(2\log2-1)$ (D6, Eq. (I3)), $c_1=0$, $c_2=-\frac{\pi}{96}\int_0^1\frac{r^2(r+14)}{(1+r)^4}dr=\frac{\pi}{1152}-\frac{\pi\log2}{96}$ (D7, Eq. (I7));
for $q>2$, $C_\infty=\frac{1}{16\pi}\cdot2\big(\tfrac{4\pi}{3}\big)^2=\frac{2\pi}{9}$ (Eq. (I4)).

**Manuscript route.** Expand $L_t$, $L_s$ and $H^{-1/2}$ to $O(q^2)$ and use $\langle t\rangle=\langle s\rangle=\langle ts\rangle=0$, $\langle t^2\rangle=\langle s^2\rangle=\tfrac13$.

**How this notebook verifies it.**
1. **(I19) is verified by brute force, not by the manuscript's moment bookkeeping.** `avgF_series(u, v)` builds $\mathsf F$ exactly as defined in App. I (with $x=L_tu$, $y=L_sv$ and the full
   $H$), expands it in $q$ to second order with `sp.series`, and integrates the resulting polynomial in $t,s$ over $[-1,1]^2$ divided by 4 (the uniform average). The result is compared with the
   right-hand side of (I19) by `exact`.

   *Why two rational points instead of symbolic $(u,v)$:* the fully symbolic expansion takes about 2.5 minutes; with $(u,v)=(\tfrac13,\tfrac25)$ and $(\tfrac7{10},\tfrac14)$ substituted first it takes
   about 15 s per point. Both the absence of a $q^1$ term and the stated $q^2$ coefficient are tested at each point. Since (I19) is a rational function of $(u,v)$ with a specific structure, agreement at
   these two points test specializations. The generic symbolic coefficient is checked in [Check 36](#check-36).
2. `exact`: the two $r$-integrals of (D6), (D7) and their closed forms; `num`: the printed values of $c_0$, $c_2$.
3. `exact`: $C_\infty$ from the two cross-sphere sectors ($2\times V_B^2$) and the normalization $1/(16\pi)$ of Eq. (I2).

**Reading the output.** Two `exact` (I19), two `exact` + two `num` ($c_0,c_2$), one `exact` ($C_\infty$). This cell is the slowest of the notebook (~30 s).

**Hand-first.** The crescent geometry, the inclusion of the polar caps ($0\le x\le L_t$), and the even analytic continuation of $X_c/q$ (App. I, before (I19)).

```python
qq, t, s_ = sp.symbols('q t s', real=True)
def avg_sfF_series(uv, vv):
    Lt = qq*t/2 + sp.sqrt(1 - qq**2*(1-t**2)/4);  Ls = qq*s_/2 + sp.sqrt(1 - qq**2*(1-s_**2)/4)
    xx, yy = Lt*uv, Ls*vv
    H = 1 + qq*xx*yy*(t+s_)/(xx+yy) + qq**2/4*((xx*t-yy*s_)**2/(xx+yy)**2 - 1)
    sfF = Lt**2*Ls**2*uv*vv/(xx+yy)**2/sp.sqrt(H)
    ser = sp.series(sfF, qq, 0, 3).removeO()
    return sp.nsimplify(sp.expand(sp.integrate(sp.integrate(sp.expand(ser), (t, -1, 1)), (s_, -1, 1))/4))
for (uv, vv) in [(R(1,3), R(2,5)), (R(7,10), R(1,4))]:
    target = uv*vv/(uv+vv)**2 + qq**2*uv**2*vv**2*(3*uv*vv - 3*uv - 3*vv - 5)/(12*(uv+vv)**4)
    exact("Eq.(I19) <sfF> to O(q^2) at (u,v)=(%s,%s): no q^1 term, q^2 coefficient as stated" % (uv, vv), avg_sfF_series(uv, vv) - target)
r = sp.symbols('r', positive=True)
c0_s = pi/4*sp.integrate(r/(1+r)**2, (r, 0, 1))
c2_s = -pi/96*sp.integrate(r**2*(r+14)/(1+r)**4, (r, 0, 1))
exact("Eq.(I3)/(I20)  c0 = pi/8 (2 log2 - 1)", c0_s - pi/8*(2*sp.log(2)-1))
exact("Eq.(I7)/(I21)  c2 = pi/1152 - pi log2/96", c2_s - (pi/1152 - pi*sp.log(2)/96))
num("Eq.(I3)  c0 = 0.151697", sp.N(c0_s, 20), mp.mpf('0.151697'), tol=5e-6)
num("Eq.(I7)  c2 = -0.0199561", sp.N(c2_s, 20), mp.mpf('-0.0199561'), tol=5e-6)
exact("Eq.(I4)  C_inf = (1/16pi) 2 (4pi/3)^2 = 2pi/9", 1/(16*pi)*2*(4*pi/3)**2 - 2*pi/9)
```

**Where we are in the argument.** Feeding $X_c=c_0q+c_2q^3+\dots$ into the static Lindhard line gives the high-density expansion of the static

$$D_0^{\rm scr}=D_0+\frac{c_0}{2}q_{\rm TF}^{2}\log q_{\rm TF}^{2}+B_1q_{\rm TF}^{2}+\Big(\frac{c_0}{12}-\frac{c_2}{2}\Big)q_{\rm TF}^{4}\log q_{\rm TF}^{2}+O(q_{\rm TF}^{4})\qquad\text{(Eq. (I8))}.$$

**Reading the output.** Three `exact`, one `num`, two `exact`.

**Hand-first.** Why the regular Taylor series of $\mathcal X(w)$ excludes half-integer powers at every order (last paragraph of App. I).

```python
exact("Eq.(I6)  (2 alpha_L/pi) c0 = alpha_L (2 log2 - 1)/4", 2*alphaL/pi*c0_s - alphaL*(2*sp.log(2)-1)/4)
num("Eq.(I6)  C_log^stat = 6 c_log/pi^3 = 0.00974 Ry  (tolerance = half a unit in the last printed digit)", sp.N(6/pi**3*alphaL*(2*sp.log(2)-1)/4, 20), mp.mpf('0.00974'), tol=5e-6)
W, c0v, c1v = sp.symbols('W c_0 c_1', positive=True)
E = q_tf_sq/2*sp.integrate((c0v + c1v*w)/(w + q_tf_sq), (w, 0, W))            # = D0 - D0^scr (small-q part)
Es = sp.expand(sp.series(E, q_tf_sq, 0, 3).removeO())
coef_tlog  = Es.coeff(sp.log(q_tf_sq)).coeff(q_tf_sq, 1)
coef_t2log = Es.coeff(sp.log(q_tf_sq)).coeff(q_tf_sq, 2)
exact("Eq.(I8)  D0^scr - D0 has q_tf_sq log q_tf_sq coefficient  +c0/2   (from -E)", -coef_tlog - c0v/2)
exact("Eq.(I8)  D0^scr - D0 has q_tf_sq^2 log q_tf_sq coefficient c0/12 - c2/2 with c1 = c2 - c0/6",
      (-coef_t2log).subs(c1v, sp.Symbol('c_2') - c0v/6) - (c0v/12 - sp.Symbol('c_2')/2))
exact("Eq.(I8)  f_L(q/2) = 1 - q^2/12 + O(q^4)",
      sp.series(R(1,2) + (1-(q/2)**2)*sp.log((1+q/2)/(1-q/2))/(4*(q/2)), q, 0, 4).removeO() - (1 - q**2/12))
```

## Check 19 — Eq. (S19) and auxiliary numerical examples: Yukawa pole match of the static Lindhard line

**Where we are in the argument.** A practical remark of Sec. 5.4: the static Lindhard line can be approximated by a single Yukawa pole
$Z/(q^2+\kappa_*^2)$ obtained by continuing $W_{\rm stat}$ to $q^2=-\kappa_*^2$; this check quantifies the pole position and residue at metallic densities.
What fraction of $\mathcal K[W_{\rm stat}]$ that pole captures in energy is outside the scope of the current manuscript.

**Claim.** Continuing $W_{\rm stat}=[q^2+q_{\rm TF}^2f_L(q/2)]^{-1}$ to $q^2=-\kappa_*^2$ with $f_L(iy)=\tfrac12+(1+y^2)\arctan y/(2y)$ and $Z=[1+q_{\rm TF}^2\,df_L/d(q^2)]^{-1}$ gives
$(\kappa_*,Z)=(0.837,1.055),(1.217,1.108),(1.814,1.209)$ at $r_s=1,2,4$; the residual has the tail $\Delta=(1-Z)q^{-2}+Z\kappa_*^2q^{-4}+O(q^{-6})$.

**How this notebook verifies it.**
1. *Numerical solution of the pole condition (the manuscript gives only the results).* `mp.findroot` solves $\kappa^2=q_{\rm TF}^{2} f_L(i\kappa/2)$ with $q_{\rm TF}^{2}=a_{\rm TF}r_s$; the residue factor is
   $Z=[1+q_{\rm TF}^{2}\,g'(s)]^{-1}$ with $g(s)=f_L(\sqrt{-s}/2)$ evaluated at $s=-\kappa_*^2$, the derivative taken numerically by `mp.diff` (the derivative with respect to $q^2$ is what the definition of
   $\mathcal Z_*$ requires). Tolerance $6\times10^{-4}$ matches the three printed digits.
2. `exact`: the large-$q$ expansion of $W_{\rm stat}-\mathcal Z_*/(q^2+\kappa_*^2)$ using the $q>2$ branch of $f_L(q/2)$ (where $\log\frac{q/2+1}{q/2-1}$ needs no absolute value), keeping $Z$ and $\kappa_*$ symbolic.

**Reading the output.** Six `num` (table), one `exact` (tail), plus the printed table.

**Hand-first.** $\mathcal K[\Delta]/\mathcal K[W_{\rm stat}]$ percentages (require the full kernel) and the Kohn nonanalyticity at $q=2$.

```python
fLi = lambda y: mp.mpf(1)/2 + (1+y*y)*mp.atan(y)/(2*y)
table = {}
for rs_v, (k_ref, Z_ref) in {1: ('0.837', '1.055'), 2: ('1.217', '1.108'), 4: ('1.814', '1.209')}.items():
    tv = aTF_mp*rs_v
    kap = mp.findroot(lambda kk: kk*kk - tv*fLi(kk/2), 1.0)
    Zv  = 1/(1 + tv*mp.diff(lambda s_: fLi(mp.sqrt(-s_)/2), -kap**2))
    num("Eq.(S19) kappa_* at r_s=%d = %s" % (rs_v, k_ref), kap, mp.mpf(k_ref), tol=6e-4)
    num("Eq.(S19) Z_* at r_s=%d = %s" % (rs_v, Z_ref), Zv, mp.mpf(Z_ref), tol=6e-4)
    table[rs_v] = (mp.nstr(kap, 4), mp.nstr(Zv, 4))
print("kappa_*, Z:", table)
Zs, ks = sp.symbols('Z kappa_*', positive=True)
Dstat = 1/(q**2 + q_tf_sq*(R(1,2) + (1-(q/2)**2)*sp.log((q/2+1)/(q/2-1))/(4*(q/2))))    # q>2 branch of f_L(q/2)
Delta = sp.series(Dstat - Zs/(q**2+ks**2), q, oo, 6).removeO()
exact("Eq.(S19) D_reg = (1-Z_*) q^-2 + Z_* kappa_*^2 q^-4 + O(q^-6)", sp.simplify(Delta - ((1-Zs)/q**2 + Zs*ks**2/q**4)))
```

## Check 20 — (S5)–(S11): correspondence with the BRM representation

**Where we are in the argument.** Supplementary Material S1 shows that the functional of Eq. (A15) is identical to the nine-dimensional integral that BRM
evaluate numerically (their Eqs. (24)–(28)): the four occupation sectors reduce to two cap integrals, the frequency integral is done by a tangent
substitution, and the spin-frame screening function $f^\sigma_\lambda$ is recovered.  This is what makes the direct numerical comparisons of Sec. 5.4
meaningful.

**Claim.** (S6): $\Re[(i\nu-a)(i\nu-b)]^{-1}=\frac{1}{a-b}\big[\frac{-a}{\nu^2+a^2}+\frac{b}{\nu^2+b^2}\big]$, $\Re[(i\nu-a)(i\nu+b)]^{-1}=-\frac1{a+b}\big[\frac{a}{\nu^2+a^2}+\frac{b}{\nu^2+b^2}\big]$;
(S7): $\int_0^\infty d\nu\,\frac{a\,h(\nu)}{\nu^2+a^2}=\int_0^{\pi/2}du\,h(a\tan u)$ for even $h$;
(S10), after dividing its common leading coupling factor $\lambda$: $x_\sigma^3q^{-2}(\epsilon_\lambda^{-1}-1)=-\frac{\lambda\alpha r_s}{\pi q^4}f^\sigma_\lambda$ with $f^\sigma_\lambda=\frac{x_\sigma^3S_\sigma}{x_\sigma^2+\lambda\alpha r_sS_\sigma/(\pi q^2)}$ and $\epsilon_\lambda=1+\lambda\alpha r_sS_\sigma/(\pi x_\sigma^2q^2)$;
(S11): $2A_0\alpha r_s/\pi^2=3\alpha r_s/(16\pi^7)$.

**How this notebook verifies it.**
1. `sp.re(1/((I*nu-a)*(I*nu-b)))` takes the real part symbolically (with $\nu,a,b$ declared positive so that SymPy can separate real and imaginary parts); `exact` compares with the partial fractions.
2. (S7) is an identity for all even $h$; it is tested numerically for two representative even functions, $e^{-\nu^2}$ and $(1+\nu^4)^{-1}$, at $a=1.3$. (The substitution $\nu=a\tan u$ is elementary; the numerical
   test simply confirms the normalization.)
3. (S10) and (S11) are rational identities (`exact`), with $S_\sigma$ kept symbolic.

**Reading the output.** Two `exact`, two `num`, two `exact`.

**Hand-first.** The pairwise reduction of the four sign sectors to $C(\bm q)^2$ (S5) and the identification with BRM's Heaviside products — the geometric content of Supplementary Material S1.

```python
nu_, a_s2, b_s2 = sp.symbols('nu a b', positive=True)
same = sp.re(sp.expand(1/((sp.I*nu_-a_s2)*(sp.I*nu_-b_s2))))
exact("Eq.(S6) Re[(i nu - a)(i nu - b)]^-1 = [-a/(nu^2+a^2) + b/(nu^2+b^2)]/(a-b)",
      sp.simplify(sp.re(1/((sp.I*nu_-a_s2)*(sp.I*nu_-b_s2))) - (-a_s2/(nu_**2+a_s2**2) + b_s2/(nu_**2+b_s2**2))/(a_s2-b_s2)))
exact("Eq.(S6) Re[(i nu - a)(i nu + b)]^-1 = -[a/(nu^2+a^2) + b/(nu^2+b^2)]/(a+b)",
      sp.simplify(sp.re(1/((sp.I*nu_-a_s2)*(sp.I*nu_+b_s2))) + (a_s2/(nu_**2+a_s2**2) + b_s2/(nu_**2+b_s2**2))/(a_s2+b_s2)))
# (A3) tangent substitution, tested on two even functions
for hname, hf in [("exp(-nu^2)", lambda t_: mp.exp(-t_**2)), ("1/(1+nu^4)", lambda t_: 1/(1+t_**4))]:
    av = mp.mpf('1.3')
    num("Eq.(S8) int a h/(nu^2+a^2) dnu = int_0^{pi/2} h(a tan u) du, h=%s" % hname,
        mp.quad(lambda t_: av*hf(t_)/(t_**2+av**2), [0, 1, mp.inf]), mp.quad(lambda u_: hf(av*mp.tan(u_)), [0, mp.pi/2]))
xs_, S_, al_r = sp.symbols('x_sigma S alpha_rs', positive=True)
eps_l = 1 + lam*al_r*S_/(pi*xs_**2*q**2)
f_sig = xs_**3*S_/(xs_**2 + lam*al_r*S_/(pi*q**2))
exact("Eq.(S10) x^3 q^-2 (1/eps - 1) = -(lam alpha r_s/(pi q^4)) f^sigma", xs_**3/q**2*(1/eps_l - 1) + lam*al_r/(pi*q**4)*f_sig)
exact("Eq.(S11) 2 A0 alpha r_s/pi^2 = 3 alpha r_s/(16 pi^7)", 2*A0*al_r/pi**2 - 3*al_r/(16*pi**7))
```

## Check 21 — Direct profile quadrature (optional)

Evaluate Eq. (D3) after $t=uy$ and $v=e^{-r}$. Gauss–Laguerre integration of $r$ absorbs the logarithmic $v=0$ endpoint into the weight $e^{-r}$. The $t$ integral is split into intervals of length $2\pi$; its lower logarithmic endpoint is integrated adaptively.

The leading nonoscillatory tail beyond $T$ is
$2yT^{-2}\{[\sin y/y][\log(y/T)-1/2]+\operatorname{Si}(y)/y\}$, from $L(t/y)\simeq2y/t$ and the nonoscillatory Bessel-product part $2\cos(vy)/t^2$. Compare $(T,n)=(512,32)$ and $(1024,48)$ to test truncation and quadrature stability. This replaces the earlier truncated $u$ quadrature that emitted a convergence warning. Compare the converged values with $y[\log y/2+\gamma/2-3/8]$; this second check tests a finite-$y$ asymptotic approximation and has a looser tolerance.

Set `RUN_SLOW` in the setup cell. The distributed executed notebook includes this check. It does not integrate the profile over all $y$, or prove either endpoint remainder uniformly.


```python
if RUN_SLOW:
    import numpy as np
    from scipy import integrate,special
    from functools import lru_cache
    @lru_cache(None)
    def lag_profile(n):return special.roots_laguerre(n)
    def Phi_rescaled(y,T=512,n=32):
        r,w=lag_profile(n);v=np.exp(-r)
        def f(t):
            prod=special.spherical_jn(1,(t+v*y)/2)*special.spherical_jn(1,(t-v*y)/2)
            return np.log1p(2*y/(t-y))*np.sum(w*(np.log(y/t)+r)*prod)
        edges=[y, max(1.,2*y), 4.]
        edges+=list(np.arange(4+2*np.pi,T,2*np.pi))+[float(T)]
        vals=[integrate.quad(f,a,b,epsabs=1e-12,epsrel=1e-10) for a,b in zip(edges[:-1],edges[1:])]
        tail=2*y/T**2*((np.sin(y)/y)*(np.log(y/T)-.5)+special.sici(y)[0]/y)
        return sum(a for a,b in vals)+tail,sum(b for a,b in vals)
    for yv in [0.02,0.05]:
        coarse,err1=Phi_rescaled(yv,512,32)
        fine,err2=Phi_rescaled(yv,1024,48)
        ref=yv*(np.log(yv)/2+float(mp.euler)/2-3/8)
        print("Phi",yv,"=",fine,"cutoff/node difference",abs(fine-coarse),"small-y residual",fine-ref)
        num("Eq.(D3): rescaled Phi cutoff/node convergence y=%s"%yv,coarse,fine,tol=5e-9)
        num("Eq.(F2): direct Phi vs small-y asymptote y=%s"%yv,fine,ref,tol=2e-3)
else:
    SKIPPED.append("Check 21: direct profile quadrature")
    print("SKIP Check 21 (set RUN_SLOW = True)")
```

## Check 22 — Spectral representation, continuum and plasmon

For positive real frequency define $B_{W,\lambda}(q,\omega)=-\operatorname{Im}W_\lambda^R(q,\omega)/\pi$. The screened line obeys
$$W_\lambda(q,i\nu)=\lambda v(q)-\int_0^\infty\frac{2\omega B_{W,\lambda}(q,\omega)}{\omega^2+\nu^2}\,d\omega.$$
For the unchanged Lorentzian kernel $b_\Lambda=q^{-2}\Lambda^2/[\Lambda^2+(\nu/q)^2]$, the change $\omega=q\Lambda$ gives
$$W_\lambda=\lambda v-\int_0^\infty a_\lambda(q,\Lambda)b_\Lambda\,d\Lambda,\qquad
a_\lambda=\frac{2q^2}{\Lambda}B_{W,\lambda}(q,q\Lambda).$$

The numerical quadrature uses the weighted loss quantity $\varrho_\lambda=(2\omega/\pi)\operatorname{Im}[-\epsilon_\lambda^{-1}]$ as an intermediate. It is connected to the physical spectrum by $B_{W,\lambda}=\lambda\varrho_\lambda/(2q^2\omega)$, hence $a_\lambda=\lambda\varrho_\lambda(q,q\Lambda)/(q\Lambda^2)$.

The retarded Lindhard shape is obtained directly by continuing its defining logarithmic integral. At $q=1.5$, $q_{\rm TF}^2=0.3$, $\lambda=0.8$, the continuum-only quadrature is compared with $1-\epsilon_\lambda^{-1}(q,i\nu)$ at $\nu=0.7$. The cell then repeats the test at $q=0.4$, $q_{\rm TF}^2=3$, where it explicitly finds the undamped plasmon and adds its point mass. The pole contribution to the inverse-dielectric integral is $2\omega_p/[\epsilon'(\omega_p)(\nu^2+\omega_p^2)]$.

The final symbolic checks include the coupling factor and the spectral change-of-variable Jacobian. A point mass is a spectral measure, not a smooth sampled density. Momentum-dependent weights remain inside $\mathcal K$.


```python
def lindhard_shape_im(qv_, nu_):
    f = lambda kk: kk*mp.log(((qv_*kk+qv_*qv_/2)**2 + nu_**2)/((qv_*kk-qv_*qv_/2)**2 + nu_**2))
    return mp.quad(f, [0, qv_/2, 1])/(2*qv_) if qv_ < 2 else mp.quad(f, [0, 1])/(2*qv_)
def lindhard_shape_retarded(qv_, wv_, eta=mp.mpf('1e-12')):          # retarded continuation nu -> -i(w + i eta)
    zc = wv_ + 1j*eta
    f = lambda kk: kk*(mp.log((qv_*kk+qv_*qv_/2)**2 - zc*zc) - mp.log((qv_*kk-qv_*qv_/2)**2 - zc*zc))
    pts = sorted({mp.mpf(0), mp.mpf(1)} | {p_ for p_ in (abs((wv_-qv_*qv_/2)/qv_), abs((wv_+qv_*qv_/2)/qv_)) if p_ < 1})
    return mp.quad(f, pts)/(2*qv_)
mp.mp.dps = 15          # nested quadratures: 15 digits keep this cell at ~45 s; the test tolerance is 1e-5
qv_, tv_, lv_ = mp.mpf('1.5'), mp.mpf('0.3'), mp.mpf('0.8')
num("Eq.(C2) consistency: lindhard_shape_retarded(q,0) = lindhard_shape_im(q,0)", lindhard_shape_retarded(qv_, mp.mpf(0)).real, lindhard_shape_im(qv_, mp.mpf('1e-15')), tol=1e-9)
eps_re = lambda w_: 1 + lv_*tv_*lindhard_shape_retarded(qv_, w_)/qv_**2
top = qv_**2/2 + qv_                                        # upper edge of the particle-hole continuum for q<2
grid = [mp.mpf(w_) for w_ in ('0.1', '0.5', '1', '2', '2.6', '3', '5')]
vals = [eps_re(w_) for w_ in grid]
print("omega:      ", [mp.nstr(w_, 3) for w_ in grid])
print("Re eps:     ", [mp.nstr(e_.real, 6) for e_ in vals])
print("Im eps:     ", [mp.nstr(e_.imag, 3) for e_ in vals])
check("Eq.(C2) Im eps(q,omega+i0) >= 0 on the grid (passivity => rho >= 0)", all(e_.imag >= -1e-12 for e_ in vals))
check("Eq.(C2) Re eps > 0 outside the continuum on the grid (no plasmon point mass here)", all(e_.real > 0 for w_, e_ in zip(grid, vals) if w_ > top))
rho = lambda w_: 2*w_/mp.pi*(-1/eps_re(w_)).imag
nu_ = mp.mpf('0.7')
lhs = 1 - 1/(1 + lv_*tv_*lindhard_shape_im(qv_, nu_)/qv_**2)
rhs = mp.quad(lambda w_: rho(w_)/(nu_**2 + w_**2), [0, mp.mpf('0.5'), 1, 2, top])
num("Eq.(C2) 1 - 1/eps(q,i nu) = int_0^oo rho(q,omega)/(nu^2+omega^2) d omega  (q=1.5, q_tf_sq=0.3, lam=0.8, nu=0.7)", rhs, lhs, tol=1e-5)
check("Eq.(C2) rho(q,omega) >= 0 at sampled omega", all(rho(w_) >= 0 for w_ in grid if w_ < top))

# --- the same identity in the case that actually matters: a plasmon pole outside the continuum
q2_, t2_, l2_ = mp.mpf('0.4'), mp.mpf(3), mp.mpf(1)          # q small, q_tf_sq = 3 (r_s about 4.5): epsilon has a zero
eps2 = lambda w_: 1 + l2_*t2_*lindhard_shape_retarded(q2_, w_)/q2_**2
top2 = q2_**2/2 + q2_
wp_ = mp.findroot(lambda w_: eps2(w_).real, mp.mpf(1))
dwp = mp.diff(lambda w_: eps2(w_).real, wp_)
print("plasmon case: q = 0.4, q_tf_sq = 3, lambda = 1;  continuum top = %s,  zero of Re eps at omega_p = %s,  d eps/d omega = %s"
      % (mp.nstr(top2, 4), mp.nstr(wp_, 8), mp.nstr(dwp, 6)))
check("Eq.(C2) the zero of Re eps lies outside the particle-hole continuum (a genuine plasmon)", wp_ > top2)
check("Eq.(C2) Im eps vanishes there, so the spectral measure has a point mass", abs(eps2(wp_).imag) < 1e-9)
check("Eq.(C2) the point mass has positive weight (d eps/d omega > 0 at the zero)", dwp > 0)
rho2 = lambda w_: 2*w_/mp.pi*(-1/eps2(w_)).imag
nu2_ = mp.mpf('0.7')
lhs2 = 1 - 1/(1 + l2_*t2_*lindhard_shape_im(q2_, nu2_)/q2_**2)
cont2 = mp.quad(lambda w_: rho2(w_)/(nu2_**2 + w_**2), [0, top2/3, 2*top2/3, top2])
pole2 = 2*wp_/(dwp*(nu2_**2 + wp_**2))
print("   1 - 1/eps(i nu) = %s ;  continuum part = %s ;  plasmon pole = %s ;  sum = %s"
      % (mp.nstr(lhs2, 10), mp.nstr(cont2, 10), mp.nstr(pole2, 10), mp.nstr(cont2 + pole2, 10)))
num("Eq.(C2) spectral identity holds WITH the plasmon point mass included (continuum + pole = LHS)", cont2 + pole2, lhs2, tol=1e-9)
check("Eq.(C2) the continuum alone is NOT enough here: it supplies less than 1% of the LHS", abs(cont2/lhs2) < 0.01)
mp.mp.dps = 30
rho_s, Lam_s, qv2, nu2 = sp.symbols('varrho Lambda q nu', positive=True)
lambda_spec = sp.symbols('lambda_spec', nonnegative=True)
a_l = lambda_spec*rho_s/(qv2*Lam_s**2)
b_L = (1/qv2**2)*Lam_s**2/(Lam_s**2 + (nu2/qv2)**2)
B_W = lambda_spec*rho_s/(2*qv2**3*Lam_s)
exact("Spectral Jacobian: lambda v rho q/(nu^2+Lambda^2 q^2) = a_lambda b_Lambda",
      lambda_spec*(1/qv2**2)*rho_s*qv2/(nu2**2+Lam_s**2*qv2**2)-a_l*b_L)
exact("Physical W spectrum: a_lambda = 2 q^2 B_W(q,q Lambda)/Lambda",a_l-2*qv2**2*B_W/Lam_s)
exact("Zero-coupling spectral measure vanishes",a_l.subs(lambda_spec,0))
```

## Check 23 — Kernel, pole and response prefactors

The elementary Lorentzian kernel identities, Mellin pole terms, screening remainder and the $A_2$ coefficient retain their original numerical values. In the common response units, the shell coefficients are
$$\Pi_{\rm vtx}^{(1)}=-\frac{q_{\rm TF}^4}{4}\alpha(z)\log(1/q)+O(q_{\rm TF}^4),\qquad
\Pi_{\rm SE}^{(1)}=+\frac{q_{\rm TF}^4}{4}\alpha(z)\log(1/q)+O(q_{\rm TF}^4),$$
where $\alpha=h+zh'$ and $X=8\pi^2\alpha\log(1/q)+O(1)$. The self-energy coefficient contains $C=\pi^2q_{\rm TF}^2$ from the response unit, the spin factor 2, the angular measure, the Fock increment and the angular integral $2\alpha$. The code assembles these factors explicitly.

The remaining checks concern $K_\Lambda(y)=\Xi(\Lambda y)/(\pi y)$, Mellin endpoint integrals, $\kappa(\Lambda)$, the rational difference of screened denominators, $2\int_0^1\lambda\,d\lambda/(2\omega_\lambda)=2/\sqrt{3q_{\rm TF}^2}$, $A_2=8/15$, and the quoted momentum scales and auxiliary integrals.


```python
Lam_, y_, t_, z_ = sp.symbols('Lambda y t z', positive=True)
K_Lam = sp.integrate(Lam_**2/(Lam_**2 + z_**2)*sp.exp(-z_*y_), (z_, 0, oo))/pi
Xi_sym = (Lam_*y_)*sp.integrate(sp.exp(-(Lam_*y_)*t_)/(1+t_**2), (t_, 0, oo))
exact("Eq.(D2)  (1/pi) int Lam^2/(Lam^2+z^2) e^{-zy} dz = Xi(Lam y)/(pi y)", sp.simplify(K_Lam - Xi_sym/(pi*y_)))
s_ = sp.symbols('s', positive=True); a0s, b0s = sp.symbols('a_0 b_0')
exact("Eq.(F3)  int_0^1 y^(s-1)(a0 y log y + b0 y) dy = -a0/(s+1)^2 + b0/(s+1)",
      sp.integrate(y_**(s_-1)*(a0s*y_*sp.log(y_) + b0s*y_), (y_, 0, 1)) - (-a0s/(s_+1)**2 + b0s/(s_+1)))
sA = sp.symbols('s', real=True)
exact("Eq.(F3)  int_1^oo y^(s-1) 2 pi log2 / y^2 dy = 2 pi log2/(2-s)   (Re s < 2)",
      sp.integrate(y_**(sA-1)*2*pi*sp.log(2)/y_**2, (y_, 1, oo), conds='none') - 2*pi*sp.log(2)/(2-sA))
eta_ = sp.symbols('eta', positive=True)
exact("Eq.(H4)  kappa(eta/q) q log(1/q) / (q^2 log(1/q)) -> 1/(2 pi^2 eta) as q -> 0", sp.limit(kappa.subs(Lam, eta_/q)/q, q, 0) - 1/(2*pi**2*eta_))
cc_, c0_ = sp.symbols('c c_0', positive=True)
exact("Sec.5.3  1/(1+lam c) - 1/(1+lam c0) = -lam (c - c0)/[(1+lam c)(1+lam c0)]", 1/(1+lam*cc_) - 1/(1+lam*c0_) + lam*(cc_-c0_)/((1+lam*cc_)*(1+lam*c0_)))
exact("Sec.5.4  2 int lam dlam/(2 sqrt(lam q_tf_sq/3)) = 2/sqrt(3 q_tf_sq) = 4/(3 q*^2)", sp.integrate(2*lam/(2*sp.sqrt(lam*q_tf_sq/3)), (lam, 0, 1)) - 4/(3*sp.sqrt(R(4,3)*q_tf_sq)))
exact("Sec.5.4  A2 = 4 pi A0 (-4/3)(-16 pi^4/15) = 8/15", 4*pi*A0*(-R(4,3))*(-16*pi**4/15) - R(8,15))
exact("Eq.(G20) vertex-response shell coefficient", -q_tf_sq**2/(32*pi**2)*8*pi**2 + q_tf_sq**2/4)
exact("Eq.(G20) self-energy-response shell coefficient", (pi**2*q_tf_sq)*2*(2*pi)**(-3)*2*pi*q_tf_sq/4*2-q_tf_sq**2/4)
for rs_v, ref in [(1, '0.97'), (2, '1.15'), (5, '1.45')]:
    num("Sec.5.4  q_* = (4 a_TF r_s/3)^(1/4) at r_s=%d = %s" % (rs_v, ref), (4*aTF_mp*rs_v/3)**mp.mpf(0.25), mp.mpf(ref), tol=6e-3)
num("auxiliary examples  int_1^oo u^-2 L(u) du = 2 log 2", mp.quad(lambda uu: L(uu)/uu**2, [1, 2, mp.inf]), 2*mp.log(2), tol=1e-20)
num("auxiliary examples  int_0^1 log(1/v)/(1-v^2) dv = pi^2/8", mp.quad(lambda vv: mp.log(1/vv)/(1-vv**2), [0, 1]), mp.pi**2/8, tol=1e-20)
```

## Check 24 — Appendix F.2.2, Eqs. (F13)–(F22): the large-$y$ endpoints of the profile

**Where we are in the argument.** The pole of $\widetilde\Phi(s)$ at $s=2$ — the small-$\Lambda$ side of the Mellin convolution and the
convergence strip of the profile transform — is governed by the large-$y$ form $\Phi(y)=\frac{2\pi\log2}{y^2}-\frac{\pi^2}{4}\frac{\log y}{y^2}\sin y+O(y^{-2})$.
The manuscript obtains it by splitting the Bessel product into three trigonometric channels and analysing the endpoints $v=0$ and $u=1$.
The exact identities of that analysis are tested here; the final asymptotic statement (F22) itself remains hand-first (it needs the
integration-by-parts estimates of the remainders).

**Claims.**
- (F13): $yj_1(Ay)j_1(By)=\frac{2}{Dy}[\cos vy+\cos uy]+\frac{8}{D^2y^3}[\cos vy-\cos uy-uy\sin uy+vy\sin vy]$, $A=\tfrac{u+v}2$, $B=\tfrac{u-v}2$, $D=u^2-v^2$.
- Amplitude of the $v$-channel: $\mathcal A(v)=\int_1^\infty\frac{L(u)\ell(u,v)}{u^2-v^2}du=2\log2\,\log\tfrac1v+b_A+O(v^2\log\tfrac1v)$ with $\int_1^\infty L(u)u^{-2}du=2\log2$.
- $\int_0^\delta\log\tfrac1v\cos(vy)\,dv=\frac{\operatorname{Si}(\delta y)-\log\delta\,\sin(\delta y)}{y}$ and $\int_0^\delta\log\tfrac1t\sin(yt)\,dt=\frac{\log y+\gamma-\operatorname{Ci}(\delta y)+\log\delta\cos(\delta y)}{y}$.
- (F19): $J(u)=\int_0^1\frac{\ell(u,v)}{u^2-v^2}dv=\frac{\operatorname{Li}_2(1/u)-\operatorname{Li}_2(-1/u)-\log u\,L(u)}{2u}$; $J(1)=\pi^2/8$; $J(u)=u^{-2}(1-\log u)+O(u^{-4}\log u)$.
- Oscillatory Mellin continuation: $\int_1^\infty y^{s-3}\sin(\alpha y)\,dy=\frac{\cos\alpha}\alpha+\frac{s-3}{\alpha}\int_1^\infty y^{s-4}\cos(\alpha y)\,dy$.

**What the cell does.** (F13) is an `exact` trigonometric identity (`expand_trig` + `simplify`).  $\mathcal A(v)-2\log2\log(1/v)$ is evaluated at
$v=10^{-2},10^{-3},10^{-4}$ and shown to converge (differences $\to0$ like $v^2\log v$), which is the stated structure; $\int L/u^2$ is `num`.
The two endpoint integrals are `num` at $\delta=0.3$, $y=7$.  $J(u)$: closed form vs. defining integral at two $u$; $J(1)$ by `sp.limit` of the closed form
(uses $\operatorname{Li}_2(1)=\pi^2/6$, $\operatorname{Li}_2(-1)=-\pi^2/12$); the large-$u$ form at $u=100$ (tolerance $10^{-3}$, consistent with the $O(u^{-4}\log u)$ remainder).
The oscillatory identity is `num` at $s=1$, $\alpha=1$ with `quadosc`.

**Where this cell departs from the manuscript.** Nowhere in substance; it evaluates displayed identities.

**Analytic scope.** The remainder estimates that turn these identities into (F22), and the statement that the sole pole at $s=2$ is $2\pi\log2/(2-s)$.

```python
uS, vS, yS = sp.symbols('u v y', positive=True)
AS, BS, DS = (uS+vS)/2, (uS-vS)/2, uS**2 - vS**2
j1s = lambda zz: sp.sin(zz)/zz**2 - sp.cos(zz)/zz
lhs = yS*j1s(AS*yS)*j1s(BS*yS)
rhs = 2/(DS*yS)*(sp.cos(vS*yS) + sp.cos(uS*yS)) + 8/(DS**2*yS**3)*(sp.cos(vS*yS) - sp.cos(uS*yS) - uS*yS*sp.sin(uS*yS) + vS*yS*sp.sin(vS*yS))
exact("Eq.(F13) y j1(Ay) j1(By) = trigonometric two-channel form", sp.simplify(sp.expand_trig(sp.expand(lhs - rhs))))
Av = lambda vv: mp.quad(lambda uu: L(uu)*mp.log(1/(uu*vv))/(uu**2 - vv**2), [1, 2, 10, mp.inf])
bA = [Av(mp.mpf(vv)) - 2*mp.log(2)*mp.log(1/mp.mpf(vv)) for vv in ('1e-2', '1e-3', '1e-4')]
print("A(v) - 2 log2 log(1/v) at v = 1e-2, 1e-3, 1e-4:", [mp.nstr(b_, 8) for b_ in bA])
check("Eq.(F14) A(v) = 2 log2 log(1/v) + b_A + o(1): successive differences shrink by ~100x", abs(bA[2]-bA[1]) < abs(bA[1]-bA[0])/30)
num("App.F.2.2  int_1^oo L(u)/u^2 du = 2 log 2", mp.quad(lambda uu: L(uu)/uu**2, [1, 2, mp.inf]), 2*mp.log(2), tol=1e-20)
dl, yv_ = mp.mpf('0.3'), mp.mpf(7)
num("App.F.2.2  int_0^delta log(1/v) cos(vy) dv = [Si(delta y) - log delta sin(delta y)]/y",
    mp.quad(lambda vv: mp.log(1/vv)*mp.cos(vv*yv_), [0, dl]), (mp.si(dl*yv_) - mp.log(dl)*mp.sin(dl*yv_))/yv_, tol=1e-20)
num("App.F.2.2  int_0^delta log(1/t) sin(yt) dt = [log y + gamma - Ci(delta y) + log delta cos(delta y)]/y",
    mp.quad(lambda tt: mp.log(1/tt)*mp.sin(yv_*tt), [0, dl]), (mp.log(yv_) + mp.euler - mp.ci(dl*yv_) + mp.log(dl)*mp.cos(dl*yv_))/yv_, tol=1e-20)
Ju = lambda uu: (mp.polylog(2, 1/uu) - mp.polylog(2, -1/uu) - mp.log(uu)*L(uu))/(2*uu)
Jdef = lambda uu: mp.quad(lambda vv: mp.log(1/(uu*vv))/(uu**2 - vv**2), [0, 1])
for uu in [mp.mpf('1.3'), mp.mpf(4)]:
    num("Eq.(F19) J(u) closed form = defining v-integral, u=%s" % uu, Ju(uu), Jdef(uu), tol=1e-20)
Js = (sp.polylog(2, 1/uS) - sp.polylog(2, -1/uS) - sp.log(uS)*sp.log((uS+1)/(uS-1)))/(2*uS)
exact("Eq.(F19) J(1) = pi^2/8", sp.limit(Js, uS, 1) - pi**2/8)
num("Eq.(F19) J(u) ~ u^-2 (1 - log u) at u=100", Ju(mp.mpf(100))*100**2, 1 - mp.log(100), tol=1e-3)
sv_, al_ = mp.mpf(1), mp.mpf(1)
num("App.F.2.2  int_1^oo y^(s-3) sin(ay) dy = cos(a)/a + (s-3)/a int_1^oo y^(s-4) cos(ay) dy   (s=1, a=1)",
    mp.quadosc(lambda yy: yy**(sv_-3)*mp.sin(al_*yy), [1, mp.inf], omega=al_),
    mp.cos(al_)/al_ + (sv_-3)/al_*mp.quadosc(lambda yy: yy**(sv_-4)*mp.cos(al_*yy), [1, mp.inf], omega=al_), tol=1e-15)
```

## Check 25 — Appendix I, Eqs. (I11)–(I15), (I22): the algebra of the fixed-domain representation of $X_c$

**Where we are in the argument.** Check 16 verified the *output* of Appendix I (the Taylor coefficients of $\langle\mathsf F\rangle$).  The
representation itself — crescent variables, the parametrization of the transverse areas, the discriminant, and the change of variables to
the fixed rectangle — consists of displayed algebraic identities that are tested here.

**Claims.**
- (D2): with $\mathsf R=1-x^2-q^2/4+qxt$, the condition $\mathsf R\ge0$ gives $0\le x\le L_t=\tfrac{qt}2+\sqrt{1-q^2(1-t^2)/4}$; $d\mathsf R\,d\mathsf S=q^2xy\,dt\,ds$.
- Unnumbered: $[R+S+(x+y)^2]^2-4RS=4(x+y)^2+4qxy(x+y)(t+s)+q^2[(xt-ys)^2-(x+y)^2]$, hence the square root in (D1) is $2(x+y)\sqrt H$ with
  $H=1+q\frac{xy(t+s)}{x+y}+\frac{q^2}4\big[\frac{(xt-ys)^2}{(x+y)^2}-1\big]$ (D3).
- (D4): with $x=L_tu$, $y=L_sv$, $v=ru$, $H$ becomes $\widetilde H=1+qu\frac{L_tL_sr(t+s)}{L_t+rL_s}+\frac{q^2}4\big[\frac{(L_tt-rL_ss)^2}{(L_t+rL_s)^2}-1\big]$.
- (D8): with $w=q^2/f_L(q/2)$ and $f_L=1-q^2/12+O(q^4)$: $\mathcal X(w)=\frac{X_c(q)}{q}\frac{d(q^2)}{dw}=c_0+(c_2-c_0/6)w+O(w^2)$.

**What the cell does.** $R$ is a downward parabola in $x$; the cell checks $R(L_t)=0$ and $\partial_x^2R=-2$, so that $R\ge0$ exactly for $x$ between the two roots, the larger of which is $L_t$ (the smaller is negative for $q<2$).  The Jacobian is $\partial(R,S)/\partial(t,s)=(qx)(qy)$.
The discriminant identity and the definition of $H$ are polynomial `exact` checks; (D4) is the substitution $x=L_tu$, $y=rL_su$ into $H$ (with $L_t,L_s$ kept as symbols).
For (D8), $w=z(1+z/12+\dots)$ is inverted as a series, $z(w)=w-w^2/12+\dots$, and $(c_0+c_2z)\,dz/dw$ is expanded in $w$.

**Where this cell departs from the manuscript.** Nowhere.

**Analytic scope.** The geometric identification of the crescents and of the azimuthal integral $4\pi^2/\sqrt{\cdots}$ in (D1).

```python
qD, xD, tD, yD, sD = sp.symbols('q x t y s', real=True)
RD = 1 - xD**2 - qD**2/4 + qD*xD*tD; SD = 1 - yD**2 - qD**2/4 + qD*yD*sD
LtD = qD*tD/2 + sp.sqrt(1 - qD**2*(1-tD**2)/4)
exact("Eq.(I12) R(x = L_t) = 0, i.e. L_t = qt/2 + sqrt(1 - q^2(1-t^2)/4) is the boundary of R >= 0", sp.expand(RD.subs(xD, LtD)))
exact("Eq.(I12) R is a downward parabola: d^2R/dx^2=-2 (boundary signs require the stated domain)", sp.diff(RD, xD, 2) + 2)
exact("Eq.(I12) dR dS = q^2 x y dt ds", sp.Matrix([[sp.diff(RD, tD), sp.diff(RD, sD)], [sp.diff(SD, tD), sp.diff(SD, sD)]]).det() - qD**2*xD*yD)
disc = sp.expand((RD + SD + (xD+yD)**2)**2 - 4*RD*SD)
target = 4*(xD+yD)**2 + 4*qD*xD*yD*(xD+yD)*(tD+sD) + qD**2*((xD*tD - yD*sD)**2 - (xD+yD)**2)
exact("App.I  [R+S+(x+y)^2]^2 - 4RS = 4(x+y)^2 + 4q xy(x+y)(t+s) + q^2[(xt-ys)^2 - (x+y)^2]", disc - sp.expand(target))
HD = 1 + qD*xD*yD*(tD+sD)/(xD+yD) + qD**2/4*((xD*tD - yD*sD)**2/(xD+yD)**2 - 1)
exact("Eq.(I13) discriminant = 4 (x+y)^2 H", sp.simplify(disc/(4*(xD+yD)**2) - HD))
Lt_, Ls_, uD, rD = sp.symbols('L_t L_s u r', positive=True)
Ht = 1 + qD*uD*Lt_*Ls_*rD*(tD+sD)/(Lt_ + rD*Ls_) + qD**2/4*((Lt_*tD - rD*Ls_*sD)**2/(Lt_ + rD*Ls_)**2 - 1)
exact("Eq.(I15) H(x = L_t u, y = r L_s u) = H-tilde", sp.simplify(HD.subs({xD: Lt_*uD, yD: rD*Ls_*uD}) - Ht))
zD, wD, c0D, c2D = sp.symbols('z w c_0 c_2')
fLz = 1 - zD/12                                             # f_L(sqrt z /2) to O(z)
w_of_z = sp.series(zD/fLz, zD, 0, 3).removeO()               # w = z + z^2/12 + ...
z_of_w = wD - wD**2/12                                       # inverse series to O(w^2)
exact("Eq.(I22) inverse series: w(z(w)) = w + O(w^3)", sp.series(w_of_z.subs(zD, z_of_w), wD, 0, 3).removeO() - wD)
Aw = sp.series((c0D + c2D*z_of_w)*sp.diff(z_of_w, wD), wD, 0, 2).removeO()
exact("Eq.(I22) X(w) = c0 + (c2 - c0/6) w + O(w^2)", sp.expand(Aw - (c0D + (c2D - c0D/6)*wD)))
```

## Check 28 — Monte-Carlo tests of the radial kernels

**What is integrated.**  Directly from Eq. (A15) and Eq. (I2)/(H1), with no reduction:
$$X_c(q)=\frac1{16\pi}\!\int\! d^3p\,d^3k\,\frac{\Delta(\bm p;\bm q)\Delta(\bm k;\bm q)}{|\bm p-\bm k|^2}\,\mathcal W(\delta_{\bm p},\delta_{\bm k}),
\qquad
e_\Lambda(q)=4\pi A_0\!\int\! d^3p\,d^3k\,\frac{\Delta\Delta}{|\bm p-\bm k|^2}\,\omega^2J(\omega;\delta_{\bm p},\delta_{\bm k}),\ \ \omega=\Lambda q,$$
where $\mathcal W$ is the static kernel of Eq. (I1) and $J$ the Lorentzian kernel of Eq. (C12) (both verified in Check 8; here they are used, not
re-derived).  The second form follows from Eq. (H1) by $\frac{\Lambda^2}{\Lambda^2+\nu^2/q^2}=\frac{\omega^2}{\omega^2+\nu^2}$, which is the one
line of algebra shown in the code.

**How the Monte Carlo is set up.**
- *Small $q$ (all four sectors).*  $\bm p$ is drawn uniformly from the ball of radius $1+q$ that contains the support of $\Delta_{\bm p}$, and
  the **relative** coordinate $\bm r=\bm p-\bm k$ from the density $\propto|\bm r|^{-2}$ on $|\bm r|<2+q$.  That density cancels the Coulomb
  factor exactly, so the estimator is bounded; with uniform sampling the variance would be infinite ($\int d^3r\,|\bm r|^{-4}=\infty$).
  Pairs with $\Delta_{\bm p}\Delta_{\bm k}=0$ contribute zero.  This is the regime where the shell and the near-coincidence
  $\hat{\bm p}\simeq\hat{\bm k}$ matter, i.e. exactly what App. I's crescent construction handles analytically.
- *Large $q$ ($q>2$, static case).*  The same-sphere sectors have $\delta_{\bm p}\delta_{\bm k}>0$, where $\mathcal W=0$ identically, so only the
  two cross sectors survive; there $|\bm p-\bm k|\ge q-2$ and plain uniform sampling in the two balls is efficient.
- *Large $q$ (Lorentzian kernel).*  Both sector types contribute; the same-sphere pair uses the $|\bm r|^{-2}$ sampling, the cross pair uniform sampling.
  The factor 2 in each case is the mirror symmetry $(\bm p,\bm k)\to(-\bm p-\bm q,-\bm k-\bm q)$, which flips both $\Delta$'s and both $\delta$'s
  and leaves $\mathcal W$, $J$ and $|\bm p-\bm k|$ unchanged.

**Reading the output.**
- 28.1 small $q$: $X_c/q$ must match $c_0+c_2q^2$ with $c_0=\pi(2\log2-1)/8=0.151697$ (Eq. (I3)) and $c_2$ of Eq. (I7).  Note that the data show
  no sign of a $\log(1/q)$: such a term would make $X_c/q$ rise by $\kappa\log 4\simeq1.4\kappa$ between $q=0.4$ and $q=0.1$.
- 28.1 large $q$: $q^4X_c\to C_\infty=2\pi/9=0.69813$ (Eq. (I4)), approached from above like $1+O(q^{-2})$.
- 28.2: $q^2e_\Lambda\to\frac{3}{\pi^3}p_{\rm loc}(s_\Lambda)$ with $p_{\rm loc}(s_\Lambda)=2\pi s_\Lambda/(1+2s_\Lambda)^2$ (Eq. (H3)), for $s_\Lambda=\Lambda/q$ held fixed.
- 28.3: the two limits together — $X_c\sim q^{-4}$ versus $e_\Lambda\sim q^{-2}$ — are the numerical signature of Eq. (I1): the same-sphere
  sectors, which dominate at large $q$, are annihilated by a static line and retained by a dynamical one.  The experiment measures
  $e_\Lambda/X_c$ at $q=8$ and $q=16$ (at fixed $s_\Lambda=\tfrac12$) and checks both that it grows by the predicted factor 4 when $q$ doubles and
  that its value agrees with $\frac{3}{\pi^3}P(\tfrac12)/C_\infty\cdot q^2$.  This is the quantitative form of the statement that decides
  $r_s^{-1/4}$ versus $r_s^{-3/4}$.

**What a pass does not establish.** The crescent parametrization itself (App. I) and the thin-shell reduction of Appendix G.1 (`app-eq:shell`) are not reproduced; what is
tested is that the numbers they produce agree with the unreduced integral.  The small-$q$ experiment reaches $q=0.1$, so it constrains the
Taylor structure of Eq. (I3) but says nothing about terms beyond $q^3$.

```python
import numpy as np
rng = np.random.default_rng(20260912)
VB_ = 4*np.pi/3; A0_np = 3/(32*np.pi**5)
c0_np = np.pi/8*(2*np.log(2) - 1); c2_np = np.pi/1152 - np.pi*np.log(2)/96; Cinf_np = 2*np.pi/9

def _ball(m, Rr=1.0):
    v = rng.normal(size=(m, 3)); v /= np.linalg.norm(v, axis=1)[:, None]
    return v*(Rr*rng.random(m)**(1/3))[:, None]
def _deltas(p, k, q):
    qv = np.array([0, 0, q])
    Dp = (np.linalg.norm(p, axis=1) < 1).astype(float) - (np.linalg.norm(p + qv, axis=1) < 1)
    Dk = (np.linalg.norm(k, axis=1) < 1).astype(float) - (np.linalg.norm(k + qv, axis=1) < 1)
    return Dp, Dk, p[:, 2]*q + q*q/2, k[:, 2]*q + q*q/2
W_st = lambda a_, b_: (np.sign(a_*b_) - 1)/(2*(np.abs(a_) + np.abs(b_)))            # Eq.(I1)
def J_bl(w_, a_, b_):                                                              # Eq.(C12)
    A_, B_ = np.abs(a_), np.abs(b_)
    return np.where(a_*b_ > 0, 1/(2*w_*(A_ + w_)*(B_ + w_)),
                    -(A_ + B_ + 2*w_)/(2*w_*(A_ + B_)*(A_ + w_)*(B_ + w_)))

def Xc_small(q, n=2_500_000, rep=4):
    Rp = 1 + q; rmax = 2 + q; out = []
    for _ in range(rep):
        p = _ball(n, Rp)
        w = rng.normal(size=(n, 3)); w /= np.linalg.norm(w, axis=1)[:, None]
        k = p - w*(rmax*rng.random(n))[:, None]
        Dp, Dk, dp, dk = _deltas(p, k, q)
        f = np.where((Dp != 0) & (Dk != 0), Dp*Dk*W_st(dp, dk), 0.0)
        out.append((4*np.pi/3*Rp**3)*(4*np.pi*rmax)*f.mean()/(16*np.pi))
    v = np.array(out); return v.mean(), v.std(ddof=1)/np.sqrt(rep)

def Xc_large(q, n=1_500_000, rep=4):
    out = []
    for _ in range(rep):
        p = _ball(n); k = _ball(n) - np.array([0, 0, q])
        dp = p[:, 2]*q + q*q/2; dk = k[:, 2]*q + q*q/2
        out.append(2*VB_**2*np.mean(-W_st(dp, dk)/np.sum((p - k)**2, axis=1))/(16*np.pi))
    v = np.array(out); return v.mean(), v.std(ddof=1)/np.sqrt(rep)

def eLam_large(q, Lam, n=800_000, rep=4):
    om = Lam*q; out = []
    for _ in range(rep):
        p = _ball(n)
        w = rng.normal(size=(n, 3)); w /= np.linalg.norm(w, axis=1)[:, None]
        rmax = 2.0; k = p - w*(rmax*rng.random(n))[:, None]
        inside = np.linalg.norm(k, axis=1) < 1
        dp = p[:, 2]*q + q*q/2; dk = k[:, 2]*q + q*q/2
        same = 2*VB_*(4*np.pi*rmax)*np.where(inside, om**2*J_bl(om, dp, dk), 0).mean()
        p2 = _ball(n); k2 = _ball(n) - np.array([0, 0, q])
        dp2 = p2[:, 2]*q + q*q/2; dk2 = k2[:, 2]*q + q*q/2
        cross = 2*VB_**2*np.mean(-om**2*J_bl(om, dp2, dk2)/np.sum((p2 - k2)**2, axis=1))
        out.append(4*np.pi*A0_np*(same + cross))
    v = np.array(out); return v.mean(), v.std(ddof=1)/np.sqrt(rep)

print("28.1  static kernel at small q:   X_c/q   vs   c0 + c2 q^2      (c0 = %.6f, c2 = %.6f)" % (c0_np, c2_np))
ok_small = True
for qv_ in [0.1, 0.2, 0.4]:
    m_, s_ = Xc_small(qv_); pred = c0_np + c2_np*qv_**2
    print("   q=%.2f :  X_c/q = %.5f +- %.5f    c0 + c2 q^2 = %.5f    (dev = %+.1f sigma)"
          % (qv_, m_/qv_, s_/qv_, pred, (m_/qv_ - pred)/(s_/qv_)))
    ok_small = ok_small and abs(m_/qv_ - pred) < 3*(s_/qv_)
check("Eqs.(I3),(I7)  X_c/q from the 6-dim Monte Carlo agrees with c0 + c2 q^2 at q = 0.1, 0.2, 0.4 (3 sigma)", ok_small)

print("28.1  static kernel at large q:   q^4 X_c  vs  C_inf = %.5f" % Cinf_np)
rat = {}
for qv_ in [3.0, 5.0, 8.0, 15.0]:
    m_, s_ = Xc_large(qv_); rat[qv_] = m_*qv_**4/Cinf_np
    print("   q=%5.1f :  q^4 X_c = %.5f +- %.5f    ratio to C_inf = %.4f" % (qv_, m_*qv_**4, s_*qv_**4, rat[qv_]))
check("Eq.(I4)  q^4 X_c -> C_inf = 2 pi/9 (within 1% at q = 15)", abs(rat[15.0] - 1) < 0.01)
r58 = (rat[5.0] - 1)/(rat[8.0] - 1); r815 = (rat[8.0] - 1)/(rat[15.0] - 1)
print("   deviation ratios:  (q=5)/(q=8) = %.2f  vs  (8/5)^2 = %.2f ;   (q=8)/(q=15) = %.2f  vs  (15/8)^2 = %.2f"
      % (r58, (8/5)**2, r815, (15/8)**2))
check("Eq.(I4)  the approach is O(q^-2): deviation ratios match (q'/q)^2 within 25%",
      0.75 < r58/(8/5)**2 < 1.25 and 0.75 < r815/(15/8)**2 < 1.25)

print("28.2  Lorentzian kernel at large q:    q^2 e_Lambda  vs  (3/pi^3) p_loc(s_lambda)")
p_loc_np = lambda s_lambda_: 2*np.pi*s_lambda_/(1 + 2*s_lambda_)**2
rat2 = {}
for qv_, s_lambda_ in [(4.0, 0.5), (8.0, 0.5), (16.0, 0.5), (8.0, 2.0)]:
    m_, s_ = eLam_large(qv_, s_lambda_*qv_); pred = 3/np.pi**3*p_loc_np(s_lambda_)
    rat2[(qv_, s_lambda_)] = m_*qv_**2/pred
    print("   q=%5.1f s_lambda=%.1f :  q^2 e_Lambda = %.6f +- %.6f    (3/pi^3)p_loc(s_lambda) = %.6f    ratio = %.4f"
          % (qv_, s_lambda_, m_*qv_**2, s_*qv_**2, pred, rat2[(qv_, s_lambda_)]))
check("Eq.(H3)  q^2 e_Lambda -> (3/pi^3) p_loc(s_lambda) at fixed s_lambda (within 1% at q = 16)", abs(rat2[(16.0, 0.5)] - 1) < 0.01)
check("Eq.(H3)  the deviation decreases monotonically with q at fixed s_lambda",
      abs(rat2[(4.0, 0.5)] - 1) > abs(rat2[(8.0, 0.5)] - 1) > abs(rat2[(16.0, 0.5)] - 1))

print("28.3  static vs dynamical at large q (the mechanism of Eq.(I1)):")
ratios_sd = {}
for qv_ in [8.0, 16.0]:
    mS, _ = Xc_large(qv_); mB, _ = eLam_large(qv_, 0.5*qv_)
    ratios_sd[qv_] = mB/mS
    print("   q=%4.1f :  X_c = %.3e (q^4 X_c = %.3f)   e_Lambda = %.3e (q^2 e_Lambda = %.5f)   e_Lambda/X_c = %.2f"
          % (qv_, mS, mS*qv_**4, mB, mB*qv_**2, mB/mS))
growth = ratios_sd[16.0]/ratios_sd[8.0]
pred_ratio = (3/np.pi**3*p_loc_np(0.5))/Cinf_np
print("   e_Lambda/X_c grows by %.2f when q doubles (q^2 growth predicts 4.00);  predicted ratio = %.4f q^2" % (growth, pred_ratio))
check("Eq.(I1)  e_Lambda/X_c grows like q^2 (factor 4 per doubling, +-15%): the same-sphere sectors survive only for the dynamical line",
      3.4 < growth < 4.6)
check("Eq.(I1)  the measured e_Lambda/X_c matches (3/pi^3)P(1/2)/C_inf times q^2 within 3% at q=16",
      abs(ratios_sd[16.0]/(pred_ratio*16.0**2) - 1) < 0.03)
```

## Check 29 — Independent loop-integral tests of the reduction and asymptotes

**29.1 and 29.2 — the master formula.**  Equation (D1) turns
$\mathcal K[b_\Lambda]=A_0\int\frac{d\nu}{\pi}\int d^3q\,b_\Lambda\,X(\bm q,\nu)$ into a single integral of $\Xi(\Lambda y)\Phi(y)/y$.  The
notebook cannot re-derive the Frullani step or the prolate-spheroidal change of variables, but it can evaluate
$\mathcal K[b_\Lambda]=\int_0^\infty e_\Lambda(q)\,dq$ directly by Monte Carlo, using only the definitions and the frequency kernel of App. C.2, and
compare with the two numbers the manuscript quotes: $\int X_c\,dq=D_0=0.24990$ (equivalently $\mathcal K[b_\infty]=e_{2x}$, the Onsager value)
and $\mathcal K[b_{\Lambda=1}]=112.868$ mRy (auxiliary examples).

**How the $q$-integration is done, and a bug this notebook had.**  Rather than evaluating $e_\Lambda(q)$ on a grid and quadrating, the cell
samples $q$ together with $(\bm p,\bm k)$: $q=s\tan t$ with $t$ uniform on $(0,\pi/2)$ and weight $s\frac\pi2(1+(q/s)^2)$, so there is no
$q$-quadrature error at all and the kink at $q=2$ (where the two displaced Fermi balls separate) costs nothing.  The momentum $\bm p$ is drawn
from the **exact support** $B\cup(B-\bm q)$ — one of the two unit balls with probability $\tfrac12$, weighted by $2V_B/m(\bm p)$ with
$m\in\{1,2\}$ the number of balls containing $\bm p$.  *An earlier version of this notebook drew $\bm p$ from a bounding ball of radius $1+q$.
That is also unbiased, but the acceptance falls like $q^{-6}$, and it produced a spurious $1\%$ excess in $\int X_c\,dq$ and lost the large-$q$
plateau that carries the $\log\Lambda$ of 29.3 entirely.*  With the support sampler, $\int X_c\,dq$ agrees with $D_0$ to within its statistical
error.

**29.3 — the contour, and the slope.**  Equation (F6) states $S(\Lambda)=1+(C_1^\Lambda\log\Lambda+C_0^\Lambda)/\Lambda+\dots$ with $C_1^\Lambda=\pi/(4D_0)>0$.
The *sign* of these coefficients is exactly what the direction of the contour closure decides (the residues are subtracted, not added), so
measuring $S$ at large $\Lambda$ tests that orientation: $S$ must approach 1 **from above**.  The experiment evaluates $S(\Lambda)=\mathcal K[b_\Lambda]/e_{2x}$
at $\Lambda=10$ and $30$, where the asymptotic formula predicts $1.645$ and $1.330$, and at $\Lambda=1.2$, where the manuscript's Fig. 6 has its
maximum $2.34$.  The sharper form of the test is the finite-difference slope of $(S-1)\Lambda$ against $\log\Lambda$, which must equal
$C_1^\Lambda=\pi/(4D_0)=3.14283$; that number comes from the double pole at $s=1$, i.e. from $a_0=\tfrac12$, $\widetilde\Xi(1)=-\pi/2$ and the direction of
the contour shift, none of which the Monte Carlo knows about.  The manuscript bounds the remainder by $O(\Lambda^{-2}(1+\log^3\Lambda))$, agreement to a few per
cent is the most that can be expected at these $\Lambda$.

**29.4 — the shell coefficient.**  In the shell limit Eq. (G2) implies $X(\bm q,qz)=A(z)\log\frac1q+B(z)+o(1)$ with
$A(z)=2\pi\int d\Omega_p\,c_p^2\frac{c_p^2-z^2}{(c_p^2+z^2)^2}=8\pi^2\big[1-2z\arctan\frac1z+\frac{z^2}{1+z^2}\big]$ (Eq. (G9)) — the function whose
frequency integral is $\mathcal A$ and hence the whole high-density double logarithm.  The experiment evaluates $X(\bm q,qz)$ by Monte Carlo at
$q=0.04,0.07,0.12,0.20$ and fits the coefficient of $\log(1/q)$.  Note that $A(z)$ changes sign between $z=\tfrac12$ and $z=1$; reproducing that
sign change is a non-trivial test of the angular structure.

*How the agreement is judged.*  The natural scale of $A(z)$ is $8\pi^2=79$ — its value at $z=0$ — and near the zero-crossing $A$ is small
($A(0.5)=7.33$, under a tenth of the scale), so a fixed *relative* tolerance there would demand an accuracy the fit cannot deliver.  The check
therefore compares the absolute deviation with $2\%$ of $8\pi^2$, which is the same accuracy requirement at every $z$.

**A warning that this notebook learned the hard way.**  Fits of the form $c_1\log(1/q)+c_0$ over a range of $q$ that reaches up to $q\sim0.5$ are
biased *high* by the neglected $o(1)$ terms — in a first attempt at the frequency-integrated version of 29.4 the bias reached $8\%$ and looked
like a discrepancy.  The fits below are therefore restricted to $q\le0.2$, and the reader should treat any single fitted slope as accurate to a
few percent at best, not to the quoted statistical error.

**What these experiments do not establish.**  They confirm the *numbers* the reductions produce at a few points; they do not reproduce the
reductions.  A compensating pair of errors that happens to leave $\mathcal K[b_\Lambda]$ unchanged at $\Lambda=1$, $1.2$ and $30$, and $A(z)$
unchanged at three values of $z$, would survive.

```python
import numpy as np, time
rng9 = np.random.default_rng(20260913)
VB9 = 4*np.pi/3; A0_9 = 3/(32*np.pi**5)
def _ball9(m, Rr=1.0):
    v = rng9.normal(size=(m, 3)); v /= np.linalg.norm(v, axis=1)[:, None]
    return v*(Rr*rng9.random(m)**(1/3))[:, None]
def _support9(q, n):
    # p uniform on the exact support B u (B - q e_z): pick one ball w.p. 1/2, weight 2 V_B/multiplicity.
    # Sampling from a bounding ball of radius 1+q is also unbiased but wastes a factor (1+q)^3/2 of the points,
    # which at q > 2 makes the estimator far noisier and at large q useless.
    p = _ball9(n)
    qv = np.zeros((n, 3)); qv[:, 2] = q
    p = np.where((rng9.random(n) < 0.5)[:, None], p - qv, p)
    mult = (np.linalg.norm(p, axis=1) < 1).astype(float) + (np.linalg.norm(p + qv, axis=1) < 1)
    return p, 2*VB9/np.maximum(mult, 1)
def _pair9(q, n):
    # Delta_p Delta_k, delta_p, delta_k and the measure weight, for one batch at momentum transfer q
    p, wp = _support9(q, n); rmax = 2 + q
    w = rng9.normal(size=(n, 3)); w /= np.linalg.norm(w, axis=1)[:, None]
    k = p - w*(rmax*rng9.random(n))[:, None]
    qv = np.zeros((n, 3)); qv[:, 2] = q
    Dp = (np.linalg.norm(p, axis=1) < 1).astype(float) - (np.linalg.norm(p + qv, axis=1) < 1)
    Dk = (np.linalg.norm(k, axis=1) < 1).astype(float) - (np.linalg.norm(k + qv, axis=1) < 1)
    return Dp*Dk, p[:, 2]*q + q*q/2, k[:, 2]*q + q*q/2, wp*(4*np.pi*rmax)
def intXc_tan(n=2_000_000, rep=6):
    # int_0^oo X_c dq in one estimator: q = tan(t), t uniform on (0,pi/2), so no q-quadrature is needed
    out = []
    for _ in range(rep):
        t = rng9.random(n)*(np.pi/2); q = np.tan(t); wq = (np.pi/2)*(1 + q**2)
        DD, dp, dk, wm = _pair9(q, n)
        f = np.where(DD != 0, DD*(np.sign(dp*dk) - 1)/(2*(np.abs(dp) + np.abs(dk))), 0.0)/(16*np.pi)
        out.append((wq*wm*f).mean())
    v = np.array(out); return v.mean(), v.std(ddof=1)/np.sqrt(rep)
def calK_lorentzian(Lam, n=2_000_000, rep=6, scale=1.0):
    out = []
    for _ in range(rep):
        t = rng9.random(n)*(np.pi/2); q = scale*np.tan(t); wq = scale*(np.pi/2)*(1 + (q/scale)**2)
        DD, dp, dk, wm = _pair9(q, n)
        om = Lam*q
        f = np.where(DD != 0, DD*om**2*_J9(om, dp, dk), 0.0)
        out.append(4*np.pi*A0_9*(wq*wm*f).mean())
    v = np.array(out); return v.mean(), v.std(ddof=1)/np.sqrt(rep)
def _J9(w_, a_, b_):
    A_, B_ = np.abs(a_), np.abs(b_)
    return np.where(a_*b_ > 0, 1/(2*w_*(A_ + w_)*(B_ + w_)),
                    -(A_ + B_ + 2*w_)/(2*w_*(A_ + B_)*(A_ + w_)*(B_ + w_)))
_W9 = lambda a_, b_: (np.sign(a_*b_) - 1)/(2*(np.abs(a_) + np.abs(b_)))
def kern9(q, Lam, n, rep, static=False):
    # X_c(q) if static (normalisation 1/16pi), else e_Lambda(q) = 4 pi A0 <Delta Delta omega^2 J / |p-k|^2>
    om = None if static else Lam*q; out = []
    for _ in range(rep):
        if q <= 3:
            Rp = 1 + q; rmax = 2 + q
            p = _ball9(n, Rp)
            w = rng9.normal(size=(n, 3)); w /= np.linalg.norm(w, axis=1)[:, None]
            k = p - w*(rmax*rng9.random(n))[:, None]
            qv = np.array([0, 0, q])
            Dp = (np.linalg.norm(p, axis=1) < 1).astype(float) - (np.linalg.norm(p + qv, axis=1) < 1)
            Dk = (np.linalg.norm(k, axis=1) < 1).astype(float) - (np.linalg.norm(k + qv, axis=1) < 1)
            dp = p[:, 2]*q + q*q/2; dk = k[:, 2]*q + q*q/2
            f = _W9(dp, dk) if static else om**2*_J9(om, dp, dk)
            est = (4*np.pi/3*Rp**3)*(4*np.pi*rmax)*np.where((Dp != 0) & (Dk != 0), Dp*Dk*f, 0.0).mean()
        else:
            p = _ball9(n)
            w = rng9.normal(size=(n, 3)); w /= np.linalg.norm(w, axis=1)[:, None]
            rmax = 2.0; k = p - w*(rmax*rng9.random(n))[:, None]
            ins = np.linalg.norm(k, axis=1) < 1
            dp = p[:, 2]*q + q*q/2; dk = k[:, 2]*q + q*q/2
            f = _W9(dp, dk) if static else om**2*_J9(om, dp, dk)
            same = 2*VB9*(4*np.pi*rmax)*np.where(ins, f, 0).mean()
            p2 = _ball9(n); k2 = _ball9(n) - np.array([0, 0, q])
            dp2 = p2[:, 2]*q + q*q/2; dk2 = k2[:, 2]*q + q*q/2
            f2 = _W9(dp2, dk2) if static else om**2*_J9(om, dp2, dk2)
            est = same + 2*VB9**2*np.mean(-f2/np.sum((p2 - k2)**2, axis=1))
        out.append(est/(16*np.pi) if static else 4*np.pi*A0_9*est)
    v = np.array(out); return v.mean(), v.std(ddof=1)/np.sqrt(rep)
def integrate9(Lam, static, hi, n=500_000, rep=2, nn=12):
    tot = 0.0; var = 0.0
    for lo_, hi_ in [(0.02, 0.5), (0.5, 2.0), (2.0, 6.0), (6.0, hi)]:
        xg, wg = np.polynomial.legendre.leggauss(nn)
        tg = (np.log(hi_) - np.log(lo_))/2*xg + (np.log(hi_) + np.log(lo_))/2
        wg = (np.log(hi_) - np.log(lo_))/2*wg
        for q_, w_ in zip(np.exp(tg), wg):
            m_, s_ = kern9(q_, Lam, n, rep, static)
            tot += w_*q_*m_; var += (w_*q_*s_)**2
    return tot, np.sqrt(var)

t9 = time.time()
D0_9 = float(sp.N(D0, 20)); e2x_9 = float(sp.N(e2x, 20))
I_, e_ = intXc_tan()
print("29.1  int X_c dq = %.5f +- %.5f    D_0 = %.5f    (%+.1f sigma)   [%.0f s]" % (I_, e_, D0_9, (I_ - D0_9)/e_, time.time() - t9))
print("      implied e_2x = (6/pi^3) int X_c dq = %.5f Ry   vs Onsager-Mittag-Stephen %.5f Ry" % (6/np.pi**3*I_, e2x_9))
check("Eqs.(D1),(I2)  the 6-dim Monte Carlo gives int X_c dq = D_0 (master formula at the static point)", abs(I_ - D0_9) < 3*e_)

I1, e1 = calK_lorentzian(1.0, scale=1.0)
print("29.2  K[b_1] = %.3f +- %.3f mRy    auxiliary examples: 112.868 mRy    S(1) = %.4f   [%.0f s]" % (I1*1000, e1*1000, I1/e2x_9, time.time() - t9))
check("Eq.(D1), auxiliary examples  the 6-dim Monte Carlo gives K[b_1] = 112.868 mRy within 3 sigma", abs(I1*1000 - 112.868) < 3*e1*1000)

C1_9 = float(sp.N(pi/(4*D0), 20)); C0_9 = float(sp.N(-pi/(16*D0), 20))
S_vals = {}
for Lam_, sc_ in [(1.2, 1.0), (10.0, 4.0), (30.0, 10.0)]:
    IL, eL = calK_lorentzian(Lam_, scale=sc_)
    S_vals[Lam_] = (IL/e2x_9, eL/e2x_9)
    pred = 1 + (C1_9*np.log(Lam_) + C0_9)/Lam_
    print("29.3  S(%.1f) = %.4f +- %.4f    asymptotic 1 + (C1 log L + C0)/L = %.4f   [%.0f s]"
          % (Lam_, IL/e2x_9, eL/e2x_9, pred, time.time() - t9))
check("Sec.4, Fig.6  S(1.2) = 2.34 (the maximum of the block enhancement)", abs(S_vals[1.2][0] - 2.34) < 0.03)
pred30 = 1 + (C1_9*np.log(30.0) + C0_9)/30.0
check("Eq.(F6)  S(30) matches 1 + (C1 log L + C0)/L within 1.5%: the contour is closed in the direction that SUBTRACTS the residues",
      abs(S_vals[30.0][0] - pred30)/pred30 < 0.015)
check("Eq.(F6)  S approaches 1 from above (C1 > 0), which fixes the sign of the residue contribution", S_vals[30.0][0] > 1)
slope9 = ((S_vals[30.0][0] - 1)*30.0 - (S_vals[10.0][0] - 1)*10.0)/np.log(3.0)
serr9 = np.hypot(S_vals[30.0][1]*30.0, S_vals[10.0][1]*10.0)/np.log(3.0)
print("29.3  slope of (S-1)*Lambda against log Lambda, from Lambda = 10 -> 30 : %.3f +- %.3f     C_1^Lambda = pi/(4 D_0) = %.5f"
      % (slope9, serr9, C1_9))
check("Eq.(F6)  the measured large-Lambda slope reproduces C_1^Lambda = pi/(4 D_0) within 10% (the double pole at s=1)",
      abs(slope9/C1_9 - 1) < 0.10)
```

<a id="check-29-4"></a>

### 29.4 — the shell coefficient $A(z)$

The cell below evaluates $X(\bm q,qz)$ — Eq. (A15) with no reduction — at four small $q$ and fits the coefficient of $\log(1/q)$.
The frequency kernel itself scales as $q^{-2}$ and the two shell thicknesses give $q^{2}$, so $X(\bm q,qz)$ tends to a *logarithm*, not to zero:
this cancellation of powers is itself part of the claim being tested.

```python
def Xqz9(q, z, n=1_500_000, rep=4):
    nu = q*z; out = []
    for _ in range(rep):
        Rp = 1 + q; rmax = 2 + q
        p = _ball9(n, Rp)
        w = rng9.normal(size=(n, 3)); w /= np.linalg.norm(w, axis=1)[:, None]
        k = p - w*(rmax*rng9.random(n))[:, None]
        qv = np.array([0, 0, q])
        Dp = (np.linalg.norm(p, axis=1) < 1).astype(float) - (np.linalg.norm(p + qv, axis=1) < 1)
        Dk = (np.linalg.norm(k, axis=1) < 1).astype(float) - (np.linalg.norm(k + qv, axis=1) < 1)
        dp = p[:, 2]*q + q*q/2; dk = k[:, 2]*q + q*q/2
        ker = (dp*dk - nu**2)/((dp**2 + nu**2)*(dk**2 + nu**2))
        out.append((4*np.pi/3*Rp**3)*(4*np.pi*rmax)*np.where((Dp != 0) & (Dk != 0), Dp*Dk*ker, 0.0).mean())
    v = np.array(out); return v.mean(), v.std(ddof=1)/np.sqrt(rep)
A_shell = lambda z: 8*np.pi**2*(1 - 2*z*np.arctan(1/z) + z**2/(1 + z**2))
qs9 = np.array([0.04, 0.07, 0.12, 0.20]); L9 = np.log(1/qs9)
dev_list = []
for z_ in [0.5, 1.0, 2.0]:
    ys = []; es = []
    for q_ in qs9:
        m_, s_ = Xqz9(q_, z_); ys.append(m_); es.append(s_)
    y9 = np.array(ys); e9 = np.array(es); W9 = 1/e9**2
    M9 = np.vstack([L9, np.ones_like(L9)]).T
    cov9 = np.linalg.inv(M9.T@(M9*W9[:, None])); c9 = cov9@(M9.T@(W9*y9))
    rel = abs(c9[0] - A_shell(z_))/abs(A_shell(z_))
    dev_list.append(c9[0] - A_shell(z_))
    print("29.4  z=%.1f :  fitted coefficient of log(1/q) = %+8.4f +- %.4f    A(z) = %+8.4f    relative deviation %.1f%%   [%.0f s]"
          % (z_, c9[0], np.sqrt(cov9[0, 0]), A_shell(z_), 100*rel, time.time() - t9))
dev_abs = max(abs(d_) for d_ in dev_list)
print("29.4  largest deviation from A(z), measured against the natural scale 8 pi^2 = %.1f :  %.4f  (%.2f %% of the scale)"
      % (8*np.pi**2, dev_abs, 100*dev_abs/(8*np.pi**2)))
check("Eqs.(G2),(G9)  the log(1/q) coefficient of the unreduced X(q,qz) equals A(z) to 2% of the scale 8 pi^2 at z = 0.5, 1, 2 (including the sign change)",
      dev_abs < 0.02*8*np.pi**2)
```

## Check 30 — The Gell-Mann–Brueckner limit of the RPA ring energy

**What is computed.**  From the same ingredients used throughout — $q_{\rm TF}^{2}=a_{\rm TF}r_s$ with $a_{\rm TF}=4\alpha_L/\pi$, $\alpha_L=(4/9\pi)^{1/3}$, the
imaginary-axis Lindhard function $\mathcal L$ normalized by $\mathcal L(0,0)=1$, momenta in $k_F$, energies in $\hbar^2k_F^2/m$ — the RPA ring energy per electron is
$$\varepsilon_{\rm ring}(r_s)=\frac{3}{2\pi\alpha_L^2}\,\frac{1}{r_s^{2}}\int_0^\infty\!\!q^2dq\int_0^\infty\!\!d\nu\,\big[\ln(1+c)-c\big],
\qquad c=\frac{q_{\rm TF}^{2}\,\mathcal L(q,\nu)}{q^{2}} .$$
The prefactor uses nothing from the manuscript beyond its conventions: $n=k_F^3/3\pi^2$ for the per-electron normalization, $d^3q/(2\pi)^3$ for
the measure, $k_F=1/(\alpha_Lr_s)$ in atomic units for the energy unit $\hbar^2k_F^2/m=k_F^2$ Ha, and a factor 2 for Rydberg; the subtraction of
$c$ removes the first-order term, leaving the ring sum from second order on.

**The anchor.**  As $r_s\to0$ this must reproduce Gell-Mann–Brueckner:
$$\varepsilon_{\rm ring}=\frac{2(1-\ln2)}{\pi^2}\ln r_s-0.142+o(1)\ \text{Ry},$$
and, added to Onsager's $e_{2x}$ (Check 1), the Carr–Maradudin constant of $\varepsilon_c=0.0622\ln r_s-0.094$ Ry.

**Why this is a real test.**  The slope is fixed by the region $q\sim q_{\rm TF}=q_{\rm TF}$, so it is sensitive to $a$ *and* to the
normalization of $\mathcal L$; the constant is sensitive in addition to the overall energy conversion.  Getting both to four digits pins down every
convention that multiplies $A_1$, $A_2$, $A_3$ (and $C_{\log^2}$, $A_{3/4}$, $\ldots$).  The quadrature is delicate: the $q$-grid must resolve
well below $q_{\rm TF}$ and the $\nu$-grid must sit at the particle-hole scale $\nu\sim q$ rather than at the plasmon scale, or the logarithm is
truncated — the cell does both explicitly.

**What it does not test.** The diagrammatic content of the kite itself; only the units, the density variable and the response normalization.

```python
import numpy as np
alphaL_np = (4/(9*np.pi))**(1/3); aTF_np = 4*alphaL_np/np.pi
ring_prefactor = 3/(2*np.pi*alphaL_np**2)                     # Ry per unit of the reduced ring integral

def lindhard_shape_midpoint(q, nu, nk=600):                       # imaginary-axis Lindhard function, from its defining k-integral
    k = (np.arange(nk) + 0.5)/nk
    return np.mean(k*np.log(((q*k + q*q/2)**2 + nu**2)/((q*k - q*q/2)**2 + nu**2)))/(2*q)
def _glA(a, b, n):
    xg, wg = np.polynomial.legendre.leggauss(n); return (b - a)/2*xg + (a + b)/2, (b - a)/2*wg
def eps_ring(rs, nq=90, nx=140):
    q_tf_sq_ = aTF_np*rs; tot = 0.0
    lo = np.log10(min(0.01, np.sqrt(q_tf_sq_))) - 3            # resolve three decades below q_TF
    hi = max(1.0, np.log10(10*(4*q_tf_sq_/3)**0.25))           # and a decade above q_*
    for A_, B_ in [(lo, lo+2), (lo+2, lo+4), (lo+4, 0), (0, 1), (1, hi), (hi, hi+2)]:
        if B_ <= A_: continue
        tq, wq = _glA(A_, B_, nq); qs_ = 10**tq; jq = wq*qs_*np.log(10)
        for qi, wi in zip(qs_, jq):
            sc = max(qi, qi**2, np.sqrt(q_tf_sq_/3))            # particle-hole (q, q^2) and plasmon (sqrt(q_tf_sq/3)) scales
            t = (np.arange(nx) + 0.5)*(np.pi/2)/nx
            nu = sc*np.tan(t); wx = sc/np.cos(t)**2*((np.pi/2)/nx)
            c_ = np.array([q_tf_sq_*lindhard_shape_midpoint(qi, x_)/qi**2 for x_ in nu])
            tot += wi*qi**2*np.sum((np.log1p(c_) - c_)*wx)
    return ring_prefactor*tot/rs**2

vals_ring = {}
for rs_v in [1e-8, 1e-6, 1e-4]:
    vals_ring[rs_v] = eps_ring(rs_v)
    print("r_s = %.0e :  eps_ring = %9.5f Ry" % (rs_v, vals_ring[rs_v]))
ks_ = sorted(vals_ring)
slope_ = (vals_ring[ks_[2]] - vals_ring[ks_[0]])/np.log(ks_[2]/ks_[0])
inter_ = vals_ring[ks_[0]] - slope_*np.log(ks_[0])
e2x_v = float(sp.N(sp.log(2)/3 - 3*sp.zeta(3)/(2*pi**2), 20))
print("slope     = %.5f    (Gell-Mann-Brueckner 2(1-ln2)/pi^2 = %.5f)" % (slope_, 2*(1-np.log(2))/np.pi**2))
print("intercept = %.4f     (RPA ring constant -0.142 Ry)" % inter_)
print("intercept + e_2x = %.4f    (Carr-Maradudin constant -0.094 Ry)" % (inter_ + e2x_v))
num("External anchor: GMB slope of the RPA ring energy, in the manuscript's conventions", mp.mpf(float(slope_)), 2*(1 - mp.log(2))/mp.pi**2, tol=2e-3)
num("External anchor: RPA ring constant = -0.142 Ry  (tolerance = half a unit in the last printed digit)", mp.mpf(float(inter_)), mp.mpf('-0.142'), tol=5e-4)
num("External anchor: ring constant + e_2x = -0.094 Ry (Carr-Maradudin)  (tolerance = half a unit in the last printed digit)", mp.mpf(float(inter_ + e2x_v)), mp.mpf('-0.094'), tol=5e-4)
```

## Check 31 — The low-density RPA ring coefficient in closed form

**Purpose.** Re-derive the companion's low-density RPA ring coefficient, $-0.803\,r_s^{-3/4}$ Ry, as an auxiliary normalization check. It is distinct from the screened-exchange coefficient.

**Re-derivation.**  The joint scaling that produces $A_1$ works for the ring sum as well: with $q=q_*x$, $\nu=q^2u$ and the large-$q$ Lindhard
limit, the screening strength is again $c\to g/(1+4u^2)$, $g=x^{-4}$.  The frequency integral is elementary,
$$\int_0^\infty\!\!du\Big[\ln\Big(1+\frac{g}{1+4u^2}\Big)-\frac{g}{1+4u^2}\Big]=\frac\pi2\Big[\sqrt{1+g}-1-\frac g2\Big],$$
from $\int_0^\infty\ln\frac{a^2+t^2}{b^2+t^2}\,dt=\pi(a-b)$ with $t=2u$.  Hence $\varepsilon_{\rm ring}\to\frac{3}{2\pi\alpha_L^2}\frac{q_*^5}{r_s^2}W$ with
$$W=\frac\pi2\int_0^\infty\!\!x^4\Big[\sqrt{1+x^{-4}}-1-\frac{x^{-4}}{2}\Big]dx
=\frac\pi8\int_0^\infty\!\!g^{-9/4}\Big[(1+g)^{1/2}-1-\frac g2\Big]dg=\frac\pi8B\Big(-\frac54,\frac34\Big),$$
the last step by the twice-subtracted Beta integral $\int_0^\infty g^{s-1}[(1+g)^{1/2}-1-g/2]dg=B(s,-\tfrac12-s)$ continued to $s=-\tfrac54$
(the two subtractions are what move the convergence strip to $-2<\Re s<-1$).  With $\Gamma(-\tfrac54)=\tfrac{16}5\Gamma(\tfrac34)$ and
$\Gamma(-\tfrac12)=-2\sqrt\pi$,
$$W=-\frac{\sqrt\pi\,\Gamma(3/4)^2}{5},\qquad
\varepsilon_{\rm ring}\longrightarrow-\frac{3}{2\pi\alpha_L^2}\Big(\frac{4a}{3}\Big)^{5/4}\frac{\sqrt\pi\,\Gamma(3/4)^2}{5}\,r_s^{-3/4}
=-0.8030786\ldots\,r_s^{-3/4}\ \text{Ry},$$
reproducing the quoted $-0.803$ in every digit.  Note that the same Beta-function machinery and the same moving region $q\sim q_*$ produce this
number and $A_1$; agreement therefore tests the machinery as well as the citation.

**Consequence.**  With $e_{\rm kite}\simeq-0.2213782\,r_s^{-1/4}$ the kite-to-ring ratio is $0.27566\,r_s^{1/2}$: the leading asymptotic terms are equal at
$r_s\simeq13$, but $q_*≈1.84$ there is not asymptotically large. This is a leading-term extrapolation, not a prediction of the finite-density crossing. Only eventual asymptotic dominance follows.

**What this does not establish.** Neither the finite-density crossing nor the accuracy of the large-density-parameter expansion at $r_s=13$ is tested here. The ring convention is checked separately by the high-density anchor in Check 30.

```python
for gv in ['0.3', '2', '10']:
    gv = mp.mpf(gv)
    num("Ring  int du [ln(1+g/(1+4u^2)) - g/(1+4u^2)] = (pi/2)[sqrt(1+g)-1-g/2], g=%s" % gv,
        mp.quad(lambda uu: mp.log1p(gv/(1 + 4*uu**2)) - gv/(1 + 4*uu**2), [0, 1, mp.inf]),
        mp.pi/2*(mp.sqrt(1 + gv) - 1 - gv/2), tol=1e-20)
W_cl = -sp.sqrt(pi)*sp.gamma(R(3,4))**2/5
exact("Ring  (pi/8) B(-5/4, 3/4) = -sqrt(pi) Gamma(3/4)^2/5", sp.gammasimp(sp.expand_func(pi/8*sp.beta(-R(5,4), R(3,4)) - W_cl)))
def _w_int(x):                      # large-x series: sqrt(1+g)-1-g/2 = -g^2/8 + g^3/16 - 5g^4/128, g = x^-4
    g_ = x**-4
    return x**4*(mp.pi/2)*(-g_**2/8 + g_**3/16 - 5*g_**4/128) if g_ < mp.mpf('1e-6') \
           else x**4*(mp.pi/2)*(mp.sqrt(1 + g_) - 1 - g_/2)
num("Ring  W = (pi/2) int x^4 [sqrt(1+x^-4) - 1 - x^-4/2] dx = -sqrt(pi) Gamma(3/4)^2/5",
    mp.quad(_w_int, [0, 1, 2, 10, mp.inf]), mp.mpf(sp.N(W_cl, 30)), tol=1e-15)
ring_coef = 3/(2*pi*alphaL**2)*(4*aTF/3)**R(5,4)*W_cl
print("closed form of the low-density ring coefficient:", sp.N(ring_coef, 12), "Ry     (quoted value: -0.803)")
num("Low-density RPA ring coefficient = -0.803 r_s^(-3/4) Ry (quoted value reproduced)",
    mp.mpf(sp.N(ring_coef, 25)), mp.mpf('-0.803'), tol=1e-3)
kite_c = mp.mpf('-0.2213782293')
rc = mp.mpf(sp.N(ring_coef, 25))
print("kite/ring = %.5f r_s^(1/2) ;  leading-term extrapolations meet at r_s = %.1f" % (float(kite_c/rc), float((rc/kite_c)**2)))
check("Leading-term extrapolation only (not a finite-density prediction): equality near r_s = 13", 10 < float((rc/kite_c)**2) < 17)
```

## Check 32 — Independent integral and moment checks

**32.1 — The master formula, end to end.**  Everything downstream of Eq. (D1) rests on the one-dimensional reduction
$\mathcal K[b_\Lambda]=64\pi^3A_0\int K_\Lambda\Phi\,dy$ and on its normalization $64\pi^2A_0=6/\pi^3$.  Check 3 verifies the *right-hand side*
(that $\frac6{\pi^3}D_0=e_{2x}$ with $D_0$ from the $h_0$ route), but it takes the reduction itself on trust: an error in the Frullani step,
in the prolate-spheroidal change of variables, or in the prefactor would produce a wrong $\Phi$, and the numerical value of $D_0$ would then
be wrong in exactly the way that makes Check 3 pass.  The bare-line functional gives an independent handle.  For a frequency-independent line
the frequency integral of the two energy denominators is the static kernel $\mathcal W$ of Eq. (I1), so
$$\mathcal K[v]=4\pi A_0\!\int_0^\infty\!dq\!\int\!d^3p\,d^3k\,\frac{\Delta\Delta}{|\bm p-\bm k|^2}\mathcal W(\delta_{\bm p},\delta_{\bm k})
=64\pi^2A_0\!\int_0^\infty\!X_c(q)\,dq=\frac{6}{\pi^3}\int_0^\infty\!X_c\,dq ,$$
using the definition (I2) of $X_c$ (the factor $q^2$ of the radial measure cancels $v=q^{-2}$).  Since $\mathcal K[v]=e_{2x}$, the manuscript's
claim $\int_0^\infty X_c\,dq=D_0$ (auxiliary examples) is equivalent to the Onsager normalization.  The cell integrates the **six-dimensional Monte-Carlo
$X_c$ of Check 28** over $q$ and compares with $D_0=0.2499019$: the small-$q$ end $[0,0.4]$ uses the verified $c_0q+c_2q^3$, the range
$[0.4,12]$ is the Monte-Carlo grid (28 points, trapezoid), and the tail uses $C_\infty/q^4$.  A wrong $\Phi$ or a wrong prefactor in Eq. (D1)
would show up here as a discrepancy, because this route never touches $\Phi$.

The cell also checks that the two independent samplers (all four sectors versus cross sectors only) agree at $q=2.05$, which they must
because $\mathcal W=0$ in the same-sphere sectors for $q>2$ — a null test of the sector bookkeeping.

**32.2 — Cross-check of $Z(q)$ against the companion manuscript.**  Eq. (H6) obtains $Z(q)=\frac{8\pi^2}{3}q^2\log\frac1q+O(q^2)$ from a
Fourier representation with the ball form factor; the companion writes the same quantity as $Z=8\pi^2-2\mathcal O(q)$ with $\mathcal O$ the cross overlap.
These are different derivations of the same object, so their agreement is a genuine cross-check.  The cell evaluates $8\pi^2-2\mathcal O(q)$ from the
closed form of $C$ (obtained by doing the angular integral analytically and the radial integral of the polynomial $V(r)$ exactly) at
$q=10^{-2},10^{-3},10^{-4}$ and extracts both the logarithmic coefficient and the constant: the data fit
$Z=\frac{8\pi^2}{3}q^2[\log\frac1q+\text{const}]$ with a *stable* constant, which is what fixes the coefficient $8\pi^2/3$ unambiguously.

**32.3 — The same-sphere kinematic expansion.**  The third-order coefficient $A_{3,\rm kin}$ rests on the statement that expanding
$\delta_{\bm p}=q^2/2+qp_z$, $\delta_{\bm k}=q^2/2+qk_z$ in the frequency kernel gives, at second order,
$q^{-6}(p_z^2+p_zk_z+k_z^2)\,\Re A^{-4}$ with $A=iu-\tfrac12$ — the combination that produces the moment $52\pi^2/27$ of Check 8.  Nothing
in the notebook tested that expansion; a wrong combination (say $p_z^2+2p_zk_z+k_z^2$) would change $A_{3,\rm kin}$ and no other check would
notice.  The cell expands $[(A-a)(A-b)]^{-1}$ with $a=p_z/q$, $b=k_z/q$ symbolically and compares order by order with
$A^{-2}+(a+b)A^{-3}+(a^2+ab+b^2)A^{-4}+(a^3+a^2b+ab^2+b^3)A^{-5}$.  The third-order term is included because it is *odd*: it cancels under
$(\bm p,\bm k)\to(-\bm p,-\bm k)$ together with the first-order term, so the first uncancelled correction after the quadratic one is quartic.
That is the reason the kinematic series proceeds in even powers and produces no $q_*^{-3}\log q_*$.

**32.4 — Exponent bookkeeping of Sec. 7.**  The statement that the kite-to-ring ratio grows as $r_s^{1/2}$ follows from $r_s^{-1/4}$ (kite)
against the quoted $r_s^{-3/4}$ (RPA ring, taken from the literature): $-\tfrac14-(-\tfrac34)=\tfrac12$.  Only the arithmetic is checked; the
ring coefficient itself is a cited value.

```python
# --- 32.1 end-to-end: Monte-Carlo integral of X_c reproduces D_0 (hence e_2x)
D0_val = float(sp.N(pi**3/18*sp.log(2) - pi/4*sp.zeta(3), 20))
qA, qB = 0.4, 12.0
grid = np.concatenate([np.arange(qA, 2.0, 0.1), np.arange(2.0, 3.01, 0.1),
                       np.arange(3.2, 6.01, 0.2), np.arange(6.5, qB + 0.01, 0.5)])
vals, errs = [], []
for qv_ in grid:
    m_, s_ = (Xc_small(qv_, 1_500_000, 2) if qv_ <= 2.0 else Xc_large(qv_, 800_000, 2))
    vals.append(m_); errs.append(s_)
vals = np.array(vals); errs = np.array(errs)
I_lo = c0_np*qA**2/2 + c2_np*qA**4/4                     # analytic small-q end (verified in 28.1)
I_mid = np.trapezoid(vals, grid)
I_hi = Cinf_np/(3*qB**3)                                  # C_inf/q^4 tail (verified in 28.1)
dq_ = np.diff(grid); wts = np.zeros_like(grid)
wts[0] = dq_[0]/2; wts[-1] = dq_[-1]/2; wts[1:-1] = (dq_[:-1] + dq_[1:])/2
sig_ = float(np.sqrt(np.sum((wts*errs)**2)))
tot = I_lo + I_mid + I_hi
print("  int_0^0.4 (analytic c0 q + c2 q^3) = %.6f" % I_lo)
print("  int_0.4^12 (Monte Carlo, %d points) = %.6f +- %.6f" % (len(grid), I_mid, sig_))
print("  int_12^oo  (C_inf/q^4 tail)         = %.6f" % I_hi)
print("  total = %.6f +- %.6f     D_0 = %.6f     deviation = %+.2f sigma" % (tot, sig_, D0_val, (tot - D0_val)/sig_))
print("  => (6/pi^3) x total = %.6f   vs   e_2x = %.6f" % (6/np.pi**3*tot, float(sp.N(sp.log(2)/3 - 3*sp.zeta(3)/(2*pi**2), 12))))
check("Eq.(D1)/(D6)  Grid integral of X_c: seed-only 3-sigma consistency; excludes q-grid and tail bias", abs(tot - D0_val) < 3*sig_)
check("auxiliary examples  int X_c dq agrees with the quoted 0.2500 to better than 1%", abs(tot - D0_val)/D0_val < 0.01)
a_s1, e_s1 = Xc_small(2.05, 2_500_000, 6); a_l1, e_l1 = Xc_large(2.05, 1_500_000, 6)
print("  null test at q=2.05:  all-sector sampler = %.6f +- %.6f    cross-only sampler = %.6f +- %.6f" % (a_s1, e_s1, a_l1, e_l1))
check("Eq.(I1)  same-sphere sectors contribute nothing to X_c for q>2 (two samplers agree at q=2.05)",
      abs(a_s1 - a_l1) < 4*np.hypot(e_s1, e_l1))

# --- 32.2 Z(q): Fourier route (Eq.(H6)) vs overlap route of the companion manuscript
def C_closed(q):
    q = np.asarray(q, float)
    Lm = np.log(np.abs(2 - q)); Lp = np.log(q + 2); lq = np.log(q)
    inner = q*(q**4*Lm - 2*q**2*(q**2 - 20)*lq - 20*q**2*Lm + 4*q**2 + 40*q*Lm
               - q*(-q**3 + 20*q + 40)*Lp + 88) + 32*(Lp - Lm)
    return np.pi**2*inner/(30*q)
num("Eq.(H6)/(H7)  overlap O(q) -> 4 pi^2 as q -> 0, i.e. Z(0) = 0  (evaluated at q = 1e-3; the closed form loses digits below that)",
    mp.mpf(float(C_closed(1e-3))), 4*mp.pi**2, tol=1e-5)
consts = []
for qv_ in [1e-2, 1e-3, 1e-4]:
    Zv = 8*np.pi**2 - 2*C_closed(qv_)
    cst = Zv/(8*np.pi**2/3*qv_**2) - np.log(1/qv_)
    consts.append(cst)
    print("  q=%8.0e :  Z = %.6e    Z/[(8pi^2/3) q^2] - log(1/q) = %.4f" % (qv_, Zv, cst))
check("Eq.(H6)  Z(q) = (8 pi^2/3) q^2 [log(1/q) + const] with a stable constant over three decades",
      max(consts) - min(consts) < 0.02)

# --- 32.3 the same-sphere kinematic expansion
uK, qK, pzK, kzK, tK = sp.symbols('u q p_z k_z t', real=True)
AK = sp.I*uK - R(1,2); aK = pzK/qK; bK = kzK/qK
lhsK = (1/((AK - aK)*(AK - bK))).subs({pzK: tK*pzK, kzK: tK*kzK})
rhsK = (AK**-2 + (aK + bK)*AK**-3 + (aK**2 + aK*bK + bK**2)*AK**-4
        + (aK**3 + aK**2*bK + aK*bK**2 + bK**3)*AK**-5).subs({pzK: tK*pzK, kzK: tK*kzK})
for nK in range(4):
    exact("Sec.5.1/(A6)  Taylor coefficient of order %d in (p_z,k_z)/q of [(A-a)(A-b)]^-1" % nK,
          sp.diff(lhsK, tK, nK).subs(tK, 0) - sp.diff(rhsK, tK, nK).subs(tK, 0))
exact("Sec.5.1  Re A^-2 = (1/4-u^2)/(1/4+u^2)^2", sp.re(AK**-2) - (R(1,4) - uK**2)/(R(1,4) + uK**2)**2)
check("Sec.5.1  orders 1 and 3 are odd in (p_z,k_z): they cancel on the inversion-symmetric same-sphere sectors",
      sp.simplify(((aK + bK) + (aK**3 + aK**2*bK + aK*bK**2 + bK**3)).subs({pzK: -pzK, kzK: -kzK})
                  + ((aK + bK) + (aK**3 + aK**2*bK + aK*bK**2 + bK**3))) == 0)

# --- 32.4 exponent arithmetic of Sec. 7
exact("Companion low-density comparison: kite/ring ratio exponent: -1/4 - (-3/4) = 1/2", (-R(1,4) - (-R(3,4))) - R(1,2))
```

<a id="check-33"></a>
## Check 33 — Direct self-energy insertion and cancellation

This Monte Carlo check evaluates
$$\Pi_{\rm SE}^{(1)}(q,i\nu)=2C\int\frac{d^3p}{(2\pi)^3}\Delta_{\bf p}
[\Sigma_x(|\mathbf p+\mathbf q|)-\Sigma_x(p)]\operatorname{Re}(i\nu-\delta_{\bf p})^{-2},\quad C=\pi^2q_{\rm TF}^2,$$
with the exact Fock function and $q_{\rm TF}^2=1$. For $\nu=qz$, the predicted logarithmic slope is $+\alpha(z)/4$. The independently sampled kite kernel supplies the vertex slope through $\Pi_{\rm vtx}^{(1)}=-X/(32\pi^2)$. Their sum is tested at $z=0.5,1,2$ using the same finite-$q$ grid as Check 29.4.

Both measured responses are now in the common units. Their ratios and cancellation accuracy are unchanged by the conversion. These finite-$q$ fits test the leading shell logarithm; they do not determine single-logarithm energy coefficients or replace the diagrammatic derivation of the selected-line energy weight.


```python
def f_x(k):
    k = np.asarray(k, float)
    out = np.where(np.abs(k - 1) < 1e-12, 0.5,
                   0.5 + (1 - k**2)*np.log(np.abs((1 + k)/np.where(np.abs(1 - k) < 1e-300, 1e-300, 1 - k)))/(4*np.maximum(k, 1e-300)))
    return np.where(k < 1e-12, 1.0, out)

def self_energy_response_mc(q, z, n=3_000_000, rep=4):
    # Eq.(A18) with q_tf_sq = 1 and C=pi^2: 2 C/(2pi)^3 int d^3p Delta_p [Sigma_x(|p+q|) - Sigma_x(p)] (delta^2-nu^2)/(delta^2+nu^2)^2
    nu = q*z; Rp = 1 + q; out = []
    for _ in range(rep):
        p = _ball9(n, Rp)
        qv = np.array([0, 0, q])
        pn = np.linalg.norm(p, axis=1); pqn = np.linalg.norm(p + qv, axis=1)
        D = (pn < 1).astype(float) - (pqn < 1)
        d_ = p[:, 2]*q + q*q/2
        dS = -0.5*(f_x(pqn) - f_x(pn))                      # Sigma_x(|p+q|) - Sigma_x(p), q_tf_sq = 1
        ker = (d_**2 - nu**2)/(d_**2 + nu**2)**2
        out.append(np.pi**2*2*(4*np.pi/3*Rp**3)*np.where(D != 0, D*dS*ker, 0.0).mean()/(2*np.pi)**3)
    v = np.array(out); return v.mean(), v.std(ddof=1)/np.sqrt(rep)

Abar = lambda z: 1 - 2*z*np.arctan(1/z) + z**2/(1 + z**2)
t33 = time.time(); sums = []; cPix_list = []; dsig_list = []; sum_abs = []
print("   z     vertex_response coefficient        self_energy_response coefficient      predicted +Abar/(4)    sum / |vertex_response coeff|")
for idx, z_ in enumerate([0.5, 1.0, 2.0]):
    ys = []; es = []
    for q_ in qs9:
        m_, s_ = self_energy_response_mc(q_, z_); ys.append(m_); es.append(s_)
    y3 = np.array(ys); e3 = np.array(es); W3 = 1/e3**2
    M3 = np.vstack([L9, np.ones_like(L9)]).T
    cov3 = np.linalg.inv(M3.T@(M3*W3[:, None])); c3 = cov3@(M3.T@(W3*y3))
    cPix = -(A_shell(z_) + dev_list[idx])/(32*np.pi**2)           # measured X coefficient of 29.4, times -1/(32 pi^2)
    rel_sum = (cPix + c3[0])/abs(cPix)
    sums.append(rel_sum); cPix_list.append(cPix)
    dsig_list.append(c3[0] - Abar(z_)/4)          # deviation of the measured self_energy_response from the claim
    sum_abs.append(cPix + c3[0])                                 # absolute residual of the cancellation
    print("  %.1f    %+.6f            %+.6f +- %.6f        %+.6f              %+.3f"
          % (z_, cPix, c3[0], np.sqrt(cov3[0, 0]), Abar(z_)/4, rel_sum))
print("   [%.0f s]  (q_tf_sq = 1; vertex_response coefficient is the 29.4 measurement of A(z) divided by -32 pi^2;" % (time.time() - t33))
print("             its accuracy, not that of self_energy_response, limits the cancellation test near z = 0.63 where alpha changes sign)")
scale33 = 1/4                 # Abar(0) = 1, so this is the natural size of both coefficients
print("   judged against the natural scale Abar(0)/(4) = %.5f, the residual sums are %s"
      % (scale33, ["%.2f%%" % (100*abs(cc)*abs(cPix_)/scale33) for cc, cPix_ in zip(sums, cPix_list)]))
check("Eq.(A18)  the measured self_energy_response logarithm equals +Abar(z)/(4) to 2% of the scale Abar(0)/(4) at z = 0.5, 1, 2",
      max(abs(np.array(dsig_list))) < 0.02*scale33)
check("Eq.(G20)  the two shell logarithms cancel to 1% of the scale Abar(0)/(4): vertex_response + self_energy_response = 0 within the accuracy of the A(z) measurement",
      max(abs(np.array(sum_abs))) < 0.01*scale33)
```

## Check 34 — Passivity on a grid

For $\lambda=1$, sample $\operatorname{Im}\epsilon(q,\omega+i0)$ on $r_s\in\{1,4,20\}$, $q\in\{0.3,0.8,1.5,3\}$ and positive frequencies. Its nonnegativity implies the nonnegative continuum loss $-\operatorname{Im}\epsilon^{-1}=\operatorname{Im}\epsilon/|\epsilon|^2$.

This cell checks the sign only. It does **not** compute $\operatorname{Re}\epsilon$, integrate the spectral measure, or find plasmon poles on this grid. Check 22 separately tests the complete spectral identity at two parameter choices, one with a plasmon point mass. Analyticity and unsampled parameters remain analytic arguments.


```python
def _fL_re(q, w):                                    # Im part of the retarded Lindhard function (standard form)
    # Positive Im Lindhard shape; Pi_0 = -q_tf_sq * lindhard_shape has negative Im Pi_0.
    nu_m = w/q - q/2; nu_p = w/q + q/2
    out = np.zeros_like(w)
    m1 = np.abs(nu_m) < 1
    out[m1] = (1 - nu_m[m1]**2)
    m2 = np.abs(nu_p) < 1
    out[m2] -= (1 - nu_p[m2]**2)
    return np.pi/(4*q)*out                            # >= 0 for w > 0
def _lindhard_shape_im(q, nu, nk=1200):
    kk = (np.arange(nk) + 0.5)/nk
    return np.mean(kk*np.log(((q*kk + q*q/2)**2 + nu**2)/((q*kk - q*q/2)**2 + nu**2)))/(2*q)

print("  r_s    q     min Im eps     passivity")
ok_pos = True
for rs_h in [1.0, 4.0, 20.0]:
    q_tf_sq_h = float(sp.N(aTF, 20))*rs_h
    for q_h in [0.3, 0.8, 1.5, 3.0]:
        w = np.linspace(1e-4, max(6.0, 3*q_h + 3), 4000)
        chi_im = _fL_re(q_h, w)                        # >= 0
        eps_im = q_tf_sq_h/q_h**2*chi_im                   # Im eps = -v Im chi_0 ... same sign convention as Check 22
        # Re eps from the Kramers-Kronig-free static-plus-continuum form is not needed for positivity of rho:
        # rho = (2 w/pi) Im[-1/eps] = (2 w/pi) Im eps/|eps|^2 >= 0 whenever Im eps >= 0.
        rho_sign_ok = np.all(eps_im >= -1e-14)
        ok_pos = ok_pos and rho_sign_ok
        print("  %4.0f  %4.1f    %.3e     %s" % (rs_h, q_h, eps_im.min(), "ok" if rho_sign_ok else "NEGATIVE"))
check("Eq.(C2)  Im eps(q, w + i0) >= 0 on the whole grid (passivity => varrho_lambda >= 0)", ok_pos)
print("  positivity of rho follows pointwise from Im eps >= 0 through rho = (2w/pi) Im eps/|eps|^2")
```

## Check 35 — Deterministic insertion and the f-sum rule

The deterministic quadrature evaluates the same common-unit response as Check 33,
$$\Pi_{\rm SE}^{(1)}=2C\int\frac{d^3p}{(2\pi)^3}\Delta_{\bf p}
\operatorname{Re}\frac{\Sigma_x(|\mathbf p+\mathbf q|)-\Sigma_x(p)}{(i\nu-\delta_{\bf p})^2},
\quad C=\pi^2q_{\rm TF}^2,\quad\Sigma_x=-q_{\rm TF}^2f_x/2.$$
The exact Fock function and occupation difference are used on the $1200\times1200$ radial-angular quadrature. At $\nu=qz$, fitting $q=0.02,0.04,0.08$ gives the logarithmic slope $q_{\rm TF}^4\alpha(z)/4$, including its sign change. The vertex slope is independently $-q_{\rm TF}^4\alpha(z)/4$.

The relative normalization also obeys the f-sum rule. The large-frequency tails are
$$\Pi_{\rm vtx}^{(1)}\sim\frac{q_{\rm TF}^4 Z(q)}{32\pi^2\nu^2},\qquad
\Pi_{\rm SE}^{(1)}\sim\frac{q_{\rm TF}^4}{8\pi\nu^2}\int d^3p\,\Delta_{\bf p}[f_x(|\mathbf p+\mathbf q|)-f_x(p)].$$
They cancel by the exact geometric identity $\int d^3p\,\Delta_{\bf p}[f_x(|\mathbf p+\mathbf q|)-f_x(p)]=-Z(q)/(4\pi)$. The code checks the shifted-ball representation using adaptive quadrature split at $k=1$, against the independently evaluated overlap formula. This check is unchanged geometrically by the common positive response rescaling.


```python
import numpy as np, time
def _F_fock(k):
    k = np.asarray(k, float); out = np.empty_like(k)
    m0 = k < 1e-8
    m = (np.abs(k - 1) < 1e-13) | m0
    out[~m] = 0.5 + (1 - k[~m]**2)*np.log(np.abs((1 + k[~m])/(1 - k[~m])))/(4*k[~m])
    out[m] = 0.5
    out[m0] = 1 - k[m0]**2/3                         # f_x(0)=1, f_x(1)=1/2
    return out
def self_energy_response(q, z, q_tf_sq_v=1.0, npn=1200, ncn=1200):
    nu_ = q*z
    pl, pw = np.polynomial.legendre.leggauss(npn)
    lo_, hi_ = max(1e-9, 1 - 1.5*q), 1 + 1.5*q       # contains the whole shell
    p_ = (hi_ - lo_)/2*pl + (hi_ + lo_)/2; wp_ = (hi_ - lo_)/2*pw
    c_, wc_ = np.polynomial.legendre.leggauss(ncn)
    P, C = np.meshgrid(p_, c_, indexing='ij'); WP, WC = np.meshgrid(wp_, wc_, indexing='ij')
    pq = np.sqrt(P**2 + 2*P*q*C + q*q)
    D_ = (P < 1).astype(float) - (pq < 1)            # Delta_p, exact
    d_ = P*q*C + q*q/2                               # delta_p, exact
    dS = (-q_tf_sq_v/2)*(_F_fock(pq) - _F_fock(P))       # Sigma_x(p+q) - Sigma_x(p), exact
    ker = ((1j*nu_ - d_)**-2).real
    return (np.pi**2*q_tf_sq_v)*2*(1/(4*np.pi**2))*np.sum(WP*WC*P**2*np.where(D_ != 0, D_*dS*ker, 0.0))

Abar_np = lambda z_: 1 - 2*z_*np.arctan(1/z_) + z_**2/(1 + z_**2)
t33 = time.time(); ok33 = True; qs33 = [0.02, 0.04, 0.08]
print("   z      fitted coefficient of log(1/q)      q_tf_sq^2 Abar(z)/4      ratio")
for z_ in [0.1, 0.3, 1.0, 2.0]:
    ys = [self_energy_response(q_, z_) for q_ in qs33]
    A_ = np.vstack([np.log(1/np.array(qs33)), np.ones(len(qs33))]).T
    coef, *_ = np.linalg.lstsq(A_, np.array(ys), rcond=None)
    pred = Abar_np(z_)/4
    ok33 = ok33 and abs(coef[0]/pred - 1) < 0.02
    print("  %4.1f        %+.6f                     %+.6f             %.3f" % (z_, coef[0], pred, coef[0]/pred))
print("   [%.0f s]" % (time.time() - t33))
check("Eq.(G20)  self_energy_response has logarithmic coefficient +q_tf_sq^2 Abar(z)/4 at z = 0.1, 0.3, 1, 2 (within 2%, including the sign change)", ok33)
exact("Eq.(G20)  vertex_response = -(q_tf_sq^2/32 pi^2) X with X's log coefficient A(z) = 8 pi^2 Abar(z) gives -(q_tf_sq^2/4) Abar(z)",
      -q_tf_sqS**2/(32*pi**2)*8*pi**2*sp.Symbol('alpha') + q_tf_sqS**2/4*sp.Symbol('alpha'))
print("   => the two logarithmic coefficients are equal and opposite, which is the cancellation of Sec. 6.2.")

# --- relative normalization of Eqs. (A17) and (A18): the f-sum rule.
# At large nu both kernels tend to -1/nu^2, so vertex_response -> +q_tf_sq^2 Z(q)/(32 pi^2 nu^2) and
# self_energy_response -> +(q_tf_sq^2/(8 pi nu^2)) int d^3p Delta_p [f_x(|p+q|) - f_x(p)].  Exchange corrections cannot change the
# f-sum rule, so these must cancel:  int d^3p Delta_p [f_x(|p+q|) - f_x(p)] = -Z(q)/(4 pi).
# Proof: f_x(p) = phi(p)/(4 pi) with phi the ball potential of Check 8; shifting p in the theta-functions gives
# 2 int_B f_x(|p+q|) - 2 int_B f_x(p) = 2 C(q)/(4 pi) - 2 pi = -(8 pi^2 - 2 C(q))/(4 pi).
def _fsum_lhs(q):
    # Shift occupations first: 2 int_B [f_x(|p+q|)-f_x(p)] d^3p.
    # Angular substitution k=|p+q| gives dc=k dk/(pq); split at k=1.
    from scipy.integrate import quad
    def inner(p):
        lo, hi = abs(p-q), p+q
        pts = [1.0] if lo < 1 < hi else None
        val = quad(lambda k: k*float(_F_fock(k)), lo, hi,
                   points=pts, epsabs=2e-11, epsrel=2e-11)[0]
        return p*val/q - 2*p*p*float(_F_fock(p))
    pts = [x for x in [q, abs(1-q)] if 0 < x < 1]
    return 4*np.pi*quad(inner, 0, 1, points=sorted(set(pts)), epsabs=2e-10, epsrel=2e-10)[0]
def _Cq33(q):
    Lm = np.log(abs(2 - q)); Lp = np.log(q + 2); lq = np.log(q)
    inner = q*(q**4*Lm - 2*q**2*(q**2 - 20)*lq - 20*q**2*Lm + 4*q**2 + 40*q*Lm - q*(-q**3 + 20*q + 40)*Lp + 88) + 32*(Lp - Lm)
    return np.pi**2*inner/(30*q)
ok_fs = True
for q_ in [0.3, 0.7, 1.5, 3.0]:
    l_ = _fsum_lhs(q_); r_ = -(8*np.pi**2 - 2*_Cq33(q_))/(4*np.pi)
    ok_fs = ok_fs and abs(l_/r_ - 1) < 2e-8
    print("   f-sum identity at q = %.1f :  int Delta_p [f_x(|p+q|) - f_x(p)] = %.5f    -Z(q)/(4 pi) = %.5f    (rel. diff %.1e)" % (q_, l_, r_, l_/r_ - 1))
check("Eqs.(A17)/(A18)  relative normalization: the nu^-2 tails of vertex_response and self_energy_response satisfy the f-sum identity numerically (relative tolerance 2e-8), at four q", ok_fs)
```

## Check 36 — Static appendix: independent quadrature and generic coefficient

The exact four-dimensional fixed-rectangle integral (I15) is evaluated with 16 and 28 Gauss–Legendre nodes per variable. The even continuation is evaluated at both signs of $q$. The cubic coefficient is extracted from $X_c(q)/q$ at $q=0.01,0.02$ and Richardson-extrapolated; no Taylor integrand is used for these quadratures. A separate generic symbolic expansion verifies (I19) for symbolic $u,v$, strengthening Check 16's two specialized evaluations. Finally, the static coupling-average factors $2/3$ and $8/5$ are derived explicitly.


```python
# Check 36: fixed-domain static kernel independent of the pointwise CAS checks.
import numpy as np, sympy as sp
from functools import lru_cache
@lru_cache(None)
def gl36(n):
    x,w=np.polynomial.legendre.leggauss(n)
    return x,w,(x+1)/2,w/2

def static_ratio36(q,n):
    ts,wt,ur,wu=gl36(n)
    U,Rr=np.meshgrid(ur,ur,indexing='ij'); weights=wu[:,None]*wu[None,:]
    total=0.0
    for t,tw in zip(ts,wt):
        Lt=q*t/2+np.sqrt(1-q*q*(1-t*t)/4)
        for s,sw in zip(ts,wt):
            Ls=q*s/2+np.sqrt(1-q*q*(1-s*s)/4)
            den=Lt+Rr*Ls
            H=1+q*U*Lt*Ls*Rr*(t+s)/den+q*q/4*((Lt*t-Rr*Ls*s)**2/den**2-1)
            total+=tw*sw*np.sum(weights*U*Lt**2*Ls**2*Rr/den**2/np.sqrt(H))
    return np.pi/8*total

for q36 in [0.01,0.02,0.2,0.8]:
    a36=static_ratio36(q36,16);b36=static_ratio36(q36,28)
    print('static fixed-domain q=',q36,'Xc/q=',b36,'node diff=',abs(a36-b36))
    num('App.I deterministic node convergence q=%g'%q36,a36,b36,tol=2e-11)
    num('App.I even continuation q=%g'%q36,static_ratio36(-q36,28),b36,tol=2e-13)
c0_36=float(sp.N(sp.pi/8*(2*sp.log(2)-1),17))
c2_36=float(sp.N(sp.pi/1152-sp.pi*sp.log(2)/96,17))
y1=(static_ratio36(.01,28)-c0_36)/.01**2
y2=(static_ratio36(.02,28)-c0_36)/.02**2
c2ext=(4*y1-y2)/3
print('c2 estimates:',y1,y2,'Richardson:',c2ext,'closed:',c2_36)
num('App.I: c2 from direct 4-D quadrature + Richardson',c2ext,c2_36,tol=3e-9)

# Generic symbolic coefficient, without specializing u,v.
u36,v36=sp.symbols('u36 v36',positive=True)
t36,s36=sp.symbols('t36 s36',real=True)
d0=u36+v36;d1=(u36*t36+v36*s36)/2
d2=-(u36*(1-t36*t36)+v36*(1-s36*s36))/8
p1=t36+s36;p2=(t36*t36+s36*s36)/2-sp.Rational(1,2)+t36*s36
h1=u36*v36*(t36+s36)/d0
h2=u36*v36/d0*((t36+s36)/2-d1/d0)*(t36+s36)+((u36*t36-v36*s36)**2/d0**2-1)/4
f2=u36*v36/d0**2*(p2+3*d1**2/d0**2-2*d2/d0-h2/2+3*h1**2/8-2*p1*d1/d0-p1*h1/2+d1*h1/d0)
poly=sp.Poly(sp.expand(f2),t36,s36)
avg=sum(coef*sp.Rational(1,it+1)*sp.Rational(1,js+1) for (it,js),coef in poly.terms() if it%2==0 and js%2==0)
expected=u36**2*v36**2*(3*u36*v36-3*u36-3*v36-5)/(12*(u36+v36)**4)
exact('Restored Eq.(I19): generic u,v Taylor coefficient (no sample-point specialization)',sp.factor(avg-expected))
lam36=sp.symbols('lambda36',positive=True)
exact('Static coupling-average factor for leading high-density logarithm',2*sp.integrate(lam36**2,(lam36,0,1))-sp.Rational(2,3))
exact('Static coupling-average factor for leading low-density coefficient',2*sp.integrate(lam36**sp.Rational(1,4),(lam36,0,1))-sp.Rational(8,5))

a36=sp.symbols('a36',positive=True)
F_s36=2*a36**2-a36+(1-2*a36**3)*sp.log(1+a36)+2*a36**3*sp.log(a36)
exact('Spin endpoint: F_s(a) - log(a) -> -2/3',sp.limit(F_s36-sp.log(a36),a36,sp.oo)+sp.Rational(2,3))
c36,t36=sp.symbols('c36 t36',positive=True)
exact('Eq.(G18) finite shell moment: 2 int_0^c t log(2/t) dt = c^2[log(2/c)+1/2]',
      2*sp.integrate(t36*sp.log(2/t36),(t36,0,c36))-c36**2*(sp.log(2/c36)+sp.Rational(1,2)))
```

## Check 37 — Physical units, static coupling average and the whole-shell Fock term

In Eq. (A10), momenta scale as $k_F$, each Coulomb line as $v_0=4\pi e^2/k_F^2$,
and the energy-frequency measure $d(\hbar\nu)$ scales as $E_0=\hbar^2k_F^2/m$.
The three momentum measures, one energy-frequency measure and two denominators supply $k_F^9/E_0$.
With the physical prefactor $1/(4n)=3\pi^2/(4k_F^3)$ and spin factor 2,
this gives $3me^4/(64\pi^5\hbar^2)$, hence $A_0$ in Ry.
The corresponding angular-frequency unit is $\omega_0=E_0/\hbar$.

The static reference in Sec. 5.4 is at full coupling. For its energy $F_{\rm stat}(q_{\rm TF}^2)$,
the kite prescription gives $2\int_0^1\lambda F_{\rm stat}(\lambda q_{\rm TF}^2)\,d\lambda$. Thus a $q_{\rm TF}^{2}\log q_{\rm TF}^{2}$ coefficient is multiplied by $2/3$,
a $q_{\rm TF}^{4}\log q_{\rm TF}^{2}$ coefficient by $1/2$, and a $q_{\rm TF}^{-3/2}$ coefficient by $8/5$.
Logarithms of $\lambda$ affect lower logarithmic orders or analytic terms. These factors do not change the quoted full-coupling coefficients.

Finally, integrate the finite term $\mathcal H(\rho,c)$ of Eq. (G18) over both signs of the shell.
For $c>0$, $\int_{-c}^0\mathcal H\,d\rho=c^2[\log(2/c)+1/2]$.
Multiplication by the occupation sign gives the same expression with $|c|$ for $c<0$.
This explicitly retains the shell interior and endpoints. It does not evaluate the finite self-energy energy correction.


```python
# Physical normalization from Eq. (A10), without setting hbar or k_F to one.
hb37, mass37, charge37, kf37 = sp.symbols('hbar m e k_F', positive=True)
energy37 = hb37**2*kf37**2/mass37
omega37 = energy37/hb37
pref37 = (3*pi**2/kf37**3)/4*2*kf37**9/(2*pi)**9 \
         *(4*pi*charge37**2/kf37**2)**2/energy37
ha37 = mass37*charge37**4/hb37**2
exact('Physical units: Eq.(A10) prefactor / Hartree = 3/(64 pi^5)',
      pref37/ha37 - 3/(64*pi**5))
exact('Physical units: Eq.(A10) prefactor / Rydberg = A0, independent of k_F',
      pref37/(ha37/2)-A0)
p37,q37 = sp.symbols('p_parallel q_reduced', real=True)
delta37 = hb37**2/(2*mass37)*kf37**2*((p37+q37)**2-p37**2)
exact('Physical units: delta/E_0 = p_parallel q + q^2/2',
      delta37/energy37 - p37*q37-q37**2/2)
l37=sp.symbols('lambda', positive=True)
for pow37,target37 in [(sp.Integer(1),R(2,3)),(sp.Integer(2),R(1,2)),(-R(3,4),R(8,5))]:
    exact('Static coupling average: 2 int lambda^(1+%s) dlambda = %s' % (pow37,target37),
          sp.integrate(2*l37**(1+pow37),(l37,0,1))-target37)
def xlog37(x):
    return mp.mpf(0) if x==0 else x*mp.log(2/abs(x))
for cv37 in [mp.mpf('0.3'),mp.mpf('0.9'),mp.mpf('-0.4')]:
    lo37,hi37=sorted([-cv37,mp.mpf(0)])
    hv37=mp.sign(cv37)*mp.quad(lambda rho:xlog37(rho+cv37)-xlog37(rho),[lo37,hi37])
    num('Eq.(G18) signed whole-shell integral of H, c=%s' % cv37,
        hv37,cv37**2*(mp.log(2/abs(cv37))+mp.mpf('0.5')),tol=1e-25)
```

## Check 38 — Coordinate transformations in the exact reduction

Sec. 4.2 / Eqs. (D1)–(D3). This supplements the sampled end-to-end integrals in Check 29.
Starting from the supplied derivation note, we check the Schwinger sign, Abel-regularized Coulomb transform,
unit-ball transform, the two occupation phases, the five-variable Jacobian, the transformed arguments,
Frullani's logarithm, prolate Jacobian, the remaining axial integral, and the assembled prefactor.
The symbolic equalities hold on the declared nondegenerate domains: $y,q,z>0$, $u>1$, $|v|<1$.
Convergence when the Abel regulator is removed and interchange of the multiple integrals remain analytic obligations;
these are not proved by multiplying the checked factors.

```python

# Symbols here are local to this check's prefix; no prior physical formula is needed.
z38, a38, eta38, k38, r38 = sp.symbols('z38 a38 eta38 k38 r38', positive=True)
exact('38 Schwinger representation: -i int exp[-(z+ia)s] ds = 1/(iz-a)',
      -sp.I/(z38+sp.I*a38)-1/(sp.I*z38-a38))
fourier38=sp.integrate(sp.exp(-eta38*r38)*sp.sin(k38*r38)/k38,(r38,0,sp.oo))
exact('38 Abel Coulomb transform after angular integration = 1/(k^2+eta^2)',fourier38-1/(k38**2+eta38**2))
exact('38 Abel Coulomb limit at nonzero k',sp.limit(fourier38,eta38,0)-1/k38**2)
rho38,p38=sp.symbols('rho38 p38',positive=True)
ball38=4*pi/rho38*sp.integrate(p38*sp.sin(p38*rho38),(p38,0,1))
exact('38 Unit-ball Fourier transform from the radial integral',ball38-4*pi*(sp.sin(rho38)-rho38*sp.cos(rho38))/rho38**3)
exact('38 Unit-ball Fourier transform at zero = volume',sp.limit(ball38,rho38,0)-4*pi/3)
q38, rz38, s138, s238=sp.symbols('q38 rz38 s138 s238',real=True)
phasep=sp.exp(-sp.I*s138*q38/2)*(1-sp.exp(-sp.I*q38*(rz38-s138)))
phasek=sp.exp(-sp.I*s238*q38/2)*(1-sp.exp(sp.I*q38*(rz38+s238)))
def trigzero38(e): return sp.trigsimp(sp.expand_complex(e))
exact('38 Occupation-difference phase for p',trigzero38(phasep-2*sp.I*sp.exp(-sp.I*q38*rz38/2)*sp.sin(q38*(rz38-s138)/2)))
exact('38 Occupation-difference phase for k',trigzero38(phasek+2*sp.I*sp.exp(sp.I*q38*rz38/2)*sp.sin(q38*(rz38+s238)/2)))
exact('38 Two Schwinger signs and sine product give the cosine difference',
      sp.expand_trig(-4*sp.sin(q38*(rz38-s138)/2)*sp.sin(q38*(rz38+s238)/2)
      -2*(sp.cos(q38*(rz38+(s238-s138)/2))-sp.cos(q38*(s138+s238)/2))).expand().trigsimp())
y38=sp.symbols('y38',positive=True)
x38,Rx38,Ry38,Rz38=sp.symbols('x38 Rx38 Ry38 Rz38',real=True)
mapping38=sp.Matrix([y38*(1+x38)/2,y38*(1-x38)/2,y38*Rx38,y38*Ry38,y38*(Rz38+x38/2)])
jac38=mapping38.jacobian([y38,x38,Rx38,Ry38,Rz38]).det()
exact('38 Full five-variable signed Jacobian = -y^4/2 (absolute value y^4/2)',jac38+y38**4/2)
exact('38 First displaced Bessel coordinate',mapping38[4]-mapping38[0]-y38*(Rz38-sp.Rational(1,2)))
exact('38 Second displaced Bessel coordinate',mapping38[4]+mapping38[1]-y38*(Rz38+sp.Rational(1,2)))
exact('38 Cosine argument becomes y R_z',mapping38[4]+(mapping38[1]-mapping38[0])/2-y38*Rz38)
aa38,bb38=sp.symbols('aa38 bb38',positive=True)
fr38=sp.log((eta38**2+bb38**2)/(eta38**2+aa38**2))/2
exact('38 Regulated Frullani derivative from sine Laplace integral',sp.diff(fr38,aa38)+aa38/(eta38**2+aa38**2))
exact('38 Regulated Frullani integration constant at equal arguments',fr38.subs(aa38,bb38))
exact('38 Frullani limit for positive arguments',sp.expand_log(sp.limit(fr38,eta38,0))-sp.log(bb38/aa38).expand(log=True))
for av,bv in [(0.6,1.7),(1.2,0.4)]:
    ev=mp.mpf('0.35'); av,bv=mp.mpf(av),mp.mpf(bv)
    val=mp.quad(lambda t:mp.exp(-ev*t)*(mp.cos(av*t)-mp.cos(bv*t))/t,[0,1,10,50,mp.inf])
    num('38 Regulated Frullani integral a=%s b=%s'%(av,bv),val,mp.log((ev**2+bv**2)/(ev**2+av**2))/2,tol=1e-10)
U38,V38=sp.symbols('U38 V38',real=True)
Rp238=(U38**2-1)*(1-V38**2)/4; Z38=U38*V38/2
det38=sp.Matrix([Rp238,Z38]).jacobian([U38,V38]).det()
exact('38 Prolate Jacobian in (R_perp^2,R_z)',det38-(U38**2-V38**2)/4)
exact('38 Positive-focus squared distance',Rp238+(Z38+sp.Rational(1,2))**2-((U38+V38)/2)**2)
exact('38 Negative-focus squared distance',Rp238+(Z38-sp.Rational(1,2))**2-((U38-V38)/2)**2)
exact('38 Axial antiderivative endpoint ratio = (u+1)/(u-1)',
      (Z38+sp.Rational(1,2)+(U38+V38)/2)/(Z38-sp.Rational(1,2)+(U38-V38)/2)-(U38+1)/(U38-1))
for uv,vv in [(1.3,0.2),(2.4,-0.6)]:
    uv,vv=mp.mpf(uv),mp.mpf(vv)
    val=mp.quad(lambda x:1/mp.sqrt((uv**2-1)*(1-vv**2)/4+(uv*vv/2+x/2)**2),[-1,1])
    num('38 Axial integral = 2 L(u), u=%s v=%s'%(uv,vv),val,2*mp.log((uv+1)/(uv-1)),tol=1e-24)
exact('38 Intermediate measure coefficient = 16 pi^2 y',4*pi/(4*pi)*2*(y38/2)*y38**2*(4*pi)**2/y38**2-16*pi**2*y38)
exact('38 Final coefficient from prolate measure, azimuth, parity and axial integral',
      16*pi**2*sp.Rational(1,2)*2*pi*2*2-64*pi**3)
```

## Check 39 — Residues, coupling average, and low-density substitutions

Eq. (A10): compute the residues for all four occupation pairs; the choice of the time-ordered continuation
and decay of the quarter-circle integrals still require a contour argument.
Eqs. (H15)–(H17): verify the three substitutions before comparing the integrals.
The exact coupling average is derived rather than merely used.
These checks do not derive the many-body contraction or the outer diagram symmetry factor.

```python

Om39,ep39,er39=sp.symbols('Om39 ep39 er39',real=True)
et39=sp.symbols('et39',positive=True)
for np39,nr39 in [(0,0),(1,0),(0,1),(1,1)]:
    zp39=Om39+ep39-sp.I*et39*(1-2*np39)
    zr39=er39-sp.I*et39*(1-2*nr39)
    integral39=sp.I*(np39/(zp39-zr39)+nr39/(zr39-zp39))
    dd39=np39-nr39
    target39=0 if dd39==0 else sp.I*dd39/(Om39-(er39-ep39)+2*sp.I*et39*dd39)
    exact('39 Particle-hole frequency residues: occupations %d,%d'%(np39,nr39),integral39-target39)
exact('39 Two residue factors times Wick differential: i*(i^2)*i = +1',sp.I*(sp.I**2)*sp.I-1)
c39,l39=sp.symbols('c39 l39',positive=True)
R39=2*(c39-sp.log(1+c39))/c39**2
exact('39 Coupling average 2 int lambda/(1+lambda c) = Upsilon(c)',2*sp.integrate(l39/(1+l39*c39),(l39,0,1))-R39)
exact('39 Upsilon(c) series including quadratic correction',sp.series(R39,c39,0,3).removeO()-(1-2*c39/3+c39**2/2))
nu39,delta39,shift39,epsilon39=sp.symbols('nu39 delta39 shift39 epsilon39')
exact('39 Self-energy insertion: derivative of shifted bubble denominator',
      sp.diff(1/(sp.I*nu39-delta39-epsilon39*shift39),epsilon39).subs(epsilon39,0)-shift39/(sp.I*nu39-delta39)**2)
exact('39 Spin-frame scaling: momenta, frequency, two lines and two denominators',
      sp.Integer(9)+2-2-2-4-3)
t39,h39=sp.symbols('t39 h39',positive=True)
x39=sp.symbols('x39',positive=True)
s_lambda39=sp.sqrt(1+l39*x39**-4)/2
integrand39=l39/x39**2*l39/(x39**4+l39)*s_lambda39/(1+2*s_lambda39)**2
s_lambda039=sp.sqrt(1+t39**-4)/2
target39=l39**sp.Rational(3,4)/t39**2/(1+t39**4)*s_lambda039/(1+2*s_lambda039)**2
exact('39 x=lambda^(1/4)t including dx and lambda measure',
      sp.simplify(integrand39.subs(x39,l39**sp.Rational(1,4)*t39)*l39**sp.Rational(1,4)-target39))
# Parameterize s>1 as sqrt(1+h), h>0; then t=h^(-1/4).
s39=sp.sqrt(1+h39)
trans39=(t39**-2/(1+t39**4)*s_lambda039/(1+2*s_lambda039)**2).subs(t39,h39**(-sp.Rational(1,4)))*s39/(2*h39**sp.Rational(5,4))
target_s39=h39**sp.Rational(1,4)/(4*(s39+1)**2)
exact('39 s=sqrt(1+t^-4), reversed limits and absolute Jacobian',trans39-target_s39)
# h^(1/4)/(s+1)^2 = (s-1)^(1/4)(s+1)^(-7/4), since h=(s-1)(s+1).
z39=sp.symbols('z39',positive=True)
sz39=(1+z39)/(1-z39)
exact('39 s=(1+z)/(1-z) Jacobian',sp.diff(sz39,z39)-2/(1-z39)**2)
for zv in [mp.mpf('.15'),mp.mpf('.7')]:
    sv=(1+zv)/(1-zv)
    original=(sv-1)**mp.mpf('.25')*(sv+1)**mp.mpf('-1.75')*2/(1-zv)**2
    num('39 Beta substitution pointwise, z=%s'%zv,original,zv**mp.mpf('.25')*(1-zv)**mp.mpf('-.5')/mp.sqrt(2),tol=1e-25)
```

## Check 40 — Bounds, local shell constant, and generic logarithmic integration

This check distinguishes algebra used in a bound from a proof of a uniform remainder.
For $|a|,|b|\le1$, the modulus of the two resolvents gives
$(1+z^2)|abR|\le1$: the squared-denominator difference below is a sum of nonnegative terms.
The shell constant is $-\int_0^1\int_0^1\log|s-t|\,ds\,dt=3/2$;
identifying this local radial average with the whole finite-part prescription still requires the shell geometry.
For the static fixed rectangle and $|q|\le1/4$, $3/4<L_t,L_s\le9/8$.
Using $AB/(A+B)\le(A+B)/4$ gives $\widetilde H\ge45/64$.
These explicit positive bounds support the hand argument for analytic Taylor integration.
The recurrence for $w^n/(w+q_{\rm TF}^{2})$ explains integer powers and single logarithms at every polynomial order.
Uniform asymptotics of the original multidimensional integrals are not implied by these local checks.

```python

a40,b40,z40=sp.symbols('a40 b40 z40',real=True)
exact('40 Shell-bound squared denominators as nonnegative polynomial for |a|,|b|<=1',
      (a40*a40+z40*z40)*(b40*b40+z40*z40)-a40*a40*b40*b40*(1+z40*z40)**2
      -z40**2*(a40**2*(1-b40**2)+b40**2*(1-a40**2))-z40**4*(1-a40**2*b40**2))
u40,v40=sp.symbols('u40 v40',positive=True)
local40=2*sp.integrate(sp.integrate(sp.log(u40-v40),(v40,0,u40)),(u40,0,1))
exact('40 Local radial coincidence integral = -3/2',local40+sp.Rational(3,2))
exact('40 Positive bound AB/(A+B) <= (A+B)/4 via square',
      (u40+v40)/4-u40*v40/(u40+v40)-(u40-v40)**2/(4*(u40+v40)))
check('40 Fixed-domain H lower bound at |q|<=1/4 is 45/64',1-sp.Rational(1,4)*sp.Rational(9,8)-sp.Rational(1,64)==sp.Rational(45,64))
w40,q_tf_sq40,W40=sp.symbols('w40 q_tf_sq40 W40',positive=True)
for order40 in range(1,7):
    polynomial40=sum((-q_tf_sq40)**j*w40**(order40-1-j) for j in range(order40))
    exact('40 Polynomial division w^n/(w+q_tf_sq), n=%d'%order40,
          w40**order40/(w40+q_tf_sq40)-polynomial40-(-q_tf_sq40)**order40/(w40+q_tf_sq40))
    primitive40=sum((-q_tf_sq40)**j*w40**(order40-j)/(order40-j) for j in range(order40))+(-q_tf_sq40)**order40*sp.log(w40+q_tf_sq40)
    exact('40 Antiderivative for generic logarithmic-integral pattern, n=%d'%order40,
          sp.diff(primitive40,w40)-w40**order40/(w40+q_tf_sq40))
Q40=sp.symbols('Q40',positive=True)
exact('40 High-density low-q bound integral = q_tf_sq(3/4-log(q_tf_sq)/4)',
      sp.integrate(Q40*(1-sp.log(Q40)),(Q40,0,sp.sqrt(q_tf_sq40)))-q_tf_sq40*(sp.Rational(3,4)-sp.log(q_tf_sq40)/4))
for power40 in [-sp.Rational(1,2),-1]:
    exact('40 Low-density coupling integrability exponent %s'%power40,
          2*sp.integrate(l39**(1+power40),(l39,0,1))-2/(2+power40))
```

## Check 41 — Mellin pole orders and inverse transform

We check representative simple and double poles of $\widetilde\Xi(s)$, including their leading Laurent coefficients.
For three positive arguments we also invert $\widetilde\Xi(s)$ on $\Re s=-1/2$ and compare with the independently evaluated Laplace form of $\Xi$.
This tests the inverse-Mellin convention for the kernel; it does not justify displacing the contour of $\widetilde\Xi(s)\widetilde\Phi(-s)$.
The latter still needs the growth and continuation bounds for $F$.

```python

s41,e41=sp.symbols('s41 e41')
XiT41=-pi/2*sp.gamma(s41+1)/sp.sin(pi*s41/2)
for point41,order41,lead41 in [(0,1,-1),(2,1,2),(4,1,-24),(-1,1,pi/2),(-2,2,-1),(-3,1,-pi/4),(-4,2,sp.Rational(1,6))]:
    got41=sp.limit(e41**order41*XiT41.subs(s41,point41+e41),e41,0)
    exact('41 Leading Laurent coefficient at s=%d, pole order %d'%(point41,order41),got41-lead41)
def inverse41(z,T):
    sig=mp.mpf('-.5')
    return mp.quad(lambda t:mp.re((-mp.pi/2*mp.gamma(sig+1+1j*t)/mp.sin(mp.pi*(sig+1j*t)/2))*z**(-sig-1j*t)),[0,2,5,T])/mp.pi
for zv41 in [mp.mpf('.2'),mp.mpf(1),mp.mpf(5)]:
    coarse41=inverse41(zv41,12);fine41=inverse41(zv41,18)
    num('41 Inverse Mellin cutoff convergence z=%s'%zv41,coarse41,fine41,tol=2e-15)
    lap41=zv41*mp.quad(lambda t:mp.exp(-zv41*t)/(1+t*t),[0,1,mp.inf])
    num('41 Inverse Mellin XiT(s) agrees with Laplace Xi(z), z=%s'%zv41,fine41,lap41,tol=1e-20)
```

## Check 42 — Frequency-first shell kernel and the high-density remainder

Appendix G.2 now integrates frequency before the shell limit. We check the signed kernel $M_\Lambda=ab\Lambda^2J$, its derivative identities, the bound on its diagonal, and independent frequency quadrature including grazing parameters. These support the short geometric shell argument in the manuscript; they are not a formal proof of the full multidimensional remainder.

For opposite signs write $H=\alpha\beta/(\alpha+\beta)$ and $U_\alpha=\Lambda/(\Lambda+\alpha)$. Then $M=H(U_\alpha+U_\beta)/2$, $|\partial_\alpha H|\le1$, and $H|\partial_\alpha U_\alpha|/2\le1/8$. This gives the uniform derivative bound $9/8$. Averaging $M_t$ over $0<t<1$ gives the $h$-weighted kernel.

```python

a42,b42,l42,z42=sp.symbols('a42 b42 l42 z42',positive=True)
ms42=a42*b42*l42/(2*(l42+a42)*(l42+b42))
mc42=a42*b42/(2*(a42+b42))*(l42/(l42+a42)+l42/(l42+b42))
exact('42 same-sign frequency kernel from J',ms42-a42*b42*l42**2/(2*l42*(a42+l42)*(b42+l42)))
exact('42 cross-sign frequency kernel from J',mc42-a42*b42*l42*(a42+b42+2*l42)/(2*(a42+b42)*(a42+l42)*(b42+l42)))
exact('42 same-sign derivative',sp.diff(ms42,a42)-b42*l42**2/(2*(a42+l42)**2*(b42+l42)))
exact('42 harmonic factor derivative',sp.diff(a42*b42/(a42+b42),a42)-b42**2/(a42+b42)**2)
exact('42 one-quarter bound is a square',R(1,4)-a42*l42/(a42+l42)**2-(a42-l42)**2/(4*(a42+l42)**2))
exact('42 diagonal bound |c|/8',a42/8-ms42.subs(b42,a42)-a42*(a42-l42)**2/(8*(a42+l42)**2))
def m42(lv,av,bv):
    aa,bb=abs(av),abs(bv)
    if av*bv>0:return aa*bb*lv/(2*(lv+aa)*(lv+bb))
    return aa*bb/(2*(aa+bb))*(lv/(lv+aa)+lv/(lv+bb))
for lv,av,bv in [('0.3','0.4','0.7'),('0.3','0.4','-0.7'),('1e-4','2e-5','-0.6'),('20','0.8','0.8')]:
    lv,av,bv=map(mp.mpf,(lv,av,bv))
    cuts=sorted(set([mp.mpf(0),abs(av),abs(bv),lv,mp.mpf(1)]))+[mp.inf]
    val=mp.quad(lambda zz:av*bv*lv**2/(lv**2+zz**2)*mp.re(1/((1j*zz-av)*(1j*zz-bv))),cuts)
    num(f'42 independent frequency integral {lv},{av},{bv}',val/mp.pi,m42(lv,av,bv),tol=1e-18)
for av,bv in [('0.2','0.6'),('0.2','-0.6'),('0.002','-0.003')]:
    av,bv=map(mp.mpf,(av,bv))
    mix=mp.quad(lambda tt:m42(tt,av,bv),[0,min(abs(av),abs(bv)),1])
    val=mp.quad(lambda zz:av*bv*(1-zz*mp.atan(1/zz))*mp.re(1/((1j*zz-av)*(1j*zz-bv))),[0,abs(av),abs(bv),1,mp.inf])/mp.pi
    num(f'42 h mixture independently integrated {av},{bv}',mix,val,tol=1e-18)
```

## Check 43 — Exact Lindhard shell measure and integrated Fock remainder

Equation (G14) follows by taking the transverse area of the two displaced spheres. Its difference from $h$ is supported only on $[1-q/2,1+q/2]$. We check the exact $L^1$ mass $q/4$, the independent Lindhard integral, and the explicit bound $|\mathcal L-h|\le q/[2(1+z^2)]$ for $0<q<1$.

For the self-energy integral the logarithmic frequency kernel is evaluated before expanding the Fock increment. The sampled Fock remainder bound is supporting evidence; the manuscript uses the endpoint modulus $|u\log|u||$ and the crescent width to establish integrability.

```python

q43,a43,z43=sp.symbols('q43 a43 z43',positive=True)
cap43=(1-(a43-q43/2)**2)/(2*q43)
exact('43 endpoint L1 mass q/4',sp.integrate(a43-cap43,(a43,1-q43/2,1))+sp.integrate(cap43,(a43,1,1+q43/2))-q43/4)
exact('43 Lindhard high-frequency moment 1/3',sp.integrate(a43*a43,(a43,0,1-q43/2))+sp.integrate(a43*cap43,(a43,1-q43/2,1+q43/2))-R(1,3))
def g43(qv,zv):
    return mp.quad(lambda aa:aa*aa/(aa*aa+zv*zv),[0,1-qv/2])+mp.quad(lambda aa:aa*(1-(aa-qv/2)**2)/(2*qv*(aa*aa+zv*zv)),[1-qv/2,1+qv/2])
for qv in map(mp.mpf,['0.02','0.2','0.8']):
    for zv in map(mp.mpf,['0','0.001','0.7','4']):
        gv=g43(qv,zv)
        ref=fL(qv/2) if zv==0 else lindhard_shape_exact(qv,qv*zv)
        num(f'43 longitudinal vs independent Lindhard q={qv},z={zv}',gv,ref,tol=1e-15)
        hv=mp.mpf(1) if zv==0 else 1-zv*mp.atan(1/zv)
        check(f'43 explicit endpoint bound q={qv},z={zv}',abs(gv-hv)<=qv/(2*(1+zv*zv)))
freq43=sp.integrate(l42/(2*(l42+a43)**2),(l42,0,1))
exact('43 integrated double resolvent',freq43-(sp.log(1+1/a43)-1/(1+a43))/2)
def ff43(kv):
    return mp.mpf('0.5') if kv==1 else mp.mpf('0.5')+(1-kv*kv)*mp.log(abs((1+kv)/(1-kv)))/(4*kv)
for qv in map(mp.mpf,['1e-3','1e-6','1e-9']):
    max_ratio=mp.mpf(0)
    for av in map(mp.mpf,['1e-5','0.01','0.7']):
        for uv in map(mp.mpf,['0.1','0.5','0.9']):
            km=mp.sqrt(1-2*qv*av*uv);kp=mp.sqrt(1+2*qv*av*(1-uv))
            # Set q_tf_sq=1; the increment and comparison bound are both linear in q_tf_sq.
            rem=-(ff43(kp)-ff43(km))/2-qv*av*mp.log(1/qv)/4
            max_ratio=max(max_ratio,abs(rem)/(qv*av*(1+abs(mp.log(av)))))
    check(f'43 sampled Fock remainder q={qv}',max_ratio<1,detail='maximum normalized remainder '+mp.nstr(max_ratio,6))
```

## Check 44 — Energy completion and independent finite part

The explicit energy functional of Section 4.3 (Eq. (A21)) reproduces the kite for $T=\Pi_{\rm vtx}^{(1)}$ and assigns the same outer weight to $\Pi_{\rm SE}^{(1)}$. This fixes the selected diagram family; it does not identify it with a full Dyson/RPAx resummation. The finite part imported from the companion is independently obtained by integrating the positive ball-overlap representation.

Added for the 2026-09-17 revision: the algebraic identity behind the combined response (Eq. (A20)), the absence of a $\nu^{-2}$ term, the normalization of $\mathcal K_T$ (Eq. (A22)), and the frequency integral $\int_0^\infty|\Re(\mathrm i\nu-\delta)^{-2}|\,\mathrm d\nu=1/|\delta|$ of Appendix C.3 (Eq. (C17)). The crescent bound of Appendix C.3 is an analytic estimate and is not sampled here.


```python

t44,lam44,b44,r44=sp.symbols('t44 lam44 b44 r44',positive=True)
exact('44 energy prefactor',64*pi**4*(3/(32*pi**5))/(pi**2*t44**2)-6/(pi**3*t44**2))
exact('44 energy functional recovers kite',(-6/(pi**3*t44**2))*(-t44**2/(32*pi**2))-2*(3/(32*pi**5)))
exact('44 one-sided first screening weight',sp.diff(lam44/(1+lam44*b44),b44).subs(b44,0)+lam44**2)
exact('44 bare double-resolvent frequency integral',sp.integrate((a43*a43-z43*z43)/(a43*a43+z43*z43)**2,(z43,0,oo)))
lens44=pi*(4+r44)*(2-r44)**2/12
exact('44 independent overlap finite part',-2*pi**3*sp.integrate(r44*lens44,(r44,0,2))+16*pi**4/15)
num('44 overlap finite part by quadrature',-2*mp.pi**3*mp.quad(lambda rr:rr*mp.pi*(4+rr)*(2-rr)**2/12,[0,1,2]),-16*mp.pi**4/15,tol=1e-25)

# --- 2026-09-17 major revision: Eqs. (A20), (A22), (C17) and the display after Eq. (A18)
nu44,dp44,dk44=sp.symbols('nu44 dp44 dk44',real=True); D44=sp.Symbol('D44',positive=True)
gp44=1/(sp.I*nu44-dp44); gk44=1/(sp.I*nu44-dk44)
exact('Eq.(A20)  g_p^2 + g_k^2 - 2 g_p g_k = (g_p - g_k)^2 (combined response)',sp.expand(gp44**2+gk44**2-2*gp44*gk44-(gp44-gk44)**2))
exact('Eq.(A20)  g_p - g_k = (delta_p - delta_k)/[(i nu - delta_p)(i nu - delta_k)]',sp.simplify(gp44-gk44-(dp44-dk44)/((sp.I*nu44-dp44)*(sp.I*nu44-dk44))))
exact('Eq.(A20)  nu^2 (g_p - g_k)^2 -> 0 as nu -> oo: no nu^-2 term at fixed q',sp.simplify(sp.limit(nu44**2*(gp44-gk44)**2,nu44,sp.oo)))
exact('Eq.(A22)  K_T normalization: -(3/(pi^3 q_tf_sq^2)) * (-q_tf_sq^2/(32 pi^2)) = A0',-3/(pi**3*t44**2)*(-t44**2/(32*pi**2))-3/(32*pi**5))
exact('Eq.(C17) Re (i nu - delta)^-2 = (delta^2 - nu^2)/(delta^2 + nu^2)^2',sp.simplify(sp.re(1/(sp.I*nu44-D44)**2)-(D44**2-nu44**2)/(D44**2+nu44**2)**2))
exact('Eq.(C17) int_0^oo |Re (i nu - delta)^-2| dnu = 1/|delta|',
      sp.integrate((D44**2-nu44**2)/(D44**2+nu44**2)**2,(nu44,0,D44))+sp.integrate((nu44**2-D44**2)/(D44**2+nu44**2)**2,(nu44,D44,oo))-1/D44)
_ent=lambda x: x*mp.log(1/x)+(1-x)*mp.log(1/(1-x))
check('Eq.(C17) a log(s/a) + b log(s/b) <= s log 2: binary entropy on 999 points is <= log 2, with equality at a = b',
      max(_ent(mp.mpf(j)/1000) for j in range(1,1000))<=mp.log(2)+mp.mpf('1e-30') and abs(_ent(mp.mpf(1)/2)-mp.log(2))<mp.mpf('1e-30'))
```

## Numerical conventions and scope

- `exact` establishes only the supplied symbolic identity under declared assumptions. Substituted parameters establish only that specialization.
- `num` requires $|a-b|/\max(1,|b|)$ below the specified tolerance; for small targets this is an absolute-error test. Independent quadrature and shared-formula algebra have different roles, identified in each check.
- `mp.quad` uses interval breakpoints to handle endpoint structure. Oscillatory integrals use `quadosc` or explicit subtractions and tail estimates. Working precision, seeds and tolerances are given in the code and output.
- Small-argument series avoid cancellation in closed forms. They do not certify an asymptotic remainder. The parametric improper integral discussed in Check 6 is also checked by quadrature because of a SymPy result that fails that comparison.
- Contour deformation, occupation-domain geometry, the Coulomb finite part and limiting operations are analytic arguments in the manuscript. The table below links every displayed statement to its computational evidence and scope.


## Analytic proof routes

Appendix G.2 controls the frequency-integrated shell remainder; Section 5.2 turns the two polarization corrections into energies and bounds the Fock remainder. Appendix I gives static-kernel analyticity, and Appendix F provides the endpoint and Mellin arguments. Their supporting calculations are linked in the opening result table.
